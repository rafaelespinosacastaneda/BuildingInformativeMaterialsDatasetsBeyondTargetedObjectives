```python
#!/usr/bin/env python3
from __future__ import annotations

from pathlib import Path
from typing import Tuple

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib.patches import Patch


# ============================================================
# CONFIG
# ============================================================
SEEDS = range(5)
DATASET = "ThermoEl"

TARGET_PROPERTIES = [
    "electric_cond",
    "seebeck_coef",
    "thermal_cond",
    "zT",
]

# ------------------------------------------------------CHANGE THE ROOT TO WHERE YOU HAVE STORED THE "al" folder containing all the data construction results---------------------------------------

ROOT = Path(
    r"C:\Users\rafae\OneDrive\Imágenes\Documentos\PhDToronto\OLD_FILES\PhD Toronto"
    r"\DatasetGeneration\Better_Acqsition\ThermoElectric_NEW"
)

#----------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------
QBC_ROOT = ROOT / "al" / "grow_QBC" / DATASET
QBC_DIV_ROOT = ROOT / "al" / "grow_EA_DeltaDiversity" / DATASET
RANDOM_ROOT = ROOT / "al" / "grow_random" / DATASET / "all"

OUTDIR = ROOT / "AUC_EXPERIMENTAL"
PLOTDIR = ROOT / "Plots" / "Plots_AUC_EXPERIMENTAL"

FIG_DPI = 300


# ============================================================
# EXACT METRIC COLUMN MAPPING
# ============================================================
METRIC_COLS = {
    "electric_cond": "rmse_elec_cond",
    "seebeck_coef": "rmse_seebeck",
    "thermal_cond": "rmse_k_total",
    "zT": "rmse_zt",
}

DISPLAY = {
    "electric_cond": "Electrical Conductivity",
    "seebeck_coef": "Seebeck Coefficient",
    "thermal_cond": "Thermal Conductivity",
    "zT": "zT",
}

COLORS = {
    "electric_cond": "#5E81AC",
    "seebeck_coef": "#A3BE8C",
    "thermal_cond": "#D08770",
    "zT": "#B48EAD",
}

POLICIES = ["QBC", "QBC + Diversity"]

HATCH = {
    "QBC": None,
    "QBC + Diversity": "...",
}


# ============================================================
# PATHS
# ============================================================
def qbc_path(target: str, seed: int) -> Path:
    return QBC_ROOT / target / str(seed) / f"ThermoEl_{target}growing_QBC_rf{seed}.csv"


def qbc_div_path(target: str, seed: int) -> Path:
    return (
        QBC_DIV_ROOT
        / target
        / str(seed)
        / f"ThermoEl_{target}growing_EA_1out_div_rf{seed}.csv"
    )


def random_path(seed: int) -> Path:
    return RANDOM_ROOT / str(seed) / f"ThermoEl_all_random_rf{seed}.csv"


# ============================================================
# HELPERS
# ============================================================
def normalize_columns(df: pd.DataFrame) -> pd.DataFrame:
    out = df.copy()
    out.columns = [str(c).strip() for c in out.columns]
    return out


def to_fraction_01(x: pd.Series) -> np.ndarray:
    arr = pd.to_numeric(x, errors="coerce").to_numpy(dtype=float)

    finite = np.isfinite(arr)
    if finite.any() and np.nanmax(arr[finite]) > 1.5:
        arr = arr / 100.0

    return arr


def auc_trapz(x: np.ndarray, y: np.ndarray) -> float:
    x = np.asarray(x, dtype=float)
    y = np.asarray(y, dtype=float)

    mask = np.isfinite(x) & np.isfinite(y)
    x = x[mask]
    y = y[mask]

    if len(x) < 2:
        return np.nan

    order = np.argsort(x)
    x = x[order]
    y = y[order]

    return float(np.trapz(y, x))


def get_xy_for_auc(df: pd.DataFrame, metric_col: str) -> Tuple[np.ndarray, np.ndarray]:
    if "train_ratio" not in df.columns:
        raise ValueError(f"'train_ratio' not found. Columns: {df.columns.tolist()}")

    if metric_col not in df.columns:
        raise ValueError(f"'{metric_col}' not found. Columns: {df.columns.tolist()}")

    tmp = df[["train_ratio", metric_col]].copy()
    tmp["x"] = to_fraction_01(tmp["train_ratio"])
    tmp["y"] = pd.to_numeric(tmp[metric_col], errors="coerce")
    tmp["_row_order"] = np.arange(len(tmp))

    tmp = tmp[np.isfinite(tmp["x"]) & np.isfinite(tmp["y"])].copy()

    rows = []

    for xval, g in tmp.groupby("x", sort=True):
        if np.isclose(xval, 1.0):
            # At train_ratio = 1, use exactly the last row only.
            yval = float(g.sort_values("_row_order").iloc[-1]["y"])
        else:
            # For duplicated intermediate ratios, average them.
            yval = float(g["y"].mean())

        rows.append((float(xval), yval))

    out = pd.DataFrame(rows, columns=["x", "y"]).sort_values("x")

    return out["x"].to_numpy(), out["y"].to_numpy()


# ============================================================
# BUILD AUC TABLE
# ============================================================
def build_auc() -> pd.DataFrame:
    rows = []

    for target in TARGET_PROPERTIES:
        for seed in SEEDS:
            for policy, path_fn in [
                ("QBC", qbc_path),
                ("QBC + Diversity", qbc_div_path),
            ]:
                path = path_fn(target, seed)

                if not path.exists():
                    print(f"[WARN] Missing file: {path}")
                    continue

                print(f"[INFO] Reading {policy}: {path}")

                df = normalize_columns(pd.read_csv(path))

                for eval_prop in TARGET_PROPERTIES:
                    metric_col = METRIC_COLS[eval_prop]

                    if metric_col not in df.columns:
                        print(f"[WARN] Missing metric column {metric_col} in {path}")
                        continue

                    x, y = get_xy_for_auc(df, metric_col)
                    auc_value = auc_trapz(x, y)

                    rows.append(
                        {
                            "dataset": DATASET,
                            "target": target,
                            "eval": eval_prop,
                            "policy": policy,
                            "seed": seed,
                            "auc": auc_value,
                            "source_file": str(path),
                            "metric_col": metric_col,
                        }
                    )

    # Random baseline
    for seed in SEEDS:
        path = random_path(seed)

        if not path.exists():
            print(f"[WARN] Missing random file: {path}")
            continue

        print(f"[INFO] Reading Random: {path}")

        df = normalize_columns(pd.read_csv(path))

        for eval_prop in TARGET_PROPERTIES:
            metric_col = METRIC_COLS[eval_prop]

            if metric_col not in df.columns:
                print(f"[WARN] Missing metric column {metric_col} in {path}")
                continue

            x, y = get_xy_for_auc(df, metric_col)
            auc_value = auc_trapz(x, y)

            rows.append(
                {
                    "dataset": DATASET,
                    "target": "all",
                    "eval": eval_prop,
                    "policy": "Random",
                    "seed": seed,
                    "auc": auc_value,
                    "source_file": str(path),
                    "metric_col": metric_col,
                }
            )

    long_df = pd.DataFrame(rows)

    if long_df.empty:
        raise RuntimeError("AUC table is empty. Check paths and metric columns.")

    agg = (
        long_df.groupby(["dataset", "target", "eval", "policy"], as_index=False)["auc"]
        .agg(auc_mean="mean", auc_std="std")
    )

    agg["auc_std"] = agg["auc_std"].fillna(0.0)

    OUTDIR.mkdir(parents=True, exist_ok=True)

    long_path = OUTDIR / "experimental_auc_per_seed.csv"
    agg_path = OUTDIR / "experimental_auc_summary.csv"

    long_df.to_csv(long_path, index=False)
    agg.to_csv(agg_path, index=False)

    print(f"[OK] Saved per-seed AUC table to: {long_path}")
    print(f"[OK] Saved AUC summary table to: {agg_path}")

    return agg


# ============================================================
# BUILD IMPROVEMENT TABLE
# ============================================================
def build_improvement(agg: pd.DataFrame) -> pd.DataFrame:
    rows = []

    for eval_prop in TARGET_PROPERTIES:
        rand = agg[
            (agg["eval"] == eval_prop)
            & (agg["policy"] == "Random")
            & (agg["target"] == "all")
        ]

        if rand.empty:
            print(f"[WARN] Missing random baseline for eval={eval_prop}")
            continue

        R = float(rand.iloc[0]["auc_mean"])
        sR = float(rand.iloc[0]["auc_std"])

        if not np.isfinite(R) or abs(R) < 1e-15:
            print(f"[WARN] Invalid random baseline for eval={eval_prop}: {R}")
            continue

        for target in TARGET_PROPERTIES:
            for policy in POLICIES:
                sub = agg[
                    (agg["target"] == target)
                    & (agg["eval"] == eval_prop)
                    & (agg["policy"] == policy)
                ]

                if sub.empty:
                    print(f"[WARN] Missing AUC row for target={target}, eval={eval_prop}, policy={policy}")
                    continue

                P = float(sub.iloc[0]["auc_mean"])
                sP = float(sub.iloc[0]["auc_std"])

                improvement = 100.0 * (1.0 - P / R)

                # Error propagation for f = 100 * (1 - P/R)
                dfdP = -100.0 / R
                dfdR = 100.0 * P / (R**2)
                improvement_std = float(np.sqrt((dfdP * sP) ** 2 + (dfdR * sR) ** 2))

                rows.append(
                    {
                        "dataset": DATASET,
                        "eval": eval_prop,
                        "eval_display": DISPLAY[eval_prop],
                        "target": target,
                        "target_display": DISPLAY[target],
                        "policy": policy,
                        "section": "Targeted" if eval_prop == target else "Untargeted",
                        "improvement_mean": improvement,
                        "improvement_std": improvement_std,
                        "auc_policy": P,
                        "auc_random": R,
                    }
                )

    out = pd.DataFrame(rows)

    if out.empty:
        raise RuntimeError("Improvement table is empty. Check AUC summary output.")

    OUTDIR.mkdir(parents=True, exist_ok=True)

    out_path = OUTDIR / "experimental_improvement_vs_random.csv"
    out.to_csv(out_path, index=False)

    print(f"[OK] Saved improvement table to: {out_path}")

    return out
# ============================================================
# PLOTTING FORMAT MATCHING YOUR REFERENCE CODE
# ============================================================
ERRORBAR_CAPSIZE = 3

TARGET_DISPLAY_ORDER = [
    "zT",
    "Electrical Conductivity",
    "Thermal Conductivity",
    "Seebeck Coefficient",
]

TARGET_KEY_FROM_DISPLAY = {
    "zT": "zT",
    "Electrical Conductivity": "electric_cond",
    "Thermal Conductivity": "thermal_cond",
    "Seebeck Coefficient": "seebeck_coef",
}

TARGET_COLORS_DISPLAY = {
    "zT": "#6C8EBF",
    "Electrical Conductivity": "#7FAF8E",
    "Thermal Conductivity": "#C07A6A",
    "Seebeck Coefficient": "#B39BC8",
}

STYLE_MAP = {
    "QBC": {"hatch": None, "alpha": 1.0},
    "QBC + Diversity": {"hatch": "///", "alpha": 1.0},
}


def format_label(label: str) -> str:
    if label == "Electrical Conductivity":
        return "Electrical\nCond."
    if label == "Thermal Conductivity":
        return "Thermal\nCond."
    if label == "Seebeck Coefficient":
        return "Seebeck\nCoeff."
    return label


def compute_global_improvement_limits(improv_df: pd.DataFrame) -> Tuple[float, float]:
    vals = improv_df["improvement_mean"].to_numpy(dtype=float)
    errs = improv_df["improvement_std"].to_numpy(dtype=float)

    ymin = np.nanmin(vals - errs)
    ymax = np.nanmax(vals + errs)

    pad = 0.08 * max(ymax - ymin, 1e-12)

    ymin -= pad
    ymax += pad

    ymin = min(ymin, 0.0)
    ymax = max(ymax, 0.0)

    return ymin, ymax


def make_group_positions():
    group_spacing = 2.25

    cases = []
    xticks = []
    xticklabels = []
    separators = []

    for i, prop in enumerate(TARGET_DISPLAY_ORDER):
        x = i * group_spacing

        cases.append((prop, x))
        xticks.append(x)
        xticklabels.append(prop)

        if i < len(TARGET_DISPLAY_ORDER) - 1:
            separators.append(x + group_spacing / 2.0)

    return cases, xticks, xticklabels, separators


def plot_targeted_panel_single_row(ax, improv_df: pd.DataFrame) -> None:
    sub = improv_df[improv_df["section"] == "Targeted"].copy()

    cases, xticks, xticklabels, separators = make_group_positions()

    width = 0.48
    offsets = {
        "QBC": -0.28,
        "QBC + Diversity": +0.28,
    }

    for policy in ["QBC", "QBC + Diversity"]:
        for eval_display, x in cases:
            eval_key = TARGET_KEY_FROM_DISPLAY[eval_display]

            r = sub[
                (sub["eval"] == eval_key)
                & (sub["target"] == eval_key)
                & (sub["policy"] == policy)
            ]

            if r.empty:
                print(f"[WARN] Missing targeted row: eval={eval_display}, policy={policy}")
                continue

            rr = r.iloc[0]

            ax.bar(
                x + offsets[policy],
                float(rr["improvement_mean"]),
                width=width,
                color=TARGET_COLORS_DISPLAY[eval_display],
                edgecolor="black",
                linewidth=0.8,
                hatch=STYLE_MAP[policy]["hatch"],
                alpha=STYLE_MAP[policy]["alpha"],
                yerr=float(rr["improvement_std"]),
                capsize=ERRORBAR_CAPSIZE,
                error_kw={"elinewidth": 1.0, "capthick": 1.0},
                zorder=3,
            )

    ax.axhline(0.0, color="black", linewidth=1.0, zorder=2)

    ax.set_xticks(xticks)
    ax.set_xticklabels(
        [format_label(l) for l in xticklabels],
        fontsize=17,
        rotation=0,
        ha="center",
        fontweight="bold",
    )

    ax.set_xlim(min(xticks) - 1.0, max(xticks) + 1.0)
    ax.grid(axis="y", linestyle="--", alpha=0.28, zorder=0)

    for x_sep in separators:
        ax.axvline(x_sep, color="black", lw=1.0, alpha=0.30)


def plot_untargeted_panel_single_row(ax, improv_df: pd.DataFrame) -> None:
    sub = improv_df[improv_df["section"] == "Untargeted"].copy()

    cases, xticks, xticklabels, separators = make_group_positions()

    width = 0.18
    offsets = [-0.75, -0.45, -0.15, 0.15, 0.45, 0.75]

    for eval_display, x in cases:
        eval_key = TARGET_KEY_FROM_DISPLAY[eval_display]

        others_display = [t for t in TARGET_DISPLAY_ORDER if t != eval_display]

        barspecs = []
        k = 0

        for target_display in others_display:
            target_key = TARGET_KEY_FROM_DISPLAY[target_display]

            for policy in ["QBC", "QBC + Diversity"]:
                barspecs.append((target_display, target_key, policy, offsets[k]))
                k += 1

        for target_display, target_key, policy, off in barspecs:
            r = sub[
                (sub["eval"] == eval_key)
                & (sub["target"] == target_key)
                & (sub["policy"] == policy)
            ]

            if r.empty:
                print(
                    f"[WARN] Missing untargeted row: "
                    f"eval={eval_display}, target={target_display}, policy={policy}"
                )
                continue

            rr = r.iloc[0]

            ax.bar(
                x + off,
                float(rr["improvement_mean"]),
                width=width,
                color=TARGET_COLORS_DISPLAY[target_display],
                edgecolor="black",
                linewidth=0.8,
                hatch=STYLE_MAP[policy]["hatch"],
                alpha=STYLE_MAP[policy]["alpha"],
                yerr=float(rr["improvement_std"]),
                capsize=ERRORBAR_CAPSIZE,
                error_kw={"elinewidth": 1.0, "capthick": 1.0},
                zorder=3,
            )

    ax.axhline(0.0, color="black", linewidth=1.0, zorder=2)

    ax.set_xticks(xticks)
    ax.set_xticklabels(
        [format_label(l) for l in xticklabels],
        fontsize=17,
        rotation=0,
        ha="center",
        fontweight="bold",
    )

    ax.set_xlim(min(xticks) - 1.2, max(xticks) + 1.2)
    ax.grid(axis="y", linestyle="--", alpha=0.28, zorder=0)

    for x_sep in separators:
        ax.axvline(x_sep, color="black", lw=1.0, alpha=0.30)


def plot_improvement(improv_df: pd.DataFrame) -> None:
    ylims = compute_global_improvement_limits(improv_df)

    fig, axes = plt.subplots(
        nrows=1,
        ncols=2,
        figsize=(15.8, 5.8),
        squeeze=False,
        sharey=True,
        gridspec_kw={"width_ratios": [1, 1.7]},
    )

    axL = axes[0, 0]
    axR = axes[0, 1]

    plot_targeted_panel_single_row(axL, improv_df)
    plot_untargeted_panel_single_row(axR, improv_df)

    axL.set_ylim(*ylims)
    axR.set_ylim(*ylims)

    axL.set_title("Targeted", fontsize=18, fontweight="bold")
    axR.set_title("Untargeted", fontsize=18, fontweight="bold")

    axL.set_ylabel("Improvement vs random (%)", fontsize=18, fontweight="bold")

    policy_handles = [
        Patch(facecolor="white", edgecolor="black", hatch=None, label="QBC"),
        Patch(facecolor="white", edgecolor="black", hatch="///", label="QBC + diversity"),
    ]

    target_handles = [
        Patch(facecolor=TARGET_COLORS_DISPLAY["zT"], edgecolor="black", label="Target: zT"),
        Patch(facecolor=TARGET_COLORS_DISPLAY["Electrical Conductivity"], edgecolor="black", label="Target: Electrical Cond."),
        Patch(facecolor=TARGET_COLORS_DISPLAY["Thermal Conductivity"], edgecolor="black", label="Target: Thermal Cond."),
        Patch(facecolor=TARGET_COLORS_DISPLAY["Seebeck Coefficient"], edgecolor="black", label="Target: Seebeck Coeff."),
    ]

    fig.legend(
        handles=policy_handles + target_handles,
        loc="lower center",
        bbox_to_anchor=(0.5, 0.01),
        ncol=3,
        frameon=True,
        fontsize=16,
    )

    fig.subplots_adjust(
        left=0.08,
        right=0.995,
        top=0.88,
        bottom=0.28,
        wspace=0.10,
    )

    PLOTDIR.mkdir(parents=True, exist_ok=True)

    outpath = PLOTDIR / "TARGETED_UNTARGETED_IMPROVEMENT_THERMOELECTRIC.png"
    fig.savefig(outpath, dpi=FIG_DPI, bbox_inches="tight")
    plt.show()

    print(f"[OK] Wrote plot to: {outpath}")

def main() -> None:
    OUTDIR.mkdir(parents=True, exist_ok=True)
    PLOTDIR.mkdir(parents=True, exist_ok=True)

    agg = build_auc()
    improvement = build_improvement(agg)

    plot_improvement(improvement)


if __name__ == "__main__":
    main()
```

```python
#!/usr/bin/env python3
from __future__ import annotations

from pathlib import Path
from typing import Tuple

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib.patches import Patch


# ============================================================
# CONFIG
# ============================================================
SEEDS = range(5)
DATASET = "ThermoEl"

TARGET_PROPERTIES = [
    "electric_cond",
    "seebeck_coef",
    "thermal_cond",
    "zT",
]

ROOT = Path(
    r"C:\Users\rafae\OneDrive\Imágenes\Documentos\PhDToronto\OLD_FILES\PhD Toronto"
    r"\DatasetGeneration\Better_Acqsition\ThermoElectric_NEW"
)

QBC_ROOT = ROOT / "al" / "grow_QBC" / DATASET
QBC_DIV_ROOT = ROOT / "al" / "grow_EA_DeltaDiversity" / DATASET
RANDOM_ROOT = ROOT / "al" / "grow_random" / DATASET / "all"

OUTDIR = ROOT / "AUC_EXPERIMENTAL"
PLOTDIR = ROOT / "Plots" / "Plots_AUC_EXPERIMENTAL"

FIG_DPI = 300


# ============================================================
# EXACT METRIC COLUMN MAPPING
# ============================================================
METRIC_COLS = {
    "electric_cond": "rmse_elec_cond",
    "seebeck_coef": "rmse_seebeck",
    "thermal_cond": "rmse_k_total",
    "zT": "rmse_zt",
}

DISPLAY = {
    "electric_cond": "Electrical Conductivity",
    "seebeck_coef": "Seebeck Coefficient",
    "thermal_cond": "Thermal Conductivity",
    "zT": "zT",
}

COLORS = {
    "electric_cond": "#5E81AC",
    "seebeck_coef": "#A3BE8C",
    "thermal_cond": "#D08770",
    "zT": "#B48EAD",
}

POLICIES = ["QBC", "QBC + Diversity"]

HATCH = {
    "QBC": None,
    "QBC + Diversity": "...",
}



MODEL = "xgb"   # <-- change to "rf" or "xgb"


# ============================================================
# PATHS
# ============================================================
def qbc_path(target: str, seed: int) -> Path:
    return QBC_ROOT / target / str(seed) / f"ThermoEl_{target}growing_QBC_{MODEL}{seed}.csv"


def qbc_div_path(target: str, seed: int) -> Path:
    return (
        QBC_DIV_ROOT
        / target
        / str(seed)
        / f"ThermoEl_{target}growing_EA_1out_div_{MODEL}{seed}.csv"
    )


def random_path(seed: int) -> Path:
    return RANDOM_ROOT / str(seed) / f"ThermoEl_all_random_{MODEL}{seed}.csv"


# ============================================================
# HELPERS
# ============================================================
def normalize_columns(df: pd.DataFrame) -> pd.DataFrame:
    out = df.copy()
    out.columns = [str(c).strip() for c in out.columns]
    return out


def to_fraction_01(x: pd.Series) -> np.ndarray:
    arr = pd.to_numeric(x, errors="coerce").to_numpy(dtype=float)

    finite = np.isfinite(arr)
    if finite.any() and np.nanmax(arr[finite]) > 1.5:
        arr = arr / 100.0

    return arr


def auc_trapz(x: np.ndarray, y: np.ndarray) -> float:
    x = np.asarray(x, dtype=float)
    y = np.asarray(y, dtype=float)

    mask = np.isfinite(x) & np.isfinite(y)
    x = x[mask]
    y = y[mask]

    if len(x) < 2:
        return np.nan

    order = np.argsort(x)
    x = x[order]
    y = y[order]

    return float(np.trapz(y, x))


def get_xy_for_auc(df: pd.DataFrame, metric_col: str) -> Tuple[np.ndarray, np.ndarray]:
    if "train_ratio" not in df.columns:
        raise ValueError(f"'train_ratio' not found. Columns: {df.columns.tolist()}")

    if metric_col not in df.columns:
        raise ValueError(f"'{metric_col}' not found. Columns: {df.columns.tolist()}")

    tmp = df[["train_ratio", metric_col]].copy()
    tmp["x"] = to_fraction_01(tmp["train_ratio"])
    tmp["y"] = pd.to_numeric(tmp[metric_col], errors="coerce")
    tmp["_row_order"] = np.arange(len(tmp))

    tmp = tmp[np.isfinite(tmp["x"]) & np.isfinite(tmp["y"])].copy()

    rows = []

    for xval, g in tmp.groupby("x", sort=True):
        if np.isclose(xval, 1.0):
            # At train_ratio = 1, use exactly the last row only.
            yval = float(g.sort_values("_row_order").iloc[-1]["y"])
        else:
            # For duplicated intermediate ratios, average them.
            yval = float(g["y"].mean())

        rows.append((float(xval), yval))

    out = pd.DataFrame(rows, columns=["x", "y"]).sort_values("x")

    return out["x"].to_numpy(), out["y"].to_numpy()


# ============================================================
# BUILD AUC TABLE
# ============================================================
def build_auc() -> pd.DataFrame:
    rows = []

    for target in TARGET_PROPERTIES:
        for seed in SEEDS:
            for policy, path_fn in [
                ("QBC", qbc_path),
                ("QBC + Diversity", qbc_div_path),
            ]:
                path = path_fn(target, seed)

                if not path.exists():
                    print(f"[WARN] Missing file: {path}")
                    continue

                print(f"[INFO] Reading {policy}: {path}")

                df = normalize_columns(pd.read_csv(path))

                for eval_prop in TARGET_PROPERTIES:
                    metric_col = METRIC_COLS[eval_prop]

                    if metric_col not in df.columns:
                        print(f"[WARN] Missing metric column {metric_col} in {path}")
                        continue

                    x, y = get_xy_for_auc(df, metric_col)
                    auc_value = auc_trapz(x, y)

                    rows.append(
                        {
                            "dataset": DATASET,
                            "target": target,
                            "eval": eval_prop,
                            "policy": policy,
                            "seed": seed,
                            "auc": auc_value,
                            "source_file": str(path),
                            "metric_col": metric_col,
                        }
                    )

    # Random baseline
    for seed in SEEDS:
        path = random_path(seed)

        if not path.exists():
            print(f"[WARN] Missing random file: {path}")
            continue

        print(f"[INFO] Reading Random: {path}")

        df = normalize_columns(pd.read_csv(path))

        for eval_prop in TARGET_PROPERTIES:
            metric_col = METRIC_COLS[eval_prop]

            if metric_col not in df.columns:
                print(f"[WARN] Missing metric column {metric_col} in {path}")
                continue

            x, y = get_xy_for_auc(df, metric_col)
            auc_value = auc_trapz(x, y)

            rows.append(
                {
                    "dataset": DATASET,
                    "target": "all",
                    "eval": eval_prop,
                    "policy": "Random",
                    "seed": seed,
                    "auc": auc_value,
                    "source_file": str(path),
                    "metric_col": metric_col,
                }
            )

    long_df = pd.DataFrame(rows)

    if long_df.empty:
        raise RuntimeError("AUC table is empty. Check paths and metric columns.")

    agg = (
        long_df.groupby(["dataset", "target", "eval", "policy"], as_index=False)["auc"]
        .agg(auc_mean="mean", auc_std="std")
    )

    agg["auc_std"] = agg["auc_std"].fillna(0.0)

    OUTDIR.mkdir(parents=True, exist_ok=True)

    long_path = OUTDIR / "experimental_auc_per_seed.csv"
    agg_path = OUTDIR / "experimental_auc_summary.csv"

    long_df.to_csv(long_path, index=False)
    agg.to_csv(agg_path, index=False)

    print(f"[OK] Saved per-seed AUC table to: {long_path}")
    print(f"[OK] Saved AUC summary table to: {agg_path}")

    return agg


# ============================================================
# BUILD IMPROVEMENT TABLE
# ============================================================
def build_improvement(agg: pd.DataFrame) -> pd.DataFrame:
    rows = []

    for eval_prop in TARGET_PROPERTIES:
        rand = agg[
            (agg["eval"] == eval_prop)
            & (agg["policy"] == "Random")
            & (agg["target"] == "all")
        ]

        if rand.empty:
            print(f"[WARN] Missing random baseline for eval={eval_prop}")
            continue

        R = float(rand.iloc[0]["auc_mean"])
        sR = float(rand.iloc[0]["auc_std"])

        if not np.isfinite(R) or abs(R) < 1e-15:
            print(f"[WARN] Invalid random baseline for eval={eval_prop}: {R}")
            continue

        for target in TARGET_PROPERTIES:
            for policy in POLICIES:
                sub = agg[
                    (agg["target"] == target)
                    & (agg["eval"] == eval_prop)
                    & (agg["policy"] == policy)
                ]

                if sub.empty:
                    print(f"[WARN] Missing AUC row for target={target}, eval={eval_prop}, policy={policy}")
                    continue

                P = float(sub.iloc[0]["auc_mean"])
                sP = float(sub.iloc[0]["auc_std"])

                improvement = 100.0 * (1.0 - P / R)

                # Error propagation for f = 100 * (1 - P/R)
                dfdP = -100.0 / R
                dfdR = 100.0 * P / (R**2)
                improvement_std = float(np.sqrt((dfdP * sP) ** 2 + (dfdR * sR) ** 2))

                rows.append(
                    {
                        "dataset": DATASET,
                        "eval": eval_prop,
                        "eval_display": DISPLAY[eval_prop],
                        "target": target,
                        "target_display": DISPLAY[target],
                        "policy": policy,
                        "section": "Targeted" if eval_prop == target else "Untargeted",
                        "improvement_mean": improvement,
                        "improvement_std": improvement_std,
                        "auc_policy": P,
                        "auc_random": R,
                    }
                )

    out = pd.DataFrame(rows)

    if out.empty:
        raise RuntimeError("Improvement table is empty. Check AUC summary output.")

    OUTDIR.mkdir(parents=True, exist_ok=True)

    out_path = OUTDIR / "experimental_improvement_vs_random.csv"
    out.to_csv(out_path, index=False)

    print(f"[OK] Saved improvement table to: {out_path}")

    return out
# ============================================================
# PLOTTING FORMAT MATCHING YOUR REFERENCE CODE
# ============================================================
ERRORBAR_CAPSIZE = 3

TARGET_DISPLAY_ORDER = [
    "zT",
    "Electrical Conductivity",
    "Thermal Conductivity",
    "Seebeck Coefficient",
]

TARGET_KEY_FROM_DISPLAY = {
    "zT": "zT",
    "Electrical Conductivity": "electric_cond",
    "Thermal Conductivity": "thermal_cond",
    "Seebeck Coefficient": "seebeck_coef",
}

TARGET_COLORS_DISPLAY = {
    "zT": "#6C8EBF",
    "Electrical Conductivity": "#7FAF8E",
    "Thermal Conductivity": "#C07A6A",
    "Seebeck Coefficient": "#B39BC8",
}

STYLE_MAP = {
    "QBC": {"hatch": None, "alpha": 1.0},
    "QBC + Diversity": {"hatch": "///", "alpha": 1.0},
}


def format_label(label: str) -> str:
    if label == "Electrical Conductivity":
        return "Electrical\nCond."
    if label == "Thermal Conductivity":
        return "Thermal\nCond."
    if label == "Seebeck Coefficient":
        return "Seebeck\nCoeff."
    return label


def compute_global_improvement_limits(improv_df: pd.DataFrame) -> Tuple[float, float]:
    vals = improv_df["improvement_mean"].to_numpy(dtype=float)
    errs = improv_df["improvement_std"].to_numpy(dtype=float)

    ymin = np.nanmin(vals - errs)
    ymax = np.nanmax(vals + errs)

    pad = 0.08 * max(ymax - ymin, 1e-12)

    ymin -= pad
    ymax += pad

    ymin = min(ymin, 0.0)
    ymax = max(ymax, 0.0)

    return ymin, ymax


def make_group_positions():
    group_spacing = 2.25

    cases = []
    xticks = []
    xticklabels = []
    separators = []

    for i, prop in enumerate(TARGET_DISPLAY_ORDER):
        x = i * group_spacing

        cases.append((prop, x))
        xticks.append(x)
        xticklabels.append(prop)

        if i < len(TARGET_DISPLAY_ORDER) - 1:
            separators.append(x + group_spacing / 2.0)

    return cases, xticks, xticklabels, separators


def plot_targeted_panel_single_row(ax, improv_df: pd.DataFrame) -> None:
    sub = improv_df[improv_df["section"] == "Targeted"].copy()

    cases, xticks, xticklabels, separators = make_group_positions()

    width = 0.48
    offsets = {
        "QBC": -0.28,
        "QBC + Diversity": +0.28,
    }

    for policy in ["QBC", "QBC + Diversity"]:
        for eval_display, x in cases:
            eval_key = TARGET_KEY_FROM_DISPLAY[eval_display]

            r = sub[
                (sub["eval"] == eval_key)
                & (sub["target"] == eval_key)
                & (sub["policy"] == policy)
            ]

            if r.empty:
                print(f"[WARN] Missing targeted row: eval={eval_display}, policy={policy}")
                continue

            rr = r.iloc[0]

            ax.bar(
                x + offsets[policy],
                float(rr["improvement_mean"]),
                width=width,
                color=TARGET_COLORS_DISPLAY[eval_display],
                edgecolor="black",
                linewidth=0.8,
                hatch=STYLE_MAP[policy]["hatch"],
                alpha=STYLE_MAP[policy]["alpha"],
                yerr=float(rr["improvement_std"]),
                capsize=ERRORBAR_CAPSIZE,
                error_kw={"elinewidth": 1.0, "capthick": 1.0},
                zorder=3,
            )

    ax.axhline(0.0, color="black", linewidth=1.0, zorder=2)

    ax.set_xticks(xticks)
    ax.set_xticklabels(
        [format_label(l) for l in xticklabels],
        fontsize=17,
        rotation=0,
        ha="center",
        fontweight="bold",
    )

    ax.set_xlim(min(xticks) - 1.0, max(xticks) + 1.0)
    ax.grid(axis="y", linestyle="--", alpha=0.28, zorder=0)

    for x_sep in separators:
        ax.axvline(x_sep, color="black", lw=1.0, alpha=0.30)


def plot_untargeted_panel_single_row(ax, improv_df: pd.DataFrame) -> None:
    sub = improv_df[improv_df["section"] == "Untargeted"].copy()

    cases, xticks, xticklabels, separators = make_group_positions()

    width = 0.18
    offsets = [-0.75, -0.45, -0.15, 0.15, 0.45, 0.75]

    for eval_display, x in cases:
        eval_key = TARGET_KEY_FROM_DISPLAY[eval_display]

        others_display = [t for t in TARGET_DISPLAY_ORDER if t != eval_display]

        barspecs = []
        k = 0

        for target_display in others_display:
            target_key = TARGET_KEY_FROM_DISPLAY[target_display]

            for policy in ["QBC", "QBC + Diversity"]:
                barspecs.append((target_display, target_key, policy, offsets[k]))
                k += 1

        for target_display, target_key, policy, off in barspecs:
            r = sub[
                (sub["eval"] == eval_key)
                & (sub["target"] == target_key)
                & (sub["policy"] == policy)
            ]

            if r.empty:
                print(
                    f"[WARN] Missing untargeted row: "
                    f"eval={eval_display}, target={target_display}, policy={policy}"
                )
                continue

            rr = r.iloc[0]

            ax.bar(
                x + off,
                float(rr["improvement_mean"]),
                width=width,
                color=TARGET_COLORS_DISPLAY[target_display],
                edgecolor="black",
                linewidth=0.8,
                hatch=STYLE_MAP[policy]["hatch"],
                alpha=STYLE_MAP[policy]["alpha"],
                yerr=float(rr["improvement_std"]),
                capsize=ERRORBAR_CAPSIZE,
                error_kw={"elinewidth": 1.0, "capthick": 1.0},
                zorder=3,
            )

    ax.axhline(0.0, color="black", linewidth=1.0, zorder=2)

    ax.set_xticks(xticks)
    ax.set_xticklabels(
        [format_label(l) for l in xticklabels],
        fontsize=17,
        rotation=0,
        ha="center",
        fontweight="bold",
    )

    ax.set_xlim(min(xticks) - 1.2, max(xticks) + 1.2)
    ax.grid(axis="y", linestyle="--", alpha=0.28, zorder=0)

    for x_sep in separators:
        ax.axvline(x_sep, color="black", lw=1.0, alpha=0.30)

"""
def plot_improvement(improv_df: pd.DataFrame) -> None:
    ylims = compute_global_improvement_limits(improv_df)

    fig, axes = plt.subplots(
        nrows=1,
        ncols=2,
        figsize=(15.8, 5.8),
        squeeze=False,
        sharey=True,
        gridspec_kw={"width_ratios": [1, 1.7]},
    )

    axL = axes[0, 0]
    axR = axes[0, 1]

    plot_targeted_panel_single_row(axL, improv_df)
    plot_untargeted_panel_single_row(axR, improv_df)

    axL.set_ylim(*ylims)
    axR.set_ylim(*ylims)

    axL.set_title("Targeted", fontsize=18, fontweight="bold")
    axR.set_title("Untargeted", fontsize=18, fontweight="bold")

    axL.set_ylabel("Improvement vs random (%)", fontsize=18, fontweight="bold")

    policy_handles = [
        Patch(facecolor="white", edgecolor="black", hatch=None, label="QBC"),
        Patch(facecolor="white", edgecolor="black", hatch="///", label="QBC + diversity"),
    ]

    target_handles = [
        Patch(facecolor=TARGET_COLORS_DISPLAY["zT"], edgecolor="black", label="Target: zT"),
        Patch(facecolor=TARGET_COLORS_DISPLAY["Electrical Conductivity"], edgecolor="black", label="Target: Electrical Cond."),
        Patch(facecolor=TARGET_COLORS_DISPLAY["Thermal Conductivity"], edgecolor="black", label="Target: Thermal Cond."),
        Patch(facecolor=TARGET_COLORS_DISPLAY["Seebeck Coefficient"], edgecolor="black", label="Target: Seebeck Coeff."),
    ]

    fig.legend(
        handles=policy_handles + target_handles,
        loc="lower center",
        bbox_to_anchor=(0.5, 0.01),
        ncol=3,
        frameon=True,
        fontsize=16,
    )

    fig.subplots_adjust(
        left=0.08,
        right=0.995,
        top=0.88,
        bottom=0.28,
        wspace=0.10,
    )

    PLOTDIR.mkdir(parents=True, exist_ok=True)



    outpath = PLOTDIR / f"TARGETED_UNTARGETED_IMPROVEMENT_THERMOELECTRIC_{MODEL}.png"
    fig.savefig(outpath, dpi=FIG_DPI, bbox_inches="tight")
    plt.show()

    print(f"[OK] Wrote plot to: {outpath}")
""" 




# (everything above remains EXACTLY the same as your file)

def plot_improvement(improv_df: pd.DataFrame) -> None:
    ylims = compute_global_improvement_limits(improv_df)

    fig, axes = plt.subplots(
        nrows=1,
        ncols=2,
        figsize=(15.8, 5.8),
        squeeze=False,
        sharey=True,
        gridspec_kw={"width_ratios": [1, 1.7]},
    )

    axL = axes[0, 0]
    axR = axes[0, 1]

    plot_targeted_panel_single_row(axL, improv_df)
    plot_untargeted_panel_single_row(axR, improv_df)

    # ------------------------------------------------------------
    # ✅ SET Y LIMITS
    # ------------------------------------------------------------
    axL.set_ylim(*ylims)
    axR.set_ylim(*ylims)

    # ------------------------------------------------------------
    # ✅ NEW: FORCE Y-TICKS EVERY 0.5
    # ------------------------------------------------------------
    ymin, ymax = ylims

    yticks = np.arange(
        np.floor(ymin / 0.5) * 0.5,
        np.ceil(ymax / 0.5) * 0.5 + 0.5,
        5
    )

    axL.set_yticks(yticks)
    axR.set_yticks(yticks)

    # ------------------------------------------------------------
    # TITLES / LABELS
    # ------------------------------------------------------------
    axL.set_title("Targeted", fontsize=18, fontweight="bold")
    axR.set_title("Untargeted", fontsize=18, fontweight="bold")

    axL.set_ylabel("Improvement vs random (%)", fontsize=18, fontweight="bold")

    # ------------------------------------------------------------
    # LEGEND
    # ------------------------------------------------------------
    policy_handles = [
        Patch(facecolor="white", edgecolor="black", hatch=None, label="QBC"),
        Patch(facecolor="white", edgecolor="black", hatch="///", label="QBC + diversity"),
    ]

    target_handles = [
        Patch(facecolor=TARGET_COLORS_DISPLAY["zT"], edgecolor="black", label="Target: zT"),
        Patch(facecolor=TARGET_COLORS_DISPLAY["Electrical Conductivity"], edgecolor="black", label="Target: Electrical Cond."),
        Patch(facecolor=TARGET_COLORS_DISPLAY["Thermal Conductivity"], edgecolor="black", label="Target: Thermal Cond."),
        Patch(facecolor=TARGET_COLORS_DISPLAY["Seebeck Coefficient"], edgecolor="black", label="Target: Seebeck Coeff."),
    ]

    fig.legend(
        handles=policy_handles + target_handles,
        loc="lower center",
        bbox_to_anchor=(0.5, 0.01),
        ncol=3,
        frameon=True,
        fontsize=16,
    )

    fig.subplots_adjust(
        left=0.08,
        right=0.995,
        top=0.88,
        bottom=0.28,
        wspace=0.10,
    )

    PLOTDIR.mkdir(parents=True, exist_ok=True)

    outpath = PLOTDIR / f"TARGETED_UNTARGETED_IMPROVEMENT_THERMOELECTRIC_{MODEL}.png"
    fig.savefig(outpath, dpi=FIG_DPI, bbox_inches="tight")
    plt.show()

    print(f"[OK] Wrote plot to: {outpath}")





def main() -> None:
    OUTDIR.mkdir(parents=True, exist_ok=True)
    PLOTDIR.mkdir(parents=True, exist_ok=True)

    agg = build_auc()
    improvement = build_improvement(agg)

    plot_improvement(improvement)


if __name__ == "__main__":
    main()
```

```python
#!/usr/bin/env python3
from __future__ import annotations

from pathlib import Path
from typing import List, Dict, Tuple, Optional
import itertools
import re

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib.patches import Patch


# ============================================================
# USER CONFIG
# ============================================================
SEEDS = range(5)
DATASET = "ThermoEl"
MODEL = "rf"   # change to "xgb" if needed

ROOT = Path(
    r"C:\Users\rafae\OneDrive\Imágenes\Documentos\PhDToronto\OLD_FILES\PhD Toronto"
    r"\DatasetGeneration\Better_Acqsition\ThermoElectric_NEW"
)

QBC_2_ROOT = ROOT / "al" / "grow_2QBC" / DATASET
NSGA_QBC_DIV_ROOT = ROOT / "al" / "grow_EA_3Obj_TwoTargets" / DATASET
NSGA_QBC_ROOT = ROOT / "al" / "grow_EA_2Obj_TwoTargets_QBC_ONLY" / DATASET
RANDOM_ROOT = ROOT / "al" / "grow_random" / DATASET / "all"

OUTDIR = ROOT / "AUC_2OBJ"
PLOTDIR = ROOT / "Plots" / "Plots_AUC"

FIG_DPI = 300
ERRORBAR_CAPSIZE = 3


# ============================================================
# PROPERTY / METRIC MAP
# ============================================================
TARGET_DISPLAY_ORDER = [
    "zT",
    "Electrical Conductivity",
    "Thermal Conductivity",
    "Seebeck Coefficient",
]

PROPERTY_KEYS = [
    "zT",
    "electric_cond",
    "thermal_cond",
    "seebeck_coef",
]

DISPLAY = {
    "zT": "zT",
    "electric_cond": "Electrical Conductivity",
    "thermal_cond": "Thermal Conductivity",
    "seebeck_coef": "Seebeck Coefficient",
}

DISPLAY_TO_KEY = {v: k for k, v in DISPLAY.items()}

METRIC_COLS = {
    "electric_cond": "rmse_elec_cond",
    "seebeck_coef": "rmse_seebeck",
    "thermal_cond": "rmse_k_total",
    "zT": "rmse_zt",
}


# ============================================================
# TWO-OUTCOME FOLDERS
# ============================================================
PAIR_FOLDER_TO_KEYS = {
    "el_seeb": ("electric_cond", "seebeck_coef"),
    "el_ThC": ("electric_cond", "thermal_cond"),
    "el_zT": ("electric_cond", "zT"),
    "seeb_ThC": ("seebeck_coef", "thermal_cond"),
    "seeb_zT": ("seebeck_coef", "zT"),
    "zT_ThC": ("zT", "thermal_cond"),
}

PAIR_FOLDERS = list(PAIR_FOLDER_TO_KEYS.keys())


def canonical_pair_from_keys(a: str, b: str) -> Tuple[str, str]:
    return tuple(sorted([DISPLAY[a], DISPLAY[b]]))


def pair_display_name(pair: Tuple[str, str]) -> str:
    a, b = tuple(sorted(pair))
    return f"{a} + {b}"


ALL_VALID_PAIRS = [
    tuple(sorted(p))
    for p in itertools.combinations(TARGET_DISPLAY_ORDER, 2)
]


# ============================================================
# POLICY DISPLAY
# ============================================================
POLICY_DISPLAY_ORDER = [
    "2QBC",
    "NSGA-II QBC",
    "NSGA-II QBC + Diversity",
]

STYLE_MAP = {
    "2QBC": {"hatch": None, "alpha": 1.0},
    "NSGA-II QBC": {"hatch": "//", "alpha": 1.0},
    "NSGA-II QBC + Diversity": {"hatch": "xx", "alpha": 1.0},
}


# ============================================================
# COLORS
# ============================================================
def canonical_metric_name(text: str) -> str:
    t = str(text).strip().lower()
    t = re.sub(r"\[[^\]]*\]", "", t)
    t = re.sub(r"\([^)]*\)", "", t)
    t = re.sub(r"\s+", " ", t).strip()

    if "random" in t:
        return "Random"
    if "zt" in t:
        return "zT"
    if "elect" in t or "elec" in t:
        return "Electrical Conductivity"
    if "thermal" in t or "k_total" in t:
        return "Thermal Conductivity"
    if "seebeck" in t:
        return "Seebeck Coefficient"

    return str(text).strip()


def canonical_pair(pair: Tuple[str, str]) -> Tuple[str, str]:
    return tuple(sorted([canonical_metric_name(pair[0]), canonical_metric_name(pair[1])]))


PAIR_COLORS = {
    canonical_pair(("zT", "Electrical Conductivity")): "#5E81AC",
    canonical_pair(("zT", "Thermal Conductivity")): "#A3BE8C",
    canonical_pair(("zT", "Seebeck Coefficient")): "#EBCB8B",
    canonical_pair(("Electrical Conductivity", "Thermal Conductivity")): "#D08770",
    canonical_pair(("Electrical Conductivity", "Seebeck Coefficient")): "#B48EAD",
    canonical_pair(("Thermal Conductivity", "Seebeck Coefficient")): "#88C0D0",
}


def pair_color(pair: Tuple[str, str]) -> str:
    cp = canonical_pair(pair)
    return PAIR_COLORS.get(cp, "#999999")


# ============================================================
# PATHS
# ============================================================
def path_2qbc(pair_folder: str, seed: int) -> Path:
    return (
        QBC_2_ROOT
        / pair_folder
        / str(seed)
        / f"ThermoEl_{pair_folder}growing_2QBC_{MODEL}{seed}.csv"
    )

def path_nsga_qbc_div(pair_folder: str, seed: int) -> Path:
    return (
        NSGA_QBC_DIV_ROOT
        / pair_folder
        / str(seed)
        / f"ThermoEl_{pair_folder}growing_EA_2out_div_{MODEL}{seed}.csv"
    )



def path_random(seed: int) -> Path:
    return RANDOM_ROOT / str(seed) / f"ThermoEl_all_random_{MODEL}{seed}.csv"


# ============================================================
# AUC HELPERS
# ============================================================
def normalize_columns(df: pd.DataFrame) -> pd.DataFrame:
    out = df.copy()
    out.columns = [str(c).strip() for c in out.columns]
    return out


def to_fraction_01(x: pd.Series) -> np.ndarray:
    arr = pd.to_numeric(x, errors="coerce").to_numpy(dtype=float)

    finite = np.isfinite(arr)
    if finite.any() and np.nanmax(arr[finite]) > 1.5:
        arr = arr / 100.0

    return arr


def auc_trapezoid(x: np.ndarray, y: np.ndarray) -> float:
    x = np.asarray(x, dtype=float)
    y = np.asarray(y, dtype=float)

    mask = np.isfinite(x) & np.isfinite(y)
    x = x[mask]
    y = y[mask]

    if len(x) < 2:
        return np.nan

    order = np.argsort(x)
    x = x[order]
    y = y[order]

    return float(np.trapezoid(y, x))


def get_xy_for_auc(df: pd.DataFrame, metric_col: str) -> Tuple[np.ndarray, np.ndarray]:
    if "train_ratio" not in df.columns:
        raise ValueError(f"'train_ratio' not found. Columns: {df.columns.tolist()}")

    if metric_col not in df.columns:
        raise ValueError(f"'{metric_col}' not found. Columns: {df.columns.tolist()}")

    tmp = df[["train_ratio", metric_col]].copy()
    tmp["x"] = to_fraction_01(tmp["train_ratio"])
    tmp["y"] = pd.to_numeric(tmp[metric_col], errors="coerce")
    tmp["_row_order"] = np.arange(len(tmp))

    tmp = tmp[np.isfinite(tmp["x"]) & np.isfinite(tmp["y"])].copy()

    rows = []

    for xval, g in tmp.groupby("x", sort=True):
        if np.isclose(xval, 1.0):
            yval = float(g.sort_values("_row_order").iloc[-1]["y"])
        else:
            yval = float(g["y"].mean())

        rows.append((float(xval), yval))

    out = pd.DataFrame(rows, columns=["x", "y"]).sort_values("x")

    return out["x"].to_numpy(), out["y"].to_numpy()


# ============================================================
# BUILD AUC TABLES FROM RAW CSVs
# ============================================================
def build_auc_tables() -> pd.DataFrame:
    rows = []

    policies = [
        ("2QBC", path_2qbc),
        ("NSGA-II QBC", path_nsga_qbc),
        ("NSGA-II QBC + Diversity", path_nsga_qbc_div),
    ]

    for pair_folder in PAIR_FOLDERS:
        pair_keys = PAIR_FOLDER_TO_KEYS[pair_folder]
        pair_display = canonical_pair_from_keys(*pair_keys)
        pair_name = pair_display_name(pair_display)

        for seed in SEEDS:
            for policy, path_fn in policies:
                path = path_fn(pair_folder, seed)

                if not path.exists():
                    print(f"[WARN] Missing file: {path}")
                    continue

                print(f"[INFO] Reading {policy}: {path}")

                df = normalize_columns(pd.read_csv(path))

                for eval_key in PROPERTY_KEYS:
                    metric_col = METRIC_COLS[eval_key]

                    if metric_col not in df.columns:
                        print(f"[WARN] Missing metric column {metric_col} in {path}")
                        continue

                    x, y = get_xy_for_auc(df, metric_col)

                    rows.append({
                        "dataset": DATASET,
                        "target_pair": pair_name,
                        "target_pair_tuple": pair_display,
                        "eval_property": DISPLAY[eval_key],
                        "policy": policy,
                        "seed": seed,
                        "auc": auc_trapezoid(x, y),
                        "source_file": str(path),
                        "metric_col": metric_col,
                    })

    # Random baseline
    for seed in SEEDS:
        path = path_random(seed)

        if not path.exists():
            print(f"[WARN] Missing random file: {path}")
            continue

        print(f"[INFO] Reading Random: {path}")

        df = normalize_columns(pd.read_csv(path))

        for eval_key in PROPERTY_KEYS:
            metric_col = METRIC_COLS[eval_key]

            if metric_col not in df.columns:
                print(f"[WARN] Missing metric column {metric_col} in {path}")
                continue

            x, y = get_xy_for_auc(df, metric_col)

            rows.append({
                "dataset": DATASET,
                "target_pair": "Random",
                "target_pair_tuple": None,
                "eval_property": DISPLAY[eval_key],
                "policy": "Random",
                "seed": seed,
                "auc": auc_trapezoid(x, y),
                "source_file": str(path),
                "metric_col": metric_col,
            })

    auc_seed_df = pd.DataFrame(rows)

    if auc_seed_df.empty:
        raise RuntimeError("AUC table is empty. Check file paths and column names.")

    auc_summary = (
        auc_seed_df
        .groupby(["dataset", "target_pair", "eval_property", "policy"], as_index=False)
        .agg(
            auc_mean=("auc", "mean"),
            auc_std=("auc", "std"),
        )
    )

    auc_summary["auc_std"] = auc_summary["auc_std"].fillna(0.0)

    OUTDIR.mkdir(parents=True, exist_ok=True)

    auc_seed_path = OUTDIR / f"auc_per_seed_2outcomes_{MODEL}.csv"
    auc_summary_path = OUTDIR / f"auc_summary_2outcomes_{MODEL}.csv"

    auc_seed_df.to_csv(auc_seed_path, index=False)
    auc_summary.to_csv(auc_summary_path, index=False)

    print(f"[OK] Saved per-seed AUC table to: {auc_seed_path}")
    print(f"[OK] Saved AUC summary table to: {auc_summary_path}")

    return auc_summary


# ============================================================
# IMPROVEMENT TABLE
# ============================================================
def build_improvement_table(auc_summary: pd.DataFrame) -> pd.DataFrame:
    rows = []

    for eval_prop in TARGET_DISPLAY_ORDER:
        sub_eval = auc_summary[auc_summary["eval_property"] == eval_prop].copy()

        if sub_eval.empty:
            print(f"[WARN] No rows found for eval property: {eval_prop}")
            continue

        rand = sub_eval[sub_eval["policy"] == "Random"]

        if rand.empty:
            print(f"[WARN] Missing random row for eval property: {eval_prop}")
            continue

        R = float(rand["auc_mean"].mean())
        sR = float(rand["auc_std"].mean())

        if not np.isfinite(R) or abs(R) < 1e-15:
            print(f"[WARN] Invalid random baseline for {eval_prop}")
            continue

        cand = sub_eval[sub_eval["policy"].isin(POLICY_DISPLAY_ORDER)].copy()

        for _, r in cand.iterrows():
            pair_name = str(r["target_pair"])
            pair = parse_pair_name(pair_name)

            if pair is None:
                continue

            P = float(r["auc_mean"])
            sP = float(r["auc_std"])

            improvement = 100.0 * (1.0 - P / R)

            dfdP = -100.0 / R
            dfdR = 100.0 * P / (R ** 2)
            s_improvement = float(np.sqrt((dfdP * sP) ** 2 + (dfdR * sR) ** 2))

            section = "Targeted" if eval_prop in pair else "Untargeted"

            rows.append({
                "eval_property": eval_prop,
                "target_pair": pair_display_name(pair),
                "target_pair_tuple": pair,
                "policy": str(r["policy"]),
                "section": section,
                "auc_mean": P,
                "auc_std": sP,
                "random_auc_mean": R,
                "random_auc_std": sR,
                "improvement_mean": improvement,
                "improvement_std": s_improvement,
            })

    out = pd.DataFrame(rows)

    if out.empty:
        raise RuntimeError("Improvement table is empty.")

    OUTDIR.mkdir(parents=True, exist_ok=True)
    out_path = OUTDIR / f"improvement_vs_random_2outcomes_{MODEL}.csv"
    out.to_csv(out_path, index=False)

    print(f"[OK] Saved improvement table to: {out_path}")

    return out


def parse_pair_name(pair_name: str) -> Optional[Tuple[str, str]]:
    if pair_name == "Random":
        return None

    parts = [p.strip() for p in pair_name.split("+")]

    if len(parts) != 2:
        print(f"[WARN] Could not parse pair name: {pair_name}")
        return None

    return canonical_pair((parts[0], parts[1]))


# ============================================================
# PLOTTING HELPERS
# ============================================================
def format_label(label: str) -> str:
    if label == "Electrical Conductivity":
        return "Electrical\nCond."
    if label == "Thermal Conductivity":
        return "Thermal\nCond."
    if label == "Seebeck Coefficient":
        return "Seebeck\nCoef."
    return label


def compute_global_improvement_limits(improv_df: pd.DataFrame) -> Tuple[float, float]:
    vals = improv_df["improvement_mean"].to_numpy(dtype=float)
    errs = improv_df["improvement_std"].to_numpy(dtype=float)

    ymin = np.nanmin(vals - errs)
    ymax = np.nanmax(vals + errs)

    pad = 0.08 * max(ymax - ymin, 1e-12)
    ymin -= pad
    ymax += pad

    ymin = min(ymin, 0.0)
    ymax = max(ymax, 0.0)

    return ymin, ymax


def make_group_positions() -> Tuple[List[Tuple[str, float]], List[float], List[str], Dict[str, float], List[float]]:
    group_spacing = 2.6

    cases = []
    xticks = []
    xticklabels = []
    group_centers = {}
    separators = []

    for i, prop in enumerate(TARGET_DISPLAY_ORDER):
        x = i * group_spacing

        cases.append((prop, x))
        xticks.append(x)
        xticklabels.append(prop)
        group_centers[prop] = x

        if i < len(TARGET_DISPLAY_ORDER) - 1:
            separators.append(x + group_spacing / 2.0)

    return cases, xticks, xticklabels, group_centers, separators


def pair_order_for_eval(eval_prop: str, section: str) -> List[Tuple[str, str]]:
    pairs = []

    for pair in ALL_VALID_PAIRS:
        is_targeted = eval_prop in pair

        if section == "Targeted" and is_targeted:
            pairs.append(pair)

        elif section == "Untargeted" and not is_targeted:
            pairs.append(pair)

    return pairs


def offsets_for_nbars(n: int, spread: float = 0.90) -> List[float]:
    if n == 1:
        return [0.0]

    return np.linspace(-spread, spread, n).tolist()


def plot_panel(ax, improv_df: pd.DataFrame, section: str) -> None:
    sub = improv_df[improv_df["section"] == section].copy()

    cases, xticks, xticklabels, _, separators = make_group_positions()

    pair_count = 3
    policy_count = len(POLICY_DISPLAY_ORDER)
    nbars = pair_count * policy_count

    width = 0.16
    offs = offsets_for_nbars(nbars, spread=0.88)

    for eval_prop, x in cases:
        pair_order = pair_order_for_eval(eval_prop, section)

        barspecs = []
        k = 0

        for pair_tuple in pair_order:
            for policy in POLICY_DISPLAY_ORDER:
                barspecs.append((pair_tuple, policy, offs[k]))
                k += 1

        for pair_tuple, policy, off in barspecs:
            pair_name = pair_display_name(pair_tuple)

            r = sub[
                (sub["eval_property"] == eval_prop)
                & (sub["target_pair"] == pair_name)
                & (sub["policy"] == policy)
            ]

            if r.empty:
                print(
                    f"[WARN] Missing row: section={section}, "
                    f"eval={eval_prop}, pair={pair_name}, policy={policy}"
                )
                continue

            rr = r.iloc[0]

            ax.bar(
                x + off,
                float(rr["improvement_mean"]),
                width=width,
                color=pair_color(pair_tuple),
                edgecolor="black",
                linewidth=0.8,
                hatch=STYLE_MAP[policy]["hatch"],
                alpha=STYLE_MAP[policy]["alpha"],
                yerr=float(rr["improvement_std"]),
                capsize=ERRORBAR_CAPSIZE,
                error_kw={"elinewidth": 1.0, "capthick": 1.0},
                zorder=3,
            )

    ax.axhline(0.0, color="black", linewidth=1.0, zorder=2)

    ax.set_xticks(xticks)
    ax.set_xticklabels(
        [format_label(l) for l in xticklabels],
        fontsize=18,
        rotation=0,
        ha="center",
        fontweight="bold",
    )

    ax.grid(axis="y", linestyle="--", alpha=0.28, zorder=0)

    for x_sep in separators:
        ax.axvline(x_sep, color="black", lw=1.0, alpha=0.30)

    ax.set_xlim(min(xticks) - 1.25, max(xticks) + 1.25)


def plot_improvement_figure(improv_df: pd.DataFrame) -> None:
    ylims = compute_global_improvement_limits(improv_df)

    fig, axes = plt.subplots(
        nrows=1,
        ncols=2,
        figsize=(18.0, 7.5),
        squeeze=False,
        sharey=True,
        gridspec_kw={"width_ratios": [1, 1]},
    )

    axL = axes[0, 0]
    axR = axes[0, 1]

    plot_panel(axL, improv_df, section="Targeted")
    plot_panel(axR, improv_df, section="Untargeted")

    axL.set_ylim(*ylims)
    axR.set_ylim(*ylims)

    axL.set_title("Targeted", fontsize=20, fontweight="bold")
    axR.set_title("Untargeted", fontsize=20, fontweight="bold")

    axL.set_ylabel("Improvement vs random (%)", fontsize=20, fontweight="bold")

    policy_handles = [
        Patch(facecolor="white", edgecolor="black", hatch=STYLE_MAP[p]["hatch"], label=p)
        for p in POLICY_DISPLAY_ORDER
    ]

    pair_handles = [
        Patch(facecolor=pair_color(("zT", "Electrical Conductivity")), edgecolor="black", label="zT & Electrical Cond."),
        Patch(facecolor=pair_color(("zT", "Thermal Conductivity")), edgecolor="black", label="zT & Thermal Cond."),
        Patch(facecolor=pair_color(("zT", "Seebeck Coefficient")), edgecolor="black", label="zT & Seebeck Coef."),
        Patch(facecolor=pair_color(("Electrical Conductivity", "Thermal Conductivity")), edgecolor="black", label="Electrical Cond. & Thermal Cond."),
        Patch(facecolor=pair_color(("Electrical Conductivity", "Seebeck Coefficient")), edgecolor="black", label="Electrical Cond. & Seebeck Coef."),
        Patch(facecolor=pair_color(("Thermal Conductivity", "Seebeck Coefficient")), edgecolor="black", label="Thermal Cond. & Seebeck Coef."),
    ]

    fig.legend(
        handles=policy_handles + pair_handles,
        loc="lower center",
        bbox_to_anchor=(0.5, 0.02),
        ncol=3,
        frameon=True,
        fontsize=15,
    )

    fig.subplots_adjust(
        left=0.075,
        right=0.995,
        top=0.88,
        bottom=0.31,
        wspace=0.08,
    )

    PLOTDIR.mkdir(parents=True, exist_ok=True)

    outpath = PLOTDIR / f"TARGETED_UNTARGETED_IMPROVEMENT_THERMOELECTRIC_2OBJECTIVES_{MODEL}.png"
    fig.savefig(outpath, dpi=FIG_DPI, bbox_inches="tight")
    plt.show()

    print(f"[OK] Wrote improvement plot to: {outpath}")


# ============================================================
# MAIN
# ============================================================
def main() -> None:
    OUTDIR.mkdir(parents=True, exist_ok=True)
    PLOTDIR.mkdir(parents=True, exist_ok=True)

    auc_summary = build_auc_tables()

    print("\n[DEBUG] AUC summary:")
    print(auc_summary.to_string(index=False))

    improv_df = build_improvement_table(auc_summary)

    print("\n[DEBUG] Improvement table:")
    print(
        improv_df.sort_values(
            ["eval_property", "section", "target_pair", "policy"]
        ).to_string(index=False)
    )

    plot_improvement_figure(improv_df)


if __name__ == "__main__":
    main()
```

```python
# -------------------------------------------------------------------------XGBOOST-----------------------------

#!/usr/bin/env python3
from __future__ import annotations

from pathlib import Path
from typing import List, Dict, Tuple, Optional
import itertools
import re

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib.patches import Patch


# ============================================================
# USER CONFIG
# ============================================================
SEEDS = range(5)
DATASET = "ThermoEl"

# XGBOOST VERSION
MODEL = "xgb"

ROOT = Path(
    r"C:\Users\rafae\OneDrive\Imágenes\Documentos\PhDToronto\OLD_FILES\PhD Toronto"
    r"\DatasetGeneration\Better_Acqsition\ThermoElectric_NEW"
)

QBC_2_ROOT = ROOT / "al" / "grow_2QBC" / DATASET
NSGA_QBC_DIV_ROOT = ROOT / "al" / "grow_EA_3Obj_TwoTargets" / DATASET
NSGA_QBC_ROOT = ROOT / "al" / "grow_EA_2Obj_TwoTargets_QBC_ONLY" / DATASET
RANDOM_ROOT = ROOT / "al" / "grow_random" / DATASET / "all"

OUTDIR = ROOT / "AUC_2OBJ_XGB"
PLOTDIR = ROOT / "Plots" / "Plots_AUC"

FIG_DPI = 300
ERRORBAR_CAPSIZE = 3


# ============================================================
# PROPERTY / METRIC MAP
# ============================================================
TARGET_DISPLAY_ORDER = [
    "zT",
    "Electrical Conductivity",
    "Thermal Conductivity",
    "Seebeck Coefficient",
]

PROPERTY_KEYS = [
    "zT",
    "electric_cond",
    "thermal_cond",
    "seebeck_coef",
]

DISPLAY = {
    "zT": "zT",
    "electric_cond": "Electrical Conductivity",
    "thermal_cond": "Thermal Conductivity",
    "seebeck_coef": "Seebeck Coefficient",
}

METRIC_COLS = {
    "electric_cond": "rmse_elec_cond",
    "seebeck_coef": "rmse_seebeck",
    "thermal_cond": "rmse_k_total",
    "zT": "rmse_zt",
}


# ============================================================
# TWO-OUTCOME FOLDERS
# ============================================================
PAIR_FOLDER_TO_KEYS = {
    "el_seeb": ("electric_cond", "seebeck_coef"),
    "el_ThC": ("electric_cond", "thermal_cond"),
    "el_zT": ("electric_cond", "zT"),
    "seeb_ThC": ("seebeck_coef", "thermal_cond"),
    "seeb_zT": ("seebeck_coef", "zT"),
    "zT_ThC": ("zT", "thermal_cond"),
}

PAIR_FOLDERS = list(PAIR_FOLDER_TO_KEYS.keys())


def canonical_pair_from_keys(a: str, b: str) -> Tuple[str, str]:
    return tuple(sorted([DISPLAY[a], DISPLAY[b]]))


def pair_display_name(pair: Tuple[str, str]) -> str:
    a, b = tuple(sorted(pair))
    return f"{a} + {b}"


ALL_VALID_PAIRS = [
    tuple(sorted(p))
    for p in itertools.combinations(TARGET_DISPLAY_ORDER, 2)
]


# ============================================================
# POLICY DISPLAY
# ============================================================
POLICY_DISPLAY_ORDER = [
    "2QBC",
    "NSGA-II QBC",
    "NSGA-II QBC + Diversity",
]

STYLE_MAP = {
    "2QBC": {"hatch": None, "alpha": 1.0},
    "NSGA-II QBC": {"hatch": "//", "alpha": 1.0},
    "NSGA-II QBC + Diversity": {"hatch": "xx", "alpha": 1.0},
}


# ============================================================
# COLORS
# ============================================================
def canonical_metric_name(text: str) -> str:
    t = str(text).strip().lower()
    t = re.sub(r"\[[^\]]*\]", "", t)
    t = re.sub(r"\([^)]*\)", "", t)
    t = re.sub(r"\s+", " ", t).strip()

    if "random" in t:
        return "Random"
    if "zt" in t:
        return "zT"
    if "elect" in t or "elec" in t:
        return "Electrical Conductivity"
    if "thermal" in t or "k_total" in t:
        return "Thermal Conductivity"
    if "seebeck" in t:
        return "Seebeck Coefficient"

    return str(text).strip()


def canonical_pair(pair: Tuple[str, str]) -> Tuple[str, str]:
    return tuple(sorted([canonical_metric_name(pair[0]), canonical_metric_name(pair[1])]))


PAIR_COLORS = {
    canonical_pair(("zT", "Electrical Conductivity")): "#5E81AC",
    canonical_pair(("zT", "Thermal Conductivity")): "#A3BE8C",
    canonical_pair(("zT", "Seebeck Coefficient")): "#EBCB8B",
    canonical_pair(("Electrical Conductivity", "Thermal Conductivity")): "#D08770",
    canonical_pair(("Electrical Conductivity", "Seebeck Coefficient")): "#B48EAD",
    canonical_pair(("Thermal Conductivity", "Seebeck Coefficient")): "#88C0D0",
}


def pair_color(pair: Tuple[str, str]) -> str:
    cp = canonical_pair(pair)
    return PAIR_COLORS.get(cp, "#999999")


# ============================================================
# PATHS — XGBOOST FILE PATTERNS
# ============================================================
def path_2qbc(pair_folder: str, seed: int) -> Path:
    return (
        QBC_2_ROOT
        / pair_folder
        / str(seed)
        / f"ThermoEl_{pair_folder}growing_2QBC_{MODEL}{seed}.csv"
    )


def path_nsga_qbc(pair_folder: str, seed: int) -> Path:
    return (
        NSGA_QBC_ROOT
        / pair_folder
        / str(seed)
        / f"ThermoEl_{pair_folder}growing_EA_2obj_qbc_only_{MODEL}{seed}.csv"
    )


def path_nsga_qbc_div(pair_folder: str, seed: int) -> Path:
    return (
        NSGA_QBC_DIV_ROOT
        / pair_folder
        / str(seed)
        / f"ThermoEl_{pair_folder}growing_EA_2out_div_{MODEL}{seed}.csv"
    )


def path_random(seed: int) -> Path:
    return RANDOM_ROOT / str(seed) / f"ThermoEl_all_random_{MODEL}{seed}.csv"


# ============================================================
# AUC HELPERS
# ============================================================
def normalize_columns(df: pd.DataFrame) -> pd.DataFrame:
    out = df.copy()
    out.columns = [str(c).strip() for c in out.columns]
    return out


def to_fraction_01(x: pd.Series) -> np.ndarray:
    arr = pd.to_numeric(x, errors="coerce").to_numpy(dtype=float)

    finite = np.isfinite(arr)
    if finite.any() and np.nanmax(arr[finite]) > 1.5:
        arr = arr / 100.0

    return arr


def auc_trapezoid(x: np.ndarray, y: np.ndarray) -> float:
    x = np.asarray(x, dtype=float)
    y = np.asarray(y, dtype=float)

    mask = np.isfinite(x) & np.isfinite(y)
    x = x[mask]
    y = y[mask]

    if len(x) < 2:
        return np.nan

    order = np.argsort(x)
    x = x[order]
    y = y[order]

    return float(np.trapezoid(y, x))


def get_xy_for_auc(df: pd.DataFrame, metric_col: str) -> Tuple[np.ndarray, np.ndarray]:
    if "train_ratio" not in df.columns:
        raise ValueError(f"'train_ratio' not found. Columns: {df.columns.tolist()}")

    if metric_col not in df.columns:
        raise ValueError(f"'{metric_col}' not found. Columns: {df.columns.tolist()}")

    tmp = df[["train_ratio", metric_col]].copy()
    tmp["x"] = to_fraction_01(tmp["train_ratio"])
    tmp["y"] = pd.to_numeric(tmp[metric_col], errors="coerce")
    tmp["_row_order"] = np.arange(len(tmp))

    tmp = tmp[np.isfinite(tmp["x"]) & np.isfinite(tmp["y"])].copy()

    rows = []

    for xval, g in tmp.groupby("x", sort=True):
        if np.isclose(xval, 1.0):
            yval = float(g.sort_values("_row_order").iloc[-1]["y"])
        else:
            yval = float(g["y"].mean())

        rows.append((float(xval), yval))

    out = pd.DataFrame(rows, columns=["x", "y"]).sort_values("x")

    return out["x"].to_numpy(), out["y"].to_numpy()


# ============================================================
# BUILD AUC TABLES FROM RAW CSVs
# ============================================================
def build_auc_tables() -> pd.DataFrame:
    rows = []

    print(f"[INFO] Running 2-objective ThermoEl pipeline with MODEL = {MODEL}")

    policies = [
        ("2QBC", path_2qbc),
        ("NSGA-II QBC", path_nsga_qbc),
        ("NSGA-II QBC + Diversity", path_nsga_qbc_div),
    ]

    for pair_folder in PAIR_FOLDERS:
        pair_keys = PAIR_FOLDER_TO_KEYS[pair_folder]
        pair_display = canonical_pair_from_keys(*pair_keys)
        pair_name = pair_display_name(pair_display)

        for seed in SEEDS:
            for policy, path_fn in policies:
                path = path_fn(pair_folder, seed)

                if not path.exists():
                    print(f"[WARN] Missing file: {path}")
                    continue

                print(f"[INFO] Reading {policy}: {path}")

                df = normalize_columns(pd.read_csv(path))

                for eval_key in PROPERTY_KEYS:
                    metric_col = METRIC_COLS[eval_key]

                    if metric_col not in df.columns:
                        print(f"[WARN] Missing metric column {metric_col} in {path}")
                        continue

                    x, y = get_xy_for_auc(df, metric_col)

                    rows.append({
                        "dataset": DATASET,
                        "target_pair": pair_name,
                        "target_pair_tuple": pair_display,
                        "eval_property": DISPLAY[eval_key],
                        "policy": policy,
                        "seed": seed,
                        "auc": auc_trapezoid(x, y),
                        "source_file": str(path),
                        "metric_col": metric_col,
                    })

    for seed in SEEDS:
        path = path_random(seed)

        if not path.exists():
            print(f"[WARN] Missing random file: {path}")
            continue

        print(f"[INFO] Reading Random: {path}")

        df = normalize_columns(pd.read_csv(path))

        for eval_key in PROPERTY_KEYS:
            metric_col = METRIC_COLS[eval_key]

            if metric_col not in df.columns:
                print(f"[WARN] Missing metric column {metric_col} in {path}")
                continue

            x, y = get_xy_for_auc(df, metric_col)

            rows.append({
                "dataset": DATASET,
                "target_pair": "Random",
                "target_pair_tuple": None,
                "eval_property": DISPLAY[eval_key],
                "policy": "Random",
                "seed": seed,
                "auc": auc_trapezoid(x, y),
                "source_file": str(path),
                "metric_col": metric_col,
            })

    auc_seed_df = pd.DataFrame(rows)

    if auc_seed_df.empty:
        raise RuntimeError("AUC table is empty. Check file paths and column names.")

    auc_summary = (
        auc_seed_df
        .groupby(["dataset", "target_pair", "eval_property", "policy"], as_index=False)
        .agg(
            auc_mean=("auc", "mean"),
            auc_std=("auc", "std"),
        )
    )

    auc_summary["auc_std"] = auc_summary["auc_std"].fillna(0.0)

    OUTDIR.mkdir(parents=True, exist_ok=True)

    auc_seed_path = OUTDIR / f"auc_per_seed_2outcomes_{MODEL}.csv"
    auc_summary_path = OUTDIR / f"auc_summary_2outcomes_{MODEL}.csv"

    auc_seed_df.to_csv(auc_seed_path, index=False)
    auc_summary.to_csv(auc_summary_path, index=False)

    print(f"[OK] Saved per-seed AUC table to: {auc_seed_path}")
    print(f"[OK] Saved AUC summary table to: {auc_summary_path}")

    return auc_summary


# ============================================================
# IMPROVEMENT TABLE
# ============================================================
def parse_pair_name(pair_name: str) -> Optional[Tuple[str, str]]:
    if pair_name == "Random":
        return None

    parts = [p.strip() for p in pair_name.split("+")]

    if len(parts) != 2:
        print(f"[WARN] Could not parse pair name: {pair_name}")
        return None

    return canonical_pair((parts[0], parts[1]))


def build_improvement_table(auc_summary: pd.DataFrame) -> pd.DataFrame:
    rows = []

    for eval_prop in TARGET_DISPLAY_ORDER:
        sub_eval = auc_summary[auc_summary["eval_property"] == eval_prop].copy()

        if sub_eval.empty:
            print(f"[WARN] No rows found for eval property: {eval_prop}")
            continue

        rand = sub_eval[sub_eval["policy"] == "Random"]

        if rand.empty:
            print(f"[WARN] Missing random row for eval property: {eval_prop}")
            continue

        R = float(rand["auc_mean"].mean())
        sR = float(rand["auc_std"].mean())

        if not np.isfinite(R) or abs(R) < 1e-15:
            print(f"[WARN] Invalid random baseline for {eval_prop}")
            continue

        cand = sub_eval[sub_eval["policy"].isin(POLICY_DISPLAY_ORDER)].copy()

        for _, r in cand.iterrows():
            pair_name = str(r["target_pair"])
            pair = parse_pair_name(pair_name)

            if pair is None:
                continue

            P = float(r["auc_mean"])
            sP = float(r["auc_std"])

            improvement = 100.0 * (1.0 - P / R)

            dfdP = -100.0 / R
            dfdR = 100.0 * P / (R ** 2)
            s_improvement = float(np.sqrt((dfdP * sP) ** 2 + (dfdR * sR) ** 2))

            section = "Targeted" if eval_prop in pair else "Untargeted"

            rows.append({
                "eval_property": eval_prop,
                "target_pair": pair_display_name(pair),
                "target_pair_tuple": pair,
                "policy": str(r["policy"]),
                "section": section,
                "auc_mean": P,
                "auc_std": sP,
                "random_auc_mean": R,
                "random_auc_std": sR,
                "improvement_mean": improvement,
                "improvement_std": s_improvement,
            })

    out = pd.DataFrame(rows)

    if out.empty:
        raise RuntimeError("Improvement table is empty.")

    OUTDIR.mkdir(parents=True, exist_ok=True)
    out_path = OUTDIR / f"improvement_vs_random_2outcomes_{MODEL}.csv"
    out.to_csv(out_path, index=False)

    print(f"[OK] Saved improvement table to: {out_path}")

    return out


# ============================================================
# PLOTTING HELPERS
# ============================================================
def format_label(label: str) -> str:
    if label == "Electrical Conductivity":
        return "Electrical\nCond."
    if label == "Thermal Conductivity":
        return "Thermal\nCond."
    if label == "Seebeck Coefficient":
        return "Seebeck\nCoef."
    return label


def compute_global_improvement_limits(improv_df: pd.DataFrame) -> Tuple[float, float]:
    vals = improv_df["improvement_mean"].to_numpy(dtype=float)
    errs = improv_df["improvement_std"].to_numpy(dtype=float)

    ymin = np.nanmin(vals - errs)
    ymax = np.nanmax(vals + errs)

    pad = 0.08 * max(ymax - ymin, 1e-12)
    ymin -= pad
    ymax += pad

    ymin = min(ymin, 0.0)
    ymax = max(ymax, 0.0)

    return ymin, ymax


def make_group_positions() -> Tuple[List[Tuple[str, float]], List[float], List[str], Dict[str, float], List[float]]:
    group_spacing = 2.6

    cases = []
    xticks = []
    xticklabels = []
    group_centers = {}
    separators = []

    for i, prop in enumerate(TARGET_DISPLAY_ORDER):
        x = i * group_spacing

        cases.append((prop, x))
        xticks.append(x)
        xticklabels.append(prop)
        group_centers[prop] = x

        if i < len(TARGET_DISPLAY_ORDER) - 1:
            separators.append(x + group_spacing / 2.0)

    return cases, xticks, xticklabels, group_centers, separators


def pair_order_for_eval(eval_prop: str, section: str) -> List[Tuple[str, str]]:
    pairs = []

    for pair in ALL_VALID_PAIRS:
        is_targeted = eval_prop in pair

        if section == "Targeted" and is_targeted:
            pairs.append(pair)

        elif section == "Untargeted" and not is_targeted:
            pairs.append(pair)

    return pairs


def offsets_for_nbars(n: int, spread: float = 0.90) -> List[float]:
    if n == 1:
        return [0.0]

    return np.linspace(-spread, spread, n).tolist()


def plot_panel(ax, improv_df: pd.DataFrame, section: str) -> None:
    sub = improv_df[improv_df["section"] == section].copy()

    cases, xticks, xticklabels, _, separators = make_group_positions()

    pair_count = 3
    policy_count = len(POLICY_DISPLAY_ORDER)
    nbars = pair_count * policy_count

    width = 0.16
    offs = offsets_for_nbars(nbars, spread=0.88)

    for eval_prop, x in cases:
        pair_order = pair_order_for_eval(eval_prop, section)

        barspecs = []
        k = 0

        for pair_tuple in pair_order:
            for policy in POLICY_DISPLAY_ORDER:
                barspecs.append((pair_tuple, policy, offs[k]))
                k += 1

        for pair_tuple, policy, off in barspecs:
            pair_name = pair_display_name(pair_tuple)

            r = sub[
                (sub["eval_property"] == eval_prop)
                & (sub["target_pair"] == pair_name)
                & (sub["policy"] == policy)
            ]

            if r.empty:
                print(
                    f"[WARN] Missing row: section={section}, "
                    f"eval={eval_prop}, pair={pair_name}, policy={policy}"
                )
                continue

            rr = r.iloc[0]

            ax.bar(
                x + off,
                float(rr["improvement_mean"]),
                width=width,
                color=pair_color(pair_tuple),
                edgecolor="black",
                linewidth=0.8,
                hatch=STYLE_MAP[policy]["hatch"],
                alpha=STYLE_MAP[policy]["alpha"],
                yerr=float(rr["improvement_std"]),
                capsize=ERRORBAR_CAPSIZE,
                error_kw={"elinewidth": 1.0, "capthick": 1.0},
                zorder=3,
            )

    ax.axhline(0.0, color="black", linewidth=1.0, zorder=2)

    ax.set_xticks(xticks)
    ax.set_xticklabels(
        [format_label(l) for l in xticklabels],
        fontsize=18,
        rotation=0,
        ha="center",
        fontweight="bold",
    )

    ax.grid(axis="y", linestyle="--", alpha=0.28, zorder=0)

    for x_sep in separators:
        ax.axvline(x_sep, color="black", lw=1.0, alpha=0.30)

    ax.set_xlim(min(xticks) - 1.25, max(xticks) + 1.25)


def plot_improvement_figure(improv_df: pd.DataFrame) -> None:
    ylims = compute_global_improvement_limits(improv_df)

    fig, axes = plt.subplots(
        nrows=1,
        ncols=2,
        figsize=(18.0, 7.5),
        squeeze=False,
        sharey=True,
        gridspec_kw={"width_ratios": [1, 1]},
    )

    axL = axes[0, 0]
    axR = axes[0, 1]

    plot_panel(axL, improv_df, section="Targeted")
    plot_panel(axR, improv_df, section="Untargeted")

    axL.set_ylim(*ylims)
    axR.set_ylim(*ylims)

    axL.set_title("Targeted", fontsize=20, fontweight="bold")
    axR.set_title("Untargeted", fontsize=20, fontweight="bold")

    axL.set_ylabel("Improvement vs random (%)", fontsize=20, fontweight="bold")

    policy_handles = [
        Patch(facecolor="white", edgecolor="black", hatch=STYLE_MAP[p]["hatch"], label=p)
        for p in POLICY_DISPLAY_ORDER
    ]

    pair_handles = [
        Patch(facecolor=pair_color(("zT", "Electrical Conductivity")), edgecolor="black", label="zT & Electrical Cond."),
        Patch(facecolor=pair_color(("zT", "Thermal Conductivity")), edgecolor="black", label="zT & Thermal Cond."),
        Patch(facecolor=pair_color(("zT", "Seebeck Coefficient")), edgecolor="black", label="zT & Seebeck Coef."),
        Patch(facecolor=pair_color(("Electrical Conductivity", "Thermal Conductivity")), edgecolor="black", label="Electrical Cond. & Thermal Cond."),
        Patch(facecolor=pair_color(("Electrical Conductivity", "Seebeck Coefficient")), edgecolor="black", label="Electrical Cond. & Seebeck Coef."),
        Patch(facecolor=pair_color(("Thermal Conductivity", "Seebeck Coefficient")), edgecolor="black", label="Thermal Cond. & Seebeck Coef."),
    ]

    fig.legend(
        handles=policy_handles + pair_handles,
        loc="lower center",
        bbox_to_anchor=(0.5, 0.02),
        ncol=3,
        frameon=True,
        fontsize=15,
    )

    fig.subplots_adjust(
        left=0.075,
        right=0.995,
        top=0.88,
        bottom=0.31,
        wspace=0.08,
    )

    PLOTDIR.mkdir(parents=True, exist_ok=True)

    outpath = PLOTDIR / f"TARGETED_UNTARGETED_IMPROVEMENT_THERMOELECTRIC_2OBJECTIVES_{MODEL}.png"
    fig.savefig(outpath, dpi=FIG_DPI, bbox_inches="tight")
    plt.show()

    print(f"[OK] Wrote improvement plot to: {outpath}")


# ============================================================
# MAIN
# ============================================================
def main() -> None:
    OUTDIR.mkdir(parents=True, exist_ok=True)
    PLOTDIR.mkdir(parents=True, exist_ok=True)

    auc_summary = build_auc_tables()

    print("\n[DEBUG] AUC summary:")
    print(auc_summary.to_string(index=False))

    improv_df = build_improvement_table(auc_summary)

    print("\n[DEBUG] Improvement table:")
    print(
        improv_df.sort_values(
            ["eval_property", "section", "target_pair", "policy"]
        ).to_string(index=False)
    )

    plot_improvement_figure(improv_df)


if __name__ == "__main__":
    main()
```

```python
#!/usr/bin/env python3
from __future__ import annotations

from pathlib import Path
from typing import List, Dict, Tuple, Optional
import itertools
import re

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib.patches import Patch


# ============================================================
# USER CONFIG
# ============================================================
SEEDS = range(5)
DATASET = "ThermoEl"
MODEL = "rf"

ROOT = Path(
    r"C:\Users\rafae\OneDrive\Imágenes\Documentos\PhDToronto\OLD_FILES\PhD Toronto"
    r"\DatasetGeneration\Better_Acqsition\ThermoElectric_NEW"
)

QBC_3_ROOT = ROOT / "al" / "grow_3QBC" / DATASET
NSGA_QBC_ROOT = ROOT / "al" / "grow_EA_3Obj_ThreeTargets_QBC_ONLY" / DATASET
NSGA_QBC_DIV_ROOT = ROOT / "al" / "grow_EA_4Obj_ThreeTargets" / DATASET
RANDOM_ROOT = ROOT / "al" / "grow_random" / DATASET / "all"

OUTDIR = ROOT / "AUC_3OBJ"
PLOTDIR = ROOT / "Plots" / "Plots_AUC"

FIG_DPI = 300
ERRORBAR_CAPSIZE = 3


# ============================================================
# PROPERTIES
# ============================================================
TARGET_DISPLAY_ORDER = [
    "zT",
    "Electrical Conductivity",
    "Thermal Conductivity",
    "Seebeck Coefficient",
]

PROPERTY_KEYS = [
    "zT",
    "electric_cond",
    "thermal_cond",
    "seebeck_coef",
]

DISPLAY = {
    "zT": "zT",
    "electric_cond": "Electrical Conductivity",
    "thermal_cond": "Thermal Conductivity",
    "seebeck_coef": "Seebeck Coefficient",
}

METRIC_COLS = {
    "electric_cond": "rmse_elec_cond",
    "seebeck_coef": "rmse_seebeck",
    "thermal_cond": "rmse_k_total",
    "zT": "rmse_zt",
}


# ============================================================
# THREE-OUTCOME FOLDERS
# Adjust folder names here if your actual folders differ.
# ============================================================
TRIPLE_FOLDER_TO_KEYS = {
    "el_seeb_ThC": ("electric_cond", "seebeck_coef", "thermal_cond"),
    "el_seeb_zT": ("electric_cond", "seebeck_coef", "zT"),
    "el_zT_ThC": ("electric_cond", "thermal_cond", "zT"),
    "seeb_zT_ThC": ("seebeck_coef", "thermal_cond", "zT"),
}

TRIPLE_FOLDERS = list(TRIPLE_FOLDER_TO_KEYS.keys())


def canonical_triple_from_keys(a: str, b: str, c: str) -> Tuple[str, str, str]:
    return tuple(sorted([DISPLAY[a], DISPLAY[b], DISPLAY[c]]))


def triple_display_name(triple: Tuple[str, str, str]) -> str:
    a, b, c = tuple(sorted(triple))
    return f"{a} + {b} + {c}"


ALL_VALID_TRIPLES = [
    tuple(sorted(t))
    for t in itertools.combinations(TARGET_DISPLAY_ORDER, 3)
]


# ============================================================
# POLICIES
# ============================================================
POLICY_DISPLAY_ORDER = [
    "3QBC",
    "NSGA-II QBC",
    "NSGA-II QBC + Diversity",
]

STYLE_MAP = {
    "3QBC": {"hatch": None, "alpha": 1.0},
    "NSGA-II QBC": {"hatch": "//", "alpha": 1.0},
    "NSGA-II QBC + Diversity": {"hatch": "xx", "alpha": 1.0},
}


# ============================================================
# COLORS
# ============================================================
def canonical_metric_name(text: str) -> str:
    t = str(text).strip().lower()
    t = re.sub(r"\[[^\]]*\]", "", t)
    t = re.sub(r"\([^)]*\)", "", t)
    t = re.sub(r"\s+", " ", t).strip()

    if "random" in t:
        return "Random"
    if "zt" in t:
        return "zT"
    if "elect" in t or "elec" in t:
        return "Electrical Conductivity"
    if "thermal" in t or "k_total" in t:
        return "Thermal Conductivity"
    if "seebeck" in t:
        return "Seebeck Coefficient"

    return str(text).strip()


def canonical_triple(triple: Tuple[str, str, str]) -> Tuple[str, str, str]:
    return tuple(sorted([canonical_metric_name(x) for x in triple]))


TRIPLE_COLORS = {
    canonical_triple(("Electrical Conductivity", "Seebeck Coefficient", "Thermal Conductivity")): "#5E81AC",
    canonical_triple(("Electrical Conductivity", "Seebeck Coefficient", "zT")): "#A3BE8C",
    canonical_triple(("Electrical Conductivity", "Thermal Conductivity", "zT")): "#D08770",
    canonical_triple(("Seebeck Coefficient", "Thermal Conductivity", "zT")): "#B48EAD",
}


def triple_color(triple: Tuple[str, str, str]) -> str:
    ct = canonical_triple(triple)
    return TRIPLE_COLORS.get(ct, "#999999")


# ============================================================
# PATH HELPERS
# ============================================================
def find_matching_csv(folder: Path, required_tokens: List[str]) -> Optional[Path]:
    """
    Robust file finder:
    - Only uses CSV files.
    - Excludes trainIndices files.
    - Requires all tokens to appear in filename.
    """
    if not folder.exists():
        print(f"[WARN] Folder does not exist: {folder}")
        return None

    csvs = sorted(folder.glob("*.csv"))
    csvs = [p for p in csvs if "trainindices" not in p.name.lower()]

    matches = []
    for p in csvs:
        name = p.name.lower()
        if all(tok.lower() in name for tok in required_tokens):
            matches.append(p)

    if not matches:
        print(f"[WARN] No matching CSV in {folder}")
        print(f"       Required tokens: {required_tokens}")
        print(f"       Available CSVs: {[p.name for p in csvs]}")
        return None

    if len(matches) > 1:
        print(f"[WARN] Multiple matches in {folder}. Using first:")
        for p in matches:
            print(f"       - {p.name}")

    return matches[0]


def path_3qbc(triple_folder: str, seed: int) -> Optional[Path]:
    folder = QBC_3_ROOT / triple_folder / str(seed)
    return find_matching_csv(
        folder,
        required_tokens=[
            f"ThermoEl_{triple_folder}".lower(),
            "3qbc",
            MODEL,
            str(seed),
        ],
    )


def path_nsga_qbc(triple_folder: str, seed: int) -> Optional[Path]:
    folder = NSGA_QBC_ROOT / triple_folder / str(seed)
    return find_matching_csv(
        folder,
        required_tokens=[
            f"ThermoEl_{triple_folder}".lower(),
            "qbc_only",
            MODEL,
            str(seed),
        ],
    )


def path_nsga_qbc_div(triple_folder: str, seed: int) -> Optional[Path]:
    folder = NSGA_QBC_DIV_ROOT / triple_folder / str(seed)
    return find_matching_csv(
        folder,
        required_tokens=[
            f"ThermoEl_{triple_folder}".lower(),
            "div",
            MODEL,
            str(seed),
        ],
    )


def path_random(seed: int) -> Path:
    return RANDOM_ROOT / str(seed) / f"ThermoEl_all_random_{MODEL}{seed}.csv"


# ============================================================
# AUC HELPERS
# ============================================================
def normalize_columns(df: pd.DataFrame) -> pd.DataFrame:
    out = df.copy()
    out.columns = [str(c).strip() for c in out.columns]
    return out


def to_fraction_01(x: pd.Series) -> np.ndarray:
    arr = pd.to_numeric(x, errors="coerce").to_numpy(dtype=float)

    finite = np.isfinite(arr)
    if finite.any() and np.nanmax(arr[finite]) > 1.5:
        arr = arr / 100.0

    return arr


def auc_trapezoid(x: np.ndarray, y: np.ndarray) -> float:
    x = np.asarray(x, dtype=float)
    y = np.asarray(y, dtype=float)

    mask = np.isfinite(x) & np.isfinite(y)
    x = x[mask]
    y = y[mask]

    if len(x) < 2:
        return np.nan

    order = np.argsort(x)
    x = x[order]
    y = y[order]

    return float(np.trapezoid(y, x))


def get_xy_for_auc(df: pd.DataFrame, metric_col: str) -> Tuple[np.ndarray, np.ndarray]:
    if "train_ratio" not in df.columns:
        raise ValueError(f"'train_ratio' not found. Columns: {df.columns.tolist()}")

    if metric_col not in df.columns:
        raise ValueError(f"'{metric_col}' not found. Columns: {df.columns.tolist()}")

    tmp = df[["train_ratio", metric_col]].copy()
    tmp["x"] = to_fraction_01(tmp["train_ratio"])
    tmp["y"] = pd.to_numeric(tmp[metric_col], errors="coerce")
    tmp["_row_order"] = np.arange(len(tmp))

    tmp = tmp[np.isfinite(tmp["x"]) & np.isfinite(tmp["y"])].copy()

    rows = []

    for xval, g in tmp.groupby("x", sort=True):
        if np.isclose(xval, 1.0):
            yval = float(g.sort_values("_row_order").iloc[-1]["y"])
        else:
            yval = float(g["y"].mean())

        rows.append((float(xval), yval))

    out = pd.DataFrame(rows, columns=["x", "y"]).sort_values("x")

    return out["x"].to_numpy(), out["y"].to_numpy()


# ============================================================
# BUILD AUC TABLES
# ============================================================
def build_auc_tables() -> pd.DataFrame:
    rows = []

    print(f"[INFO] Running 3-objective ThermoEl pipeline with MODEL = {MODEL}")

    policies = [
        ("3QBC", path_3qbc),
        ("NSGA-II QBC", path_nsga_qbc),
        ("NSGA-II QBC + Diversity", path_nsga_qbc_div),
    ]

    for triple_folder in TRIPLE_FOLDERS:
        triple_keys = TRIPLE_FOLDER_TO_KEYS[triple_folder]
        triple_display = canonical_triple_from_keys(*triple_keys)
        triple_name = triple_display_name(triple_display)

        for seed in SEEDS:
            for policy, path_fn in policies:
                path = path_fn(triple_folder, seed)

                if path is None or not path.exists():
                    print(f"[WARN] Missing file for policy={policy}, triple={triple_folder}, seed={seed}")
                    continue

                print(f"[INFO] Reading {policy}: {path}")

                df = normalize_columns(pd.read_csv(path))

                for eval_key in PROPERTY_KEYS:
                    metric_col = METRIC_COLS[eval_key]

                    if metric_col not in df.columns:
                        print(f"[WARN] Missing metric column {metric_col} in {path}")
                        continue

                    x, y = get_xy_for_auc(df, metric_col)

                    rows.append({
                        "dataset": DATASET,
                        "target_triple": triple_name,
                        "target_triple_tuple": triple_display,
                        "eval_property": DISPLAY[eval_key],
                        "policy": policy,
                        "seed": seed,
                        "auc": auc_trapezoid(x, y),
                        "source_file": str(path),
                        "metric_col": metric_col,
                    })

    # Random baseline
    for seed in SEEDS:
        path = path_random(seed)

        if not path.exists():
            print(f"[WARN] Missing random file: {path}")
            continue

        print(f"[INFO] Reading Random: {path}")

        df = normalize_columns(pd.read_csv(path))

        for eval_key in PROPERTY_KEYS:
            metric_col = METRIC_COLS[eval_key]

            if metric_col not in df.columns:
                print(f"[WARN] Missing metric column {metric_col} in {path}")
                continue

            x, y = get_xy_for_auc(df, metric_col)

            rows.append({
                "dataset": DATASET,
                "target_triple": "Random",
                "target_triple_tuple": None,
                "eval_property": DISPLAY[eval_key],
                "policy": "Random",
                "seed": seed,
                "auc": auc_trapezoid(x, y),
                "source_file": str(path),
                "metric_col": metric_col,
            })

    auc_seed_df = pd.DataFrame(rows)

    if auc_seed_df.empty:
        raise RuntimeError("AUC table is empty. Check file paths and column names.")

    auc_summary = (
        auc_seed_df
        .groupby(["dataset", "target_triple", "eval_property", "policy"], as_index=False)
        .agg(
            auc_mean=("auc", "mean"),
            auc_std=("auc", "std"),
        )
    )

    auc_summary["auc_std"] = auc_summary["auc_std"].fillna(0.0)

    OUTDIR.mkdir(parents=True, exist_ok=True)

    auc_seed_path = OUTDIR / f"auc_per_seed_3outcomes_{MODEL}.csv"
    auc_summary_path = OUTDIR / f"auc_summary_3outcomes_{MODEL}.csv"

    auc_seed_df.to_csv(auc_seed_path, index=False)
    auc_summary.to_csv(auc_summary_path, index=False)

    print(f"[OK] Saved per-seed AUC table to: {auc_seed_path}")
    print(f"[OK] Saved AUC summary table to: {auc_summary_path}")

    return auc_summary


# ============================================================
# IMPROVEMENT TABLE
# ============================================================
def parse_triple_name(triple_name: str) -> Optional[Tuple[str, str, str]]:
    if triple_name == "Random":
        return None

    parts = [p.strip() for p in triple_name.split("+")]

    if len(parts) != 3:
        print(f"[WARN] Could not parse triple name: {triple_name}")
        return None

    return canonical_triple((parts[0], parts[1], parts[2]))


def build_improvement_table(auc_summary: pd.DataFrame) -> pd.DataFrame:
    rows = []

    for eval_prop in TARGET_DISPLAY_ORDER:
        sub_eval = auc_summary[auc_summary["eval_property"] == eval_prop].copy()

        if sub_eval.empty:
            print(f"[WARN] No rows found for eval property: {eval_prop}")
            continue

        rand = sub_eval[sub_eval["policy"] == "Random"]

        if rand.empty:
            print(f"[WARN] Missing random row for eval property: {eval_prop}")
            continue

        R = float(rand["auc_mean"].mean())
        sR = float(rand["auc_std"].mean())

        if not np.isfinite(R) or abs(R) < 1e-15:
            print(f"[WARN] Invalid random baseline for {eval_prop}")
            continue

        cand = sub_eval[sub_eval["policy"].isin(POLICY_DISPLAY_ORDER)].copy()

        for _, r in cand.iterrows():
            triple_name = str(r["target_triple"])
            triple = parse_triple_name(triple_name)

            if triple is None:
                continue

            P = float(r["auc_mean"])
            sP = float(r["auc_std"])

            improvement = 100.0 * (1.0 - P / R)

            dfdP = -100.0 / R
            dfdR = 100.0 * P / (R ** 2)
            s_improvement = float(np.sqrt((dfdP * sP) ** 2 + (dfdR * sR) ** 2))

            section = "Targeted" if eval_prop in triple else "Untargeted"

            rows.append({
                "eval_property": eval_prop,
                "target_triple": triple_display_name(triple),
                "target_triple_tuple": triple,
                "policy": str(r["policy"]),
                "section": section,
                "auc_mean": P,
                "auc_std": sP,
                "random_auc_mean": R,
                "random_auc_std": sR,
                "improvement_mean": improvement,
                "improvement_std": s_improvement,
            })

    out = pd.DataFrame(rows)

    if out.empty:
        raise RuntimeError("Improvement table is empty.")

    OUTDIR.mkdir(parents=True, exist_ok=True)

    out_path = OUTDIR / f"improvement_vs_random_3outcomes_{MODEL}.csv"
    out.to_csv(out_path, index=False)

    print(f"[OK] Saved improvement table to: {out_path}")

    return out


# ============================================================
# PLOTTING
# ============================================================
def format_label(label: str) -> str:
    if label == "Electrical Conductivity":
        return "Electrical\nCond."
    if label == "Thermal Conductivity":
        return "Thermal\nCond."
    if label == "Seebeck Coefficient":
        return "Seebeck\nCoef."
    return label


def compute_global_improvement_limits(improv_df: pd.DataFrame) -> Tuple[float, float]:
    vals = improv_df["improvement_mean"].to_numpy(dtype=float)
    errs = improv_df["improvement_std"].to_numpy(dtype=float)

    ymin = np.nanmin(vals - errs)
    ymax = np.nanmax(vals + errs)

    pad = 0.08 * max(ymax - ymin, 1e-12)
    ymin -= pad
    ymax += pad

    ymin = min(ymin, 0.0)
    ymax = max(ymax, 0.0)

    return ymin, ymax


def make_group_positions() -> Tuple[List[Tuple[str, float]], List[float], List[str], Dict[str, float], List[float]]:
    group_spacing = 2.6

    cases = []
    xticks = []
    xticklabels = []
    group_centers = {}
    separators = []

    for i, prop in enumerate(TARGET_DISPLAY_ORDER):
        x = i * group_spacing

        cases.append((prop, x))
        xticks.append(x)
        xticklabels.append(prop)
        group_centers[prop] = x

        if i < len(TARGET_DISPLAY_ORDER) - 1:
            separators.append(x + group_spacing / 2.0)

    return cases, xticks, xticklabels, group_centers, separators


def triple_order_for_eval(eval_prop: str, section: str) -> List[Tuple[str, str, str]]:
    triples = []

    for triple in ALL_VALID_TRIPLES:
        is_targeted = eval_prop in triple

        if section == "Targeted" and is_targeted:
            triples.append(triple)

        elif section == "Untargeted" and not is_targeted:
            triples.append(triple)

    return triples


def offsets_for_nbars(n: int, spread: float = 0.90) -> List[float]:
    if n == 1:
        return [0.0]

    return np.linspace(-spread, spread, n).tolist()


def plot_panel(ax, improv_df: pd.DataFrame, section: str) -> None:
    sub = improv_df[improv_df["section"] == section].copy()

    cases, xticks, xticklabels, _, separators = make_group_positions()

    if section == "Targeted":
        triple_count = 3
    else:
        triple_count = 1

    policy_count = len(POLICY_DISPLAY_ORDER)
    nbars = triple_count * policy_count

    width = 0.16 if section == "Targeted" else 0.28
    offs = offsets_for_nbars(nbars, spread=0.88 if section == "Targeted" else 0.38)

    for eval_prop, x in cases:
        triple_order = triple_order_for_eval(eval_prop, section)

        barspecs = []
        k = 0

        for triple_tuple in triple_order:
            for policy in POLICY_DISPLAY_ORDER:
                barspecs.append((triple_tuple, policy, offs[k]))
                k += 1

        for triple_tuple, policy, off in barspecs:
            triple_name = triple_display_name(triple_tuple)

            r = sub[
                (sub["eval_property"] == eval_prop)
                & (sub["target_triple"] == triple_name)
                & (sub["policy"] == policy)
            ]

            if r.empty:
                print(
                    f"[WARN] Missing row: section={section}, "
                    f"eval={eval_prop}, triple={triple_name}, policy={policy}"
                )
                continue

            rr = r.iloc[0]

            ax.bar(
                x + off,
                float(rr["improvement_mean"]),
                width=width,
                color=triple_color(triple_tuple),
                edgecolor="black",
                linewidth=0.8,
                hatch=STYLE_MAP[policy]["hatch"],
                alpha=STYLE_MAP[policy]["alpha"],
                yerr=float(rr["improvement_std"]),
                capsize=ERRORBAR_CAPSIZE,
                error_kw={"elinewidth": 1.0, "capthick": 1.0},
                zorder=3,
            )

    ax.axhline(0.0, color="black", linewidth=1.0, zorder=2)

    ax.set_xticks(xticks)
    ax.set_xticklabels(
        [format_label(l) for l in xticklabels],
        fontsize=18,
        rotation=0,
        ha="center",
        fontweight="bold",
    )

    ax.grid(axis="y", linestyle="--", alpha=0.28, zorder=0)

    for x_sep in separators:
        ax.axvline(x_sep, color="black", lw=1.0, alpha=0.30)

    ax.set_xlim(min(xticks) - 1.25, max(xticks) + 1.25)


def plot_improvement_figure(improv_df: pd.DataFrame) -> None:
    ylims = compute_global_improvement_limits(improv_df)

    fig, axes = plt.subplots(
        nrows=1,
        ncols=2,
        figsize=(18.0, 7.5),
        squeeze=False,
        sharey=True,
        gridspec_kw={"width_ratios": [1, 1]},
    )

    axL = axes[0, 0]
    axR = axes[0, 1]

    plot_panel(axL, improv_df, section="Targeted")
    plot_panel(axR, improv_df, section="Untargeted")

    axL.set_ylim(*ylims)
    axR.set_ylim(*ylims)

    axL.set_title("Targeted", fontsize=20, fontweight="bold")
    axR.set_title("Untargeted", fontsize=20, fontweight="bold")

    axL.set_ylabel("Improvement vs random (%)", fontsize=20, fontweight="bold")

    policy_handles = [
        Patch(facecolor="white", edgecolor="black", hatch=STYLE_MAP[p]["hatch"], label=p)
        for p in POLICY_DISPLAY_ORDER
    ]

    triple_handles = [
        Patch(
            facecolor=triple_color(("Electrical Conductivity", "Seebeck Coefficient", "Thermal Conductivity")),
            edgecolor="black",
            label="Electrical Cond. & Seebeck Coef. & Thermal Cond.",
        ),
        Patch(
            facecolor=triple_color(("Electrical Conductivity", "Seebeck Coefficient", "zT")),
            edgecolor="black",
            label="Electrical Cond. & Seebeck Coef. & zT",
        ),
        Patch(
            facecolor=triple_color(("Electrical Conductivity", "Thermal Conductivity", "zT")),
            edgecolor="black",
            label="Electrical Cond. & Thermal Cond. & zT",
        ),
        Patch(
            facecolor=triple_color(("Seebeck Coefficient", "Thermal Conductivity", "zT")),
            edgecolor="black",
            label="Seebeck Coef. & Thermal Cond. & zT",
        ),
    ]

    fig.legend(
        handles=policy_handles + triple_handles,
        loc="lower center",
        bbox_to_anchor=(0.5, 0.02),
        ncol=3,
        frameon=True,
        fontsize=14,
    )

    fig.subplots_adjust(
        left=0.075,
        right=0.995,
        top=0.88,
        bottom=0.31,
        wspace=0.08,
    )

    PLOTDIR.mkdir(parents=True, exist_ok=True)

    outpath = PLOTDIR / f"TARGETED_UNTARGETED_IMPROVEMENT_THERMOELECTRIC_3OBJECTIVES_{MODEL}.png"
    fig.savefig(outpath, dpi=FIG_DPI, bbox_inches="tight")
    plt.show()

    print(f"[OK] Wrote improvement plot to: {outpath}")


# ============================================================
# MAIN
# ============================================================
def main() -> None:
    OUTDIR.mkdir(parents=True, exist_ok=True)
    PLOTDIR.mkdir(parents=True, exist_ok=True)

    auc_summary = build_auc_tables()

    print("\n[DEBUG] AUC summary:")
    print(auc_summary.to_string(index=False))

    improv_df = build_improvement_table(auc_summary)

    print("\n[DEBUG] Improvement table:")
    print(
        improv_df.sort_values(
            ["eval_property", "section", "target_triple", "policy"]
        ).to_string(index=False)
    )

    plot_improvement_figure(improv_df)


if __name__ == "__main__":
    main()
```

```python
#------------------------------------------------------------------------------------------XGBoost----------------------------------------------

#!/usr/bin/env python3
from __future__ import annotations

from pathlib import Path
from typing import List, Dict, Tuple, Optional
import itertools
import re

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib.patches import Patch


# ============================================================
# USER CONFIG
# ============================================================
SEEDS = range(5)
DATASET = "ThermoEl"
MODEL = "xgb"

ROOT = Path(
    r"C:\Users\rafae\OneDrive\Imágenes\Documentos\PhDToronto\OLD_FILES\PhD Toronto"
    r"\DatasetGeneration\Better_Acqsition\ThermoElectric_NEW"
)

QBC_3_ROOT = ROOT / "al" / "grow_3QBC" / DATASET
NSGA_QBC_ROOT = ROOT / "al" / "grow_EA_3Obj_ThreeTargets_QBC_ONLY" / DATASET
NSGA_QBC_DIV_ROOT = ROOT / "al" / "grow_EA_4Obj_ThreeTargets" / DATASET
RANDOM_ROOT = ROOT / "al" / "grow_random" / DATASET / "all"

OUTDIR = ROOT / "AUC_3OBJ"
PLOTDIR = ROOT / "Plots" / "Plots_AUC"

FIG_DPI = 300
ERRORBAR_CAPSIZE = 3


# ============================================================
# PROPERTIES
# ============================================================
TARGET_DISPLAY_ORDER = [
    "zT",
    "Electrical Conductivity",
    "Thermal Conductivity",
    "Seebeck Coefficient",
]

PROPERTY_KEYS = [
    "zT",
    "electric_cond",
    "thermal_cond",
    "seebeck_coef",
]

DISPLAY = {
    "zT": "zT",
    "electric_cond": "Electrical Conductivity",
    "thermal_cond": "Thermal Conductivity",
    "seebeck_coef": "Seebeck Coefficient",
}

METRIC_COLS = {
    "electric_cond": "rmse_elec_cond",
    "seebeck_coef": "rmse_seebeck",
    "thermal_cond": "rmse_k_total",
    "zT": "rmse_zt",
}


# ============================================================
# THREE-OUTCOME FOLDERS
# Adjust folder names here if your actual folders differ.
# ============================================================
TRIPLE_FOLDER_TO_KEYS = {
    "el_seeb_ThC": ("electric_cond", "seebeck_coef", "thermal_cond"),
    "el_seeb_zT": ("electric_cond", "seebeck_coef", "zT"),
    "el_zT_ThC": ("electric_cond", "thermal_cond", "zT"),
    "seeb_zT_ThC": ("seebeck_coef", "thermal_cond", "zT"),
}

TRIPLE_FOLDERS = list(TRIPLE_FOLDER_TO_KEYS.keys())


def canonical_triple_from_keys(a: str, b: str, c: str) -> Tuple[str, str, str]:
    return tuple(sorted([DISPLAY[a], DISPLAY[b], DISPLAY[c]]))


def triple_display_name(triple: Tuple[str, str, str]) -> str:
    a, b, c = tuple(sorted(triple))
    return f"{a} + {b} + {c}"


ALL_VALID_TRIPLES = [
    tuple(sorted(t))
    for t in itertools.combinations(TARGET_DISPLAY_ORDER, 3)
]


# ============================================================
# POLICIES
# ============================================================
POLICY_DISPLAY_ORDER = [
    "3QBC",
    "NSGA-II QBC",
    "NSGA-II QBC + Diversity",
]

STYLE_MAP = {
    "3QBC": {"hatch": None, "alpha": 1.0},
    "NSGA-II QBC": {"hatch": "//", "alpha": 1.0},
    "NSGA-II QBC + Diversity": {"hatch": "xx", "alpha": 1.0},
}


# ============================================================
# COLORS
# ============================================================
def canonical_metric_name(text: str) -> str:
    t = str(text).strip().lower()
    t = re.sub(r"\[[^\]]*\]", "", t)
    t = re.sub(r"\([^)]*\)", "", t)
    t = re.sub(r"\s+", " ", t).strip()

    if "random" in t:
        return "Random"
    if "zt" in t:
        return "zT"
    if "elect" in t or "elec" in t:
        return "Electrical Conductivity"
    if "thermal" in t or "k_total" in t:
        return "Thermal Conductivity"
    if "seebeck" in t:
        return "Seebeck Coefficient"

    return str(text).strip()


def canonical_triple(triple: Tuple[str, str, str]) -> Tuple[str, str, str]:
    return tuple(sorted([canonical_metric_name(x) for x in triple]))


TRIPLE_COLORS = {
    canonical_triple(("Electrical Conductivity", "Seebeck Coefficient", "Thermal Conductivity")): "#5E81AC",
    canonical_triple(("Electrical Conductivity", "Seebeck Coefficient", "zT")): "#A3BE8C",
    canonical_triple(("Electrical Conductivity", "Thermal Conductivity", "zT")): "#D08770",
    canonical_triple(("Seebeck Coefficient", "Thermal Conductivity", "zT")): "#B48EAD",
}


def triple_color(triple: Tuple[str, str, str]) -> str:
    ct = canonical_triple(triple)
    return TRIPLE_COLORS.get(ct, "#999999")


# ============================================================
# PATH HELPERS
# ============================================================
def find_matching_csv(folder: Path, required_tokens: List[str]) -> Optional[Path]:
    """
    Robust file finder:
    - Only uses CSV files.
    - Excludes trainIndices files.
    - Requires all tokens to appear in filename.
    """
    if not folder.exists():
        print(f"[WARN] Folder does not exist: {folder}")
        return None

    csvs = sorted(folder.glob("*.csv"))
    csvs = [p for p in csvs if "trainindices" not in p.name.lower()]

    matches = []
    for p in csvs:
        name = p.name.lower()
        if all(tok.lower() in name for tok in required_tokens):
            matches.append(p)

    if not matches:
        print(f"[WARN] No matching CSV in {folder}")
        print(f"       Required tokens: {required_tokens}")
        print(f"       Available CSVs: {[p.name for p in csvs]}")
        return None

    if len(matches) > 1:
        print(f"[WARN] Multiple matches in {folder}. Using first:")
        for p in matches:
            print(f"       - {p.name}")

    return matches[0]


def path_3qbc(triple_folder: str, seed: int) -> Optional[Path]:
    folder = QBC_3_ROOT / triple_folder / str(seed)
    return find_matching_csv(
        folder,
        required_tokens=[
            f"ThermoEl_{triple_folder}".lower(),
            "3qbc",
            MODEL,
            str(seed),
        ],
    )


def path_nsga_qbc(triple_folder: str, seed: int) -> Optional[Path]:
    folder = NSGA_QBC_ROOT / triple_folder / str(seed)
    return find_matching_csv(
        folder,
        required_tokens=[
            f"ThermoEl_{triple_folder}".lower(),
            "qbc_only",
            MODEL,
            str(seed),
        ],
    )


def path_nsga_qbc_div(triple_folder: str, seed: int) -> Optional[Path]:
    folder = NSGA_QBC_DIV_ROOT / triple_folder / str(seed)
    return find_matching_csv(
        folder,
        required_tokens=[
            f"ThermoEl_{triple_folder}".lower(),
            "div",
            MODEL,
            str(seed),
        ],
    )


def path_random(seed: int) -> Path:
    return RANDOM_ROOT / str(seed) / f"ThermoEl_all_random_{MODEL}{seed}.csv"


# ============================================================
# AUC HELPERS
# ============================================================
def normalize_columns(df: pd.DataFrame) -> pd.DataFrame:
    out = df.copy()
    out.columns = [str(c).strip() for c in out.columns]
    return out


def to_fraction_01(x: pd.Series) -> np.ndarray:
    arr = pd.to_numeric(x, errors="coerce").to_numpy(dtype=float)

    finite = np.isfinite(arr)
    if finite.any() and np.nanmax(arr[finite]) > 1.5:
        arr = arr / 100.0

    return arr


def auc_trapezoid(x: np.ndarray, y: np.ndarray) -> float:
    x = np.asarray(x, dtype=float)
    y = np.asarray(y, dtype=float)

    mask = np.isfinite(x) & np.isfinite(y)
    x = x[mask]
    y = y[mask]

    if len(x) < 2:
        return np.nan

    order = np.argsort(x)
    x = x[order]
    y = y[order]

    return float(np.trapezoid(y, x))


def get_xy_for_auc(df: pd.DataFrame, metric_col: str) -> Tuple[np.ndarray, np.ndarray]:
    if "train_ratio" not in df.columns:
        raise ValueError(f"'train_ratio' not found. Columns: {df.columns.tolist()}")

    if metric_col not in df.columns:
        raise ValueError(f"'{metric_col}' not found. Columns: {df.columns.tolist()}")

    tmp = df[["train_ratio", metric_col]].copy()
    tmp["x"] = to_fraction_01(tmp["train_ratio"])
    tmp["y"] = pd.to_numeric(tmp[metric_col], errors="coerce")
    tmp["_row_order"] = np.arange(len(tmp))

    tmp = tmp[np.isfinite(tmp["x"]) & np.isfinite(tmp["y"])].copy()

    rows = []

    for xval, g in tmp.groupby("x", sort=True):
        if np.isclose(xval, 1.0):
            yval = float(g.sort_values("_row_order").iloc[-1]["y"])
        else:
            yval = float(g["y"].mean())

        rows.append((float(xval), yval))

    out = pd.DataFrame(rows, columns=["x", "y"]).sort_values("x")

    return out["x"].to_numpy(), out["y"].to_numpy()


# ============================================================
# BUILD AUC TABLES
# ============================================================
def build_auc_tables() -> pd.DataFrame:
    rows = []

    print(f"[INFO] Running 3-objective ThermoEl pipeline with MODEL = {MODEL}")

    policies = [
        ("3QBC", path_3qbc),
        ("NSGA-II QBC", path_nsga_qbc),
        ("NSGA-II QBC + Diversity", path_nsga_qbc_div),
    ]

    for triple_folder in TRIPLE_FOLDERS:
        triple_keys = TRIPLE_FOLDER_TO_KEYS[triple_folder]
        triple_display = canonical_triple_from_keys(*triple_keys)
        triple_name = triple_display_name(triple_display)

        for seed in SEEDS:
            for policy, path_fn in policies:
                path = path_fn(triple_folder, seed)

                if path is None or not path.exists():
                    print(f"[WARN] Missing file for policy={policy}, triple={triple_folder}, seed={seed}")
                    continue

                print(f"[INFO] Reading {policy}: {path}")

                df = normalize_columns(pd.read_csv(path))

                for eval_key in PROPERTY_KEYS:
                    metric_col = METRIC_COLS[eval_key]

                    if metric_col not in df.columns:
                        print(f"[WARN] Missing metric column {metric_col} in {path}")
                        continue

                    x, y = get_xy_for_auc(df, metric_col)

                    rows.append({
                        "dataset": DATASET,
                        "target_triple": triple_name,
                        "target_triple_tuple": triple_display,
                        "eval_property": DISPLAY[eval_key],
                        "policy": policy,
                        "seed": seed,
                        "auc": auc_trapezoid(x, y),
                        "source_file": str(path),
                        "metric_col": metric_col,
                    })

    # Random baseline
    for seed in SEEDS:
        path = path_random(seed)

        if not path.exists():
            print(f"[WARN] Missing random file: {path}")
            continue

        print(f"[INFO] Reading Random: {path}")

        df = normalize_columns(pd.read_csv(path))

        for eval_key in PROPERTY_KEYS:
            metric_col = METRIC_COLS[eval_key]

            if metric_col not in df.columns:
                print(f"[WARN] Missing metric column {metric_col} in {path}")
                continue

            x, y = get_xy_for_auc(df, metric_col)

            rows.append({
                "dataset": DATASET,
                "target_triple": "Random",
                "target_triple_tuple": None,
                "eval_property": DISPLAY[eval_key],
                "policy": "Random",
                "seed": seed,
                "auc": auc_trapezoid(x, y),
                "source_file": str(path),
                "metric_col": metric_col,
            })

    auc_seed_df = pd.DataFrame(rows)

    if auc_seed_df.empty:
        raise RuntimeError("AUC table is empty. Check file paths and column names.")

    auc_summary = (
        auc_seed_df
        .groupby(["dataset", "target_triple", "eval_property", "policy"], as_index=False)
        .agg(
            auc_mean=("auc", "mean"),
            auc_std=("auc", "std"),
        )
    )

    auc_summary["auc_std"] = auc_summary["auc_std"].fillna(0.0)

    OUTDIR.mkdir(parents=True, exist_ok=True)

    auc_seed_path = OUTDIR / f"auc_per_seed_3outcomes_{MODEL}.csv"
    auc_summary_path = OUTDIR / f"auc_summary_3outcomes_{MODEL}.csv"

    auc_seed_df.to_csv(auc_seed_path, index=False)
    auc_summary.to_csv(auc_summary_path, index=False)

    print(f"[OK] Saved per-seed AUC table to: {auc_seed_path}")
    print(f"[OK] Saved AUC summary table to: {auc_summary_path}")

    return auc_summary


# ============================================================
# IMPROVEMENT TABLE
# ============================================================
def parse_triple_name(triple_name: str) -> Optional[Tuple[str, str, str]]:
    if triple_name == "Random":
        return None

    parts = [p.strip() for p in triple_name.split("+")]

    if len(parts) != 3:
        print(f"[WARN] Could not parse triple name: {triple_name}")
        return None

    return canonical_triple((parts[0], parts[1], parts[2]))


def build_improvement_table(auc_summary: pd.DataFrame) -> pd.DataFrame:
    rows = []

    for eval_prop in TARGET_DISPLAY_ORDER:
        sub_eval = auc_summary[auc_summary["eval_property"] == eval_prop].copy()

        if sub_eval.empty:
            print(f"[WARN] No rows found for eval property: {eval_prop}")
            continue

        rand = sub_eval[sub_eval["policy"] == "Random"]

        if rand.empty:
            print(f"[WARN] Missing random row for eval property: {eval_prop}")
            continue

        R = float(rand["auc_mean"].mean())
        sR = float(rand["auc_std"].mean())

        if not np.isfinite(R) or abs(R) < 1e-15:
            print(f"[WARN] Invalid random baseline for {eval_prop}")
            continue

        cand = sub_eval[sub_eval["policy"].isin(POLICY_DISPLAY_ORDER)].copy()

        for _, r in cand.iterrows():
            triple_name = str(r["target_triple"])
            triple = parse_triple_name(triple_name)

            if triple is None:
                continue

            P = float(r["auc_mean"])
            sP = float(r["auc_std"])

            improvement = 100.0 * (1.0 - P / R)

            dfdP = -100.0 / R
            dfdR = 100.0 * P / (R ** 2)
            s_improvement = float(np.sqrt((dfdP * sP) ** 2 + (dfdR * sR) ** 2))

            section = "Targeted" if eval_prop in triple else "Untargeted"

            rows.append({
                "eval_property": eval_prop,
                "target_triple": triple_display_name(triple),
                "target_triple_tuple": triple,
                "policy": str(r["policy"]),
                "section": section,
                "auc_mean": P,
                "auc_std": sP,
                "random_auc_mean": R,
                "random_auc_std": sR,
                "improvement_mean": improvement,
                "improvement_std": s_improvement,
            })

    out = pd.DataFrame(rows)

    if out.empty:
        raise RuntimeError("Improvement table is empty.")

    OUTDIR.mkdir(parents=True, exist_ok=True)

    out_path = OUTDIR / f"improvement_vs_random_3outcomes_{MODEL}.csv"
    out.to_csv(out_path, index=False)

    print(f"[OK] Saved improvement table to: {out_path}")

    return out


# ============================================================
# PLOTTING
# ============================================================
def format_label(label: str) -> str:
    if label == "Electrical Conductivity":
        return "Electrical\nCond."
    if label == "Thermal Conductivity":
        return "Thermal\nCond."
    if label == "Seebeck Coefficient":
        return "Seebeck\nCoef."
    return label


def compute_global_improvement_limits(improv_df: pd.DataFrame) -> Tuple[float, float]:
    vals = improv_df["improvement_mean"].to_numpy(dtype=float)
    errs = improv_df["improvement_std"].to_numpy(dtype=float)

    ymin = np.nanmin(vals - errs)
    ymax = np.nanmax(vals + errs)

    pad = 0.08 * max(ymax - ymin, 1e-12)
    ymin -= pad
    ymax += pad

    ymin = min(ymin, 0.0)
    ymax = max(ymax, 0.0)

    return ymin, ymax


def make_group_positions() -> Tuple[List[Tuple[str, float]], List[float], List[str], Dict[str, float], List[float]]:
    group_spacing = 2.6

    cases = []
    xticks = []
    xticklabels = []
    group_centers = {}
    separators = []

    for i, prop in enumerate(TARGET_DISPLAY_ORDER):
        x = i * group_spacing

        cases.append((prop, x))
        xticks.append(x)
        xticklabels.append(prop)
        group_centers[prop] = x

        if i < len(TARGET_DISPLAY_ORDER) - 1:
            separators.append(x + group_spacing / 2.0)

    return cases, xticks, xticklabels, group_centers, separators


def triple_order_for_eval(eval_prop: str, section: str) -> List[Tuple[str, str, str]]:
    triples = []

    for triple in ALL_VALID_TRIPLES:
        is_targeted = eval_prop in triple

        if section == "Targeted" and is_targeted:
            triples.append(triple)

        elif section == "Untargeted" and not is_targeted:
            triples.append(triple)

    return triples


def offsets_for_nbars(n: int, spread: float = 0.90) -> List[float]:
    if n == 1:
        return [0.0]

    return np.linspace(-spread, spread, n).tolist()


def plot_panel(ax, improv_df: pd.DataFrame, section: str) -> None:
    sub = improv_df[improv_df["section"] == section].copy()

    cases, xticks, xticklabels, _, separators = make_group_positions()

    if section == "Targeted":
        triple_count = 3
    else:
        triple_count = 1

    policy_count = len(POLICY_DISPLAY_ORDER)
    nbars = triple_count * policy_count

    width = 0.16 if section == "Targeted" else 0.28
    offs = offsets_for_nbars(nbars, spread=0.88 if section == "Targeted" else 0.38)

    for eval_prop, x in cases:
        triple_order = triple_order_for_eval(eval_prop, section)

        barspecs = []
        k = 0

        for triple_tuple in triple_order:
            for policy in POLICY_DISPLAY_ORDER:
                barspecs.append((triple_tuple, policy, offs[k]))
                k += 1

        for triple_tuple, policy, off in barspecs:
            triple_name = triple_display_name(triple_tuple)

            r = sub[
                (sub["eval_property"] == eval_prop)
                & (sub["target_triple"] == triple_name)
                & (sub["policy"] == policy)
            ]

            if r.empty:
                print(
                    f"[WARN] Missing row: section={section}, "
                    f"eval={eval_prop}, triple={triple_name}, policy={policy}"
                )
                continue

            rr = r.iloc[0]

            ax.bar(
                x + off,
                float(rr["improvement_mean"]),
                width=width,
                color=triple_color(triple_tuple),
                edgecolor="black",
                linewidth=0.8,
                hatch=STYLE_MAP[policy]["hatch"],
                alpha=STYLE_MAP[policy]["alpha"],
                yerr=float(rr["improvement_std"]),
                capsize=ERRORBAR_CAPSIZE,
                error_kw={"elinewidth": 1.0, "capthick": 1.0},
                zorder=3,
            )

    ax.axhline(0.0, color="black", linewidth=1.0, zorder=2)

    ax.set_xticks(xticks)
    ax.set_xticklabels(
        [format_label(l) for l in xticklabels],
        fontsize=18,
        rotation=0,
        ha="center",
        fontweight="bold",
    )

    ax.grid(axis="y", linestyle="--", alpha=0.28, zorder=0)

    for x_sep in separators:
        ax.axvline(x_sep, color="black", lw=1.0, alpha=0.30)

    ax.set_xlim(min(xticks) - 1.25, max(xticks) + 1.25)


def plot_improvement_figure(improv_df: pd.DataFrame) -> None:
    ylims = compute_global_improvement_limits(improv_df)

    fig, axes = plt.subplots(
        nrows=1,
        ncols=2,
        figsize=(18.0, 7.5),
        squeeze=False,
        sharey=True,
        gridspec_kw={"width_ratios": [1, 1]},
    )

    axL = axes[0, 0]
    axR = axes[0, 1]

    plot_panel(axL, improv_df, section="Targeted")
    plot_panel(axR, improv_df, section="Untargeted")

    axL.set_ylim(*ylims)
    axR.set_ylim(*ylims)

    axL.set_title("Targeted", fontsize=20, fontweight="bold")
    axR.set_title("Untargeted", fontsize=20, fontweight="bold")

    axL.set_ylabel("Improvement vs random (%)", fontsize=20, fontweight="bold")

    policy_handles = [
        Patch(facecolor="white", edgecolor="black", hatch=STYLE_MAP[p]["hatch"], label=p)
        for p in POLICY_DISPLAY_ORDER
    ]

    triple_handles = [
        Patch(
            facecolor=triple_color(("Electrical Conductivity", "Seebeck Coefficient", "Thermal Conductivity")),
            edgecolor="black",
            label="Electrical Cond. & Seebeck Coef. & Thermal Cond.",
        ),
        Patch(
            facecolor=triple_color(("Electrical Conductivity", "Seebeck Coefficient", "zT")),
            edgecolor="black",
            label="Electrical Cond. & Seebeck Coef. & zT",
        ),
        Patch(
            facecolor=triple_color(("Electrical Conductivity", "Thermal Conductivity", "zT")),
            edgecolor="black",
            label="Electrical Cond. & Thermal Cond. & zT",
        ),
        Patch(
            facecolor=triple_color(("Seebeck Coefficient", "Thermal Conductivity", "zT")),
            edgecolor="black",
            label="Seebeck Coef. & Thermal Cond. & zT",
        ),
    ]

    fig.legend(
        handles=policy_handles + triple_handles,
        loc="lower center",
        bbox_to_anchor=(0.5, 0.02),
        ncol=3,
        frameon=True,
        fontsize=14,
    )

    fig.subplots_adjust(
        left=0.075,
        right=0.995,
        top=0.88,
        bottom=0.31,
        wspace=0.08,
    )

    PLOTDIR.mkdir(parents=True, exist_ok=True)

    outpath = PLOTDIR / f"TARGETED_UNTARGETED_IMPROVEMENT_THERMOELECTRIC_3OBJECTIVES_{MODEL}.png"
    fig.savefig(outpath, dpi=FIG_DPI, bbox_inches="tight")
    plt.show()

    print(f"[OK] Wrote improvement plot to: {outpath}")


# ============================================================
# MAIN
# ============================================================
def main() -> None:
    OUTDIR.mkdir(parents=True, exist_ok=True)
    PLOTDIR.mkdir(parents=True, exist_ok=True)

    auc_summary = build_auc_tables()

    print("\n[DEBUG] AUC summary:")
    print(auc_summary.to_string(index=False))

    improv_df = build_improvement_table(auc_summary)

    print("\n[DEBUG] Improvement table:")
    print(
        improv_df.sort_values(
            ["eval_property", "section", "target_triple", "policy"]
        ).to_string(index=False)
    )

    plot_improvement_figure(improv_df)


if __name__ == "__main__":
    main()
```
