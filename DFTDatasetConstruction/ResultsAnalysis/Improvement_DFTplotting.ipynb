```python
from __future__ import annotations

import os
import re
from pathlib import Path
from typing import Optional, List, Dict, Tuple

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib.patches import Patch

# =========================================================
# USER CONFIG
# =========================================================
DATASETS = ["jarvis18", "jarvis22", "mp18", "mp21"]
seeds = range(5)

metrics_internal = {
    "rmse_bandgap": "Bandgap",
    "rmse_bulk_modulus": "Bulk Modulus",
    "rmse_eform": "Formation Energy",
}

OUT_BASE = Path(
    r"C:\Users\rafae\OneDrive\Imágenes\Documentos\PhDToronto\OLD_FILES\PhD Toronto\DatasetGeneration\Better_Acqsition"
)

AUC_SUMMARY_DIR = OUT_BASE / "ThermoElectric_NEW" / "AUC_DFT_MULTI"
PLOTS_OUTDIR_BASE = OUT_BASE / "ThermoElectric_NEW" / "Plots" / "Plots_AUC_DFT_MULTI"

# Common AUC grid.
# If your train_ratio starts at 0.01 and ends at 1.00, this is correct.
# If your curves start at 0.05, the function below automatically restricts to the available range.
COMMON_GRID = np.linspace(0.01, 1.00, 100)

DEBUG = True

# =========================================================
# IMPORTANT: BASE ROOT TEMPLATES
# =========================================================
BASE_ROOT_QBC = r"{OUT}\NewAlg_DFT2\al\QBC\{ds}\{target}"
BASE_ROOT_RANDOM = r"{OUT}\NewAlg_DFT2\al\random\{ds}\all"
BASE_ROOT_QBCDIV = r"{OUT}\NewAlg_DFT2\al\QBC_DIVERSITY\{ds}\{target}"

TARGET_TO_FOLDER = {
    "bandgap": "bandgap",
    "bulk_modulus": "bulk_modulus",
    "eform": "e_form",
}

TARGET_DISPLAY_ORDER = ["Bandgap", "Bulk Modulus", "Formation Energy"]

def qbc_filename(ds: str, target: str) -> str:
    if target == "eform":
        return f"{ds}_e_form_QBC_rf.csv"
    return f"{ds}_{target}_QBC_rf.csv"

def qbcdiv_filename(ds: str, target: str, seed: int) -> str:
    if target == "eform":
        return f"{ds}_e_form_QBCdiversity_rf{seed}.csv"
    return f"{ds}_{target}_QBCdiversity_rf{seed}.csv"

def random_filename(ds: str, seed: int) -> str:
    return f"{ds}_all_random_rf{seed}.csv"

COLMAP_QBC = {
    "rmse_bandgap": "rmse_bandgap",
    "rmse_bulk_modulus": "rmse_bulk_modulus",
    "rmse_eform": "rmse_eform",
}

COLMAP_RANDOM = {
    "rmse_bandgap": "rmse_bandgap",
    "rmse_bulk_modulus": "rmse_bulk_modulus",
    "rmse_eform": "rmse_eform",
}

COLMAP_QBCDIV = {
    "rmse_bandgap": "rmse_bandgap",
    "rmse_bulk_modulus": "rmse_bulk_modulus",
    "rmse_eform": "rmse_e_form",
}

# =========================================================
# STYLE
# =========================================================
FIG_DPI = 100
ERRORBAR_CAPSIZE = 3

TITLE_FONTSIZE = 20
AXIS_LABEL_FONTSIZE = 18
XTICK_FONTSIZE = 15
YTICK_FONTSIZE = 14
GROUP_LABEL_FONTSIZE = 16
LEGEND_FONTSIZE = 14

TARGET_COLORS = {
    "Bandgap": "#749AC4",
    "Bulk Modulus": "#94C48E",
    "Formation Energy": "#E89C9F",
}

STYLE_MAP = {
    "QBC": {"hatch": None, "alpha": 1.0},
    "QBC with diversity": {"hatch": "///", "alpha": 1.0},
}

# =========================================================
# OUTPUT TABLE SCHEMA
# =========================================================
COL_DATASET = "dataset"
COL_TARGET = "targets"
COL_PROPERTY = "RMSE"
COL_POLICY = "policy"
COL_MEAN = "auc_mean"
COL_STD = "auc_std"

# =========================================================
# UTILITIES
# =========================================================
def _normalize_colnames(df: pd.DataFrame) -> pd.DataFrame:
    df = df.copy()
    df.columns = [str(c).strip() for c in df.columns]
    return df


def _find_x_column(df: pd.DataFrame) -> str:
    """
    Strictly identify the x-axis column.

    Important:
    We do NOT fall back to the first numeric column.
    That behavior can silently use train_size, an index column, or another variable.
    """
    cols = list(df.columns)
    lower = {c: c.strip().lower() for c in cols}

    preferred = [
        "train_ratio",
        "train_pct",
        "pct",
        "pcts",
        "percent",
        "percentage",
        "data_pct",
        "pool_pct",
        "train_percent",
        "data_percent",
        "fraction",
        "frac",
        "acq_pct",
        "acq_percent",
        "x",
    ]

    for p in preferred:
        for c in cols:
            if lower[c] == p:
                return c

    raise ValueError(
        "Could not safely identify the x-axis column. "
        f"Available columns are: {cols}"
    )


def _to_fraction01(x: np.ndarray) -> np.ndarray:
    """
    Convert x-axis to fraction scale [0, 1].

    If x is already train_ratio, it remains unchanged.
    If x is percent, e.g. 1--100, it is divided by 100.
    """
    x = np.asarray(x, dtype=float)

    finite = x[np.isfinite(x)]
    if finite.size == 0:
        return x

    if np.nanmax(finite) > 1.5:
        return x / 100.0

    return x


def _clean_xy(x: np.ndarray, y: np.ndarray) -> Tuple[np.ndarray, np.ndarray]:
    x = np.asarray(x, dtype=float)
    y = np.asarray(y, dtype=float)

    mask = np.isfinite(x) & np.isfinite(y)
    x = x[mask]
    y = y[mask]

    if x.size < 2:
        return x, y

    order = np.argsort(x)
    x = x[order]
    y = y[order]

    # Handle duplicated x values by averaging y at the same x.
    df_tmp = pd.DataFrame({"x": x, "y": y})
    df_tmp = df_tmp.groupby("x", as_index=False)["y"].mean()

    return df_tmp["x"].to_numpy(), df_tmp["y"].to_numpy()


def auc_trapz_raw(x: np.ndarray, y: np.ndarray) -> float:
    """
    Raw trapezoidal AUC on the curve's own x-grid.
    Useful for diagnostics.
    """
    x, y = _clean_xy(x, y)

    if x.size < 2:
        return np.nan

    return float(np.trapz(y, x))


def auc_trapz_common_grid(x: np.ndarray, y: np.ndarray) -> float:
    """
    Fair AUC comparison using a common x-grid.

    This avoids comparing policies on different x grids.
    """
    x, y = _clean_xy(x, y)

    if x.size < 2:
        return np.nan

    # Restrict common grid to the available curve range.
    grid = COMMON_GRID[(COMMON_GRID >= x.min()) & (COMMON_GRID <= x.max())]

    if grid.size < 2:
        return np.nan

    y_interp = np.interp(grid, x, y)

    return float(np.trapz(y_interp, grid))


def infer_policy_from_runlabel(run_label: str) -> str:
    rl = run_label.strip().lower()

    if "random" in rl:
        return "Random"

    if "divers" in rl:
        return "QBC with diversity"

    if rl.startswith("qbc"):
        return "QBC"

    return "Other"


def _target_display_from_runlabel(run_label: str) -> str:
    rl = str(run_label).strip()

    if rl.lower().startswith("random"):
        return "Random"

    m = re.search(r"target\s*:\s*(.*)$", rl, flags=re.IGNORECASE)
    if m:
        return m.group(1).strip()

    return rl


def _resolve_metric_column(df: pd.DataFrame, expected_col: str) -> str:
    """
    Resolve metric column robustly using exact match first,
    then case-insensitive match.
    """
    if expected_col in df.columns:
        return expected_col

    expected_lower = expected_col.strip().lower()
    for c in df.columns:
        if c.strip().lower() == expected_lower:
            return c

    raise ValueError(
        f"Missing metric column '{expected_col}'. "
        f"Available columns are: {list(df.columns)}"
    )


# =========================================================
# BUILD RUNS PER DATASET
# =========================================================
def make_runs_for_dataset(ds: str) -> Dict[str, dict]:
    out = str(OUT_BASE)

    return {
        "QBC target: Bandgap": {
            "base_root": BASE_ROOT_QBC.format(
                OUT=out,
                ds=ds,
                target=TARGET_TO_FOLDER["bandgap"],
            ),
            "path_fn": lambda s, _ds=ds: os.path.join(
                str(s),
                qbc_filename(_ds, "bandgap"),
            ),
            "colmap": COLMAP_QBC,
        },

        "QBC target: Bulk Modulus": {
            "base_root": BASE_ROOT_QBC.format(
                OUT=out,
                ds=ds,
                target=TARGET_TO_FOLDER["bulk_modulus"],
            ),
            "path_fn": lambda s, _ds=ds: os.path.join(
                str(s),
                qbc_filename(_ds, "bulk_modulus"),
            ),
            "colmap": COLMAP_QBC,
        },

        "QBC target: Formation Energy": {
            "base_root": BASE_ROOT_QBC.format(
                OUT=out,
                ds=ds,
                target=TARGET_TO_FOLDER["eform"],
            ),
            "path_fn": lambda s, _ds=ds: os.path.join(
                str(s),
                qbc_filename(_ds, "eform"),
            ),
            "colmap": COLMAP_QBC,
        },

        "Random sampling": {
            "base_root": BASE_ROOT_RANDOM.format(
                OUT=out,
                ds=ds,
            ),
            "path_fn": lambda s, _ds=ds: os.path.join(
                str(s),
                random_filename(_ds, s),
            ),
            "colmap": COLMAP_RANDOM,
        },

        "QBC+Diversity target: Bandgap": {
            "base_root": BASE_ROOT_QBCDIV.format(
                OUT=out,
                ds=ds,
                target=TARGET_TO_FOLDER["bandgap"],
            ),
            "path_fn": lambda s, _ds=ds: os.path.join(
                str(s),
                qbcdiv_filename(_ds, "bandgap", s),
            ),
            "colmap": COLMAP_QBCDIV,
        },

        "QBC+Diversity target: Bulk Modulus": {
            "base_root": BASE_ROOT_QBCDIV.format(
                OUT=out,
                ds=ds,
                target=TARGET_TO_FOLDER["bulk_modulus"],
            ),
            "path_fn": lambda s, _ds=ds: os.path.join(
                str(s),
                qbcdiv_filename(_ds, "bulk_modulus", s),
            ),
            "colmap": COLMAP_QBCDIV,
        },

        "QBC+Diversity target: Formation Energy": {
            "base_root": BASE_ROOT_QBCDIV.format(
                OUT=out,
                ds=ds,
                target=TARGET_TO_FOLDER["eform"],
            ),
            "path_fn": lambda s, _ds=ds: os.path.join(
                str(s),
                qbcdiv_filename(_ds, "eform", s),
            ),
            "colmap": COLMAP_QBCDIV,
        },
    }


# =========================================================
# STEP 1: BUILD LONG AUC SUMMARY FOR ONE DATASET
# =========================================================
def build_auc_summary_long_for_dataset(
    ds: str,
    runs: Dict[str, dict],
) -> Tuple[pd.DataFrame, pd.DataFrame]:

    seed_rows: List[dict] = []

    for run_label, cfg in runs.items():
        base_root = Path(cfg["base_root"])
        path_fn = cfg["path_fn"]
        colmap = cfg["colmap"]
        policy = infer_policy_from_runlabel(run_label)

        for s in seeds:
            csv_path = base_root / path_fn(s)

            if not csv_path.exists():
                print(f"[WARN] Missing file: {csv_path}")
                continue

            df = pd.read_csv(csv_path)
            df = _normalize_colnames(df)

            try:
                xcol = _find_x_column(df)
            except ValueError as e:
                print(f"[WARN] {csv_path}")
                print(f"       {e}")
                continue

            x = _to_fraction01(pd.to_numeric(df[xcol], errors="coerce").to_numpy())

            for mkey, display in metrics_internal.items():
                expected_col = colmap.get(mkey, mkey)

                try:
                    actual_col = _resolve_metric_column(df, expected_col)
                except ValueError as e:
                    print(f"[WARN] {csv_path}")
                    print(f"       {e}")
                    continue

                y = pd.to_numeric(df[actual_col], errors="coerce").to_numpy()

                auc_common = auc_trapz_common_grid(x, y)
                auc_raw = auc_trapz_raw(x, y)

                if DEBUG:
                    print(
                        "[DEBUG]",
                        f"dataset={ds}",
                        f"run='{run_label}'",
                        f"policy='{policy}'",
                        f"seed={s}",
                        f"eval='{display}'",
                        f"xcol='{xcol}'",
                        f"ycol='{actual_col}'",
                        f"x_min={np.nanmin(x):.4f}",
                        f"x_max={np.nanmax(x):.4f}",
                        f"n={np.isfinite(x).sum()}",
                        f"auc_common={auc_common:.6f}",
                        f"auc_raw={auc_raw:.6f}",
                        sep=" | ",
                    )

                seed_rows.append(
                    {
                        COL_DATASET: ds,
                        COL_TARGET: run_label,
                        COL_PROPERTY: display,
                        COL_POLICY: policy,
                        "seed": int(s),
                        "x_column": xcol,
                        "y_column": actual_col,
                        "x_min": float(np.nanmin(x)),
                        "x_max": float(np.nanmax(x)),
                        "n_points": int(np.isfinite(x).sum()),
                        "auc": auc_common,
                        "auc_raw": auc_raw,
                    }
                )

    seed_long = pd.DataFrame(seed_rows)

    if seed_long.empty:
        raise RuntimeError(f"No AUC rows were generated for dataset '{ds}'.")

    agg = (
        seed_long
        .groupby(
            [COL_DATASET, COL_TARGET, COL_PROPERTY, COL_POLICY],
            as_index=False,
        )["auc"]
        .agg(auc_mean="mean", auc_std="std")
    )

    agg[COL_STD] = agg[COL_STD].fillna(0.0)

    return agg, seed_long


# =========================================================
# IMPROVEMENT CALCULATION
# =========================================================
def build_improvement_table(df_all: pd.DataFrame) -> pd.DataFrame:
    df = df_all.copy()
    df["target_display"] = df[COL_TARGET].map(_target_display_from_runlabel)

    rows = []

    for ds in DATASETS:
        for eval_prop in TARGET_DISPLAY_ORDER:
            rand = df[
                (df[COL_DATASET] == ds)
                & (df[COL_PROPERTY] == eval_prop)
                & (df[COL_POLICY] == "Random")
            ]

            if rand.empty:
                print(f"[WARN] Missing random baseline for {ds}, {eval_prop}")
                continue

            # There should be only one random row per dataset/eval_property.
            # If more exist, use the first and warn.
            if len(rand) > 1:
                print(f"[WARN] Multiple random rows for {ds}, {eval_prop}. Using first row.")

            R = float(rand.iloc[0][COL_MEAN])
            sR = float(rand.iloc[0][COL_STD])

            if not np.isfinite(R) or abs(R) < 1e-15:
                print(f"[WARN] Invalid random AUC for {ds}, {eval_prop}: {R}")
                continue

            cand = df[
                (df[COL_DATASET] == ds)
                & (df[COL_PROPERTY] == eval_prop)
                & (df[COL_POLICY].isin(["QBC", "QBC with diversity"]))
            ].copy()

            for _, r in cand.iterrows():
                P = float(r[COL_MEAN])
                sP = float(r[COL_STD])
                target_display = str(r["target_display"])

                improvement = 100.0 * (1.0 - P / R)

                # Error propagation for f = 100 * (1 - P/R)
                dfdP = -100.0 / R
                dfdR = 100.0 * P / (R ** 2)
                s_improvement = float(
                    np.sqrt((dfdP * sP) ** 2 + (dfdR * sR) ** 2)
                )

                section = "Targeted" if target_display == eval_prop else "Untargeted"

                rows.append(
                    {
                        "dataset": ds,
                        "eval_property": eval_prop,
                        "acq_target": target_display,
                        "policy": str(r[COL_POLICY]),
                        "section": section,
                        "auc_policy": P,
                        "auc_policy_std": sP,
                        "auc_random": R,
                        "auc_random_std": sR,
                        "improvement_mean": improvement,
                        "improvement_std": s_improvement,
                    }
                )

    out = pd.DataFrame(rows)

    if out.empty:
        raise RuntimeError("Improvement table is empty. Check random/policy matching.")

    return out


def compute_global_improvement_limits(improv_df: pd.DataFrame) -> Tuple[float, float]:
    vals = improv_df["improvement_mean"].to_numpy(dtype=float)
    errs = improv_df["improvement_std"].to_numpy(dtype=float)

    ymin = np.min(vals - errs)
    ymax = np.max(vals + errs)

    pad = 0.08 * max(ymax - ymin, 1e-12)
    ymin -= pad
    ymax += pad

    ymin = min(ymin, 0.0)
    ymax = max(ymax, 0.0)

    return ymin, ymax


# =========================================================
# PLOTTING HELPERS
# =========================================================
def make_grouped_case_positions():
    """
    Group by evaluated RMSE property.
    Inside each group, show datasets.
    """
    n_datasets = len(DATASETS)

    dataset_spacing = 1.7
    group_spacing = 1.45

    cases = []
    xticks = []
    xticklabels = []
    group_centers = {}
    separators = []

    for p_idx, prop in enumerate(TARGET_DISPLAY_ORDER):
        start = p_idx * ((n_datasets - 1) * dataset_spacing + group_spacing + 1.0)

        xs_this_group = []

        for d_idx, ds in enumerate(DATASETS):
            x = start + d_idx * dataset_spacing

            cases.append((prop, ds, x))
            xticks.append(x)
            xticklabels.append(ds.upper())

            xs_this_group.append(x)

        group_centers[prop] = float(np.mean(xs_this_group))

        if p_idx < len(TARGET_DISPLAY_ORDER) - 1:
            separators.append(xs_this_group[-1] + 0.5 * dataset_spacing + 0.5 * group_spacing)

    return cases, xticks, xticklabels, group_centers, separators


def add_bottom_group_labels(ax, group_centers):
    ymin, ymax = ax.get_ylim()
    y_text = ymin - 0.1 * (ymax - ymin)

    display_map = {
        "Bandgap": "Bandgap",
        "Bulk Modulus": "Bulk Modulus",
        "Formation Energy": "Formation\nEnergy",
    }

    x_shift = {
        "Bandgap": 0.0,
        "Bulk Modulus": 0.5,
        "Formation Energy": 0.0,
    }

    for prop in TARGET_DISPLAY_ORDER:
        ax.text(
            group_centers[prop] + x_shift[prop],
            y_text,
            display_map[prop],
            ha="center",
            va="top",
            fontsize=GROUP_LABEL_FONTSIZE,
            fontweight="bold",
            clip_on=False,
        )


def add_bottom_group_labels_inside(ax, group_centers):
    ymin, ymax = ax.get_ylim()
    y_text = ymin + 0.08 * (ymax - ymin)

    display_map = {
        "Bandgap": "Bandgap",
        "Bulk Modulus": "Bulk Modulus",
        "Formation Energy": "Formation\nEnergy",
    }

    x_shift = {
        "Bandgap": 0.0,
        "Bulk Modulus": -0.5,
        "Formation Energy": 0.0,
    }

    for prop in TARGET_DISPLAY_ORDER:
        ax.text(
            group_centers[prop] + x_shift[prop],
            y_text,
            display_map[prop],
            ha="center",
            va="bottom",
            fontsize=GROUP_LABEL_FONTSIZE,
            fontweight="bold",
        )


def plot_targeted_panel_single_row(ax, improv_df: pd.DataFrame) -> None:
    sub = improv_df[improv_df["section"] == "Targeted"].copy()

    cases, xticks, xticklabels, group_centers, separators = make_grouped_case_positions()

    width = 0.62
    offsets = {
        "QBC": -0.34,
        "QBC with diversity": +0.34,
    }

    for policy in ["QBC", "QBC with diversity"]:
        for eval_prop in TARGET_DISPLAY_ORDER:
            xs, ys, es = [], [], []
            acq_target_for_color = eval_prop

            for prop, ds, x in cases:
                if prop != eval_prop:
                    continue

                r = sub[
                    (sub["dataset"] == ds)
                    & (sub["eval_property"] == prop)
                    & (sub["policy"] == policy)
                ]

                if r.empty:
                    continue

                rr = r.iloc[0]
                acq_target_for_color = str(rr["acq_target"])

                xs.append(x + offsets[policy])
                ys.append(float(rr["improvement_mean"]))
                es.append(float(rr["improvement_std"]))

            if xs:
                ax.bar(
                    xs,
                    ys,
                    width=width,
                    color=TARGET_COLORS[acq_target_for_color],
                    edgecolor="black",
                    linewidth=0.8,
                    hatch=STYLE_MAP[policy]["hatch"],
                    yerr=es,
                    capsize=ERRORBAR_CAPSIZE,
                    error_kw={"elinewidth": 1.0, "capthick": 1.0},
                    zorder=3,
                )

    ax.axhline(0.0, color="black", linewidth=1.0, zorder=2)

    ax.set_xticks(xticks)
    ax.set_xticklabels(
        xticklabels,
        fontsize=XTICK_FONTSIZE,
        rotation=45,
        ha="right",
        fontweight="bold",
    )

    ax.tick_params(axis="y", labelsize=YTICK_FONTSIZE)
    ax.set_xlim(min(xticks) - 1.0, max(xticks) + 1.0)
    ax.grid(axis="y", linestyle="--", alpha=0.25, zorder=0)

    for x_sep in separators:
        ax.axvline(x_sep, color="black", lw=1.0, alpha=0.25)

    add_bottom_group_labels(ax, group_centers)


def plot_untargeted_panel_single_row(ax, improv_df: pd.DataFrame) -> None:
    sub = improv_df[improv_df["section"] == "Untargeted"].copy()

    cases, xticks, xticklabels, group_centers, separators = make_grouped_case_positions()

    width = 0.34
    offsets = [-0.57, -0.19, 0.19, 0.57]

    for eval_prop in TARGET_DISPLAY_ORDER:
        for prop, ds, x in cases:
            if prop != eval_prop:
                continue

            others = [t for t in TARGET_DISPLAY_ORDER if t != eval_prop]

            barspecs = [
                (others[0], "QBC", offsets[0]),
                (others[0], "QBC with diversity", offsets[1]),
                (others[1], "QBC", offsets[2]),
                (others[1], "QBC with diversity", offsets[3]),
            ]

            for acq_target, policy, off in barspecs:
                r = sub[
                    (sub["dataset"] == ds)
                    & (sub["eval_property"] == eval_prop)
                    & (sub["acq_target"] == acq_target)
                    & (sub["policy"] == policy)
                ]

                if r.empty:
                    continue

                rr = r.iloc[0]

                ax.bar(
                    x + off,
                    float(rr["improvement_mean"]),
                    width=width,
                    color=TARGET_COLORS[acq_target],
                    edgecolor="black",
                    linewidth=0.8,
                    hatch=STYLE_MAP[policy]["hatch"],
                    yerr=float(rr["improvement_std"]),
                    capsize=ERRORBAR_CAPSIZE,
                    error_kw={"elinewidth": 1.0, "capthick": 1.0},
                    zorder=3,
                )

    ax.axhline(0.0, color="black", linewidth=1.0, zorder=2)

    ax.set_xticks(xticks)
    ax.set_xticklabels(
        xticklabels,
        fontsize=XTICK_FONTSIZE,
        rotation=45,
        ha="right",
        fontweight="bold",
    )

    ax.tick_params(axis="y", labelsize=YTICK_FONTSIZE)
    ax.set_xlim(min(xticks) - 1.0, max(xticks) + 1.0)
    ax.grid(axis="y", linestyle="--", alpha=0.25, zorder=0)

    for x_sep in separators:
        ax.axvline(x_sep, color="black", lw=1.0, alpha=0.25)

    add_bottom_group_labels_inside(ax, group_centers)


# =========================================================
# FULL FIGURE
# =========================================================
def plot_improvement_figure(improv_df: pd.DataFrame) -> None:
    ylims = compute_global_improvement_limits(improv_df)

    fig, axes = plt.subplots(
        nrows=1,
        ncols=2,
        figsize=(15.5, 6.4),
        squeeze=False,
        sharey=True,
        gridspec_kw={"width_ratios": [1, 1.55]},
    )

    axL = axes[0, 0]
    axR = axes[0, 1]

    plot_targeted_panel_single_row(axL, improv_df)
    plot_untargeted_panel_single_row(axR, improv_df)

    axL.set_ylim(*ylims)
    axR.set_ylim(*ylims)

    axL.set_title("Targeted", fontsize=TITLE_FONTSIZE, fontweight="bold")
    axR.set_title("Untargeted", fontsize=TITLE_FONTSIZE, fontweight="bold")

    axL.set_ylabel(
        "Improvement vs random (%)",
        fontsize=AXIS_LABEL_FONTSIZE,
        fontweight="bold",
    )

    policy_handles = [
        Patch(facecolor="white", edgecolor="black", hatch=None, label="QBC"),
        Patch(facecolor="white", edgecolor="black", hatch="///", label="QBC with diversity"),
    ]

    target_handles = [
        Patch(facecolor=TARGET_COLORS["Bandgap"], edgecolor="black", label="Target: Bandgap"),
        Patch(facecolor=TARGET_COLORS["Bulk Modulus"], edgecolor="black", label="Target: Bulk Modulus"),
        Patch(facecolor=TARGET_COLORS["Formation Energy"], edgecolor="black", label="Target: Formation\nEnergy"),
    ]

    fig.legend(
        handles=policy_handles + target_handles,
        loc="lower center",
        bbox_to_anchor=(0.5, 0.03),
        ncol=5,
        frameon=True,
        fontsize=LEGEND_FONTSIZE,
        title_fontsize=12,
    )

    fig.subplots_adjust(
        left=0.07,
        right=0.995,
        top=0.88,
        bottom=0.31,
        wspace=0.035,
    )

    PLOTS_OUTDIR_BASE.mkdir(parents=True, exist_ok=True)

    outpath = (
        PLOTS_OUTDIR_BASE
        / "TARGETED_UNTARGETED_IMPROVEMENT_SINGLE_ROW_GROUPED_BY_PROPERTY_TARGET_COLORED.png"
    )

    fig.savefig(outpath, dpi=FIG_DPI, bbox_inches="tight")
    plt.show()

    print(f"[OK] Wrote improvement plot to: {outpath}")


# =========================================================
# OPTIONAL DIAGNOSTIC PLOT FOR ONE CASE
# =========================================================
def diagnostic_plot_one_case(
    ds: str = "jarvis18",
    eval_metric_key: str = "rmse_bandgap",
    acq_target: str = "bandgap",
) -> None:
    """
    Plots Random, QBC, and QBC with diversity for one dataset/property/target.
    Useful for checking whether the AUC ranking matches the visible curves.

    Example:
        diagnostic_plot_one_case(
            ds="jarvis18",
            eval_metric_key="rmse_bandgap",
            acq_target="bandgap",
        )
    """
    runs = make_runs_for_dataset(ds)

    run_labels = [
        f"QBC target: {_display_from_target_key(acq_target)}",
        "Random sampling",
        f"QBC+Diversity target: {_display_from_target_key(acq_target)}",
    ]

    fig, ax = plt.subplots(figsize=(7, 5))

    colors = {
        "QBC": "tab:blue",
        "Random": "tab:gray",
        "QBC with diversity": "tab:orange",
    }

    for run_label in run_labels:
        if run_label not in runs:
            print(f"[WARN] Diagnostic run not found: {run_label}")
            continue

        cfg = runs[run_label]
        base_root = Path(cfg["base_root"])
        path_fn = cfg["path_fn"]
        colmap = cfg["colmap"]
        policy = infer_policy_from_runlabel(run_label)

        curves = []

        for s in seeds:
            csv_path = base_root / path_fn(s)

            if not csv_path.exists():
                print(f"[WARN] Missing diagnostic file: {csv_path}")
                continue

            df = pd.read_csv(csv_path)
            df = _normalize_colnames(df)

            xcol = _find_x_column(df)
            x = _to_fraction01(pd.to_numeric(df[xcol], errors="coerce").to_numpy())

            expected_col = colmap.get(eval_metric_key, eval_metric_key)
            actual_col = _resolve_metric_column(df, expected_col)
            y = pd.to_numeric(df[actual_col], errors="coerce").to_numpy()

            x_clean, y_clean = _clean_xy(x, y)

            grid = COMMON_GRID[
                (COMMON_GRID >= x_clean.min())
                & (COMMON_GRID <= x_clean.max())
            ]

            y_interp = np.interp(grid, x_clean, y_clean)
            curves.append(y_interp)

        if not curves:
            continue

        curves = np.asarray(curves)
        y_mean = np.nanmean(curves, axis=0)
        y_std = np.nanstd(curves, axis=0)

        auc_value = np.trapz(y_mean, grid)

        ax.plot(
            grid * 100,
            y_mean,
            label=f"{policy}, AUC={auc_value:.4f}",
            color=colors.get(policy, None),
            linewidth=2,
        )

        ax.fill_between(
            grid * 100,
            y_mean - y_std,
            y_mean + y_std,
            color=colors.get(policy, None),
            alpha=0.18,
            linewidth=0,
        )

    ax.set_xlabel("Training data (%)", fontsize=14)
    ax.set_ylabel(metrics_internal[eval_metric_key], fontsize=14)
    ax.set_title(f"{ds.upper()} | {metrics_internal[eval_metric_key]}", fontsize=15)
    ax.grid(alpha=0.3)
    ax.legend()
    plt.show()


def _display_from_target_key(target_key: str) -> str:
    if target_key == "bandgap":
        return "Bandgap"
    if target_key == "bulk_modulus":
        return "Bulk Modulus"
    if target_key == "eform":
        return "Formation Energy"
    raise ValueError(f"Unknown target key: {target_key}")


# =========================================================
# MAIN
# =========================================================
def main() -> None:
    AUC_SUMMARY_DIR.mkdir(parents=True, exist_ok=True)
    PLOTS_OUTDIR_BASE.mkdir(parents=True, exist_ok=True)

    all_summaries = []
    all_seed_level = []

    for ds in DATASETS:
        runs_ds = make_runs_for_dataset(ds)

        auc_df_ds, seed_level_ds = build_auc_summary_long_for_dataset(ds, runs_ds)

        out_csv = AUC_SUMMARY_DIR / f"auc_summary_series_long_dft_{ds}.csv"
        auc_df_ds.to_csv(out_csv, index=False)
        print(f"[OK] Wrote AUC summary for {ds} to: {out_csv}")

        seed_csv = AUC_SUMMARY_DIR / f"auc_seed_level_dft_{ds}.csv"
        seed_level_ds.to_csv(seed_csv, index=False)
        print(f"[OK] Wrote seed-level AUC diagnostics for {ds} to: {seed_csv}")

        all_summaries.append(auc_df_ds)
        all_seed_level.append(seed_level_ds)

    df_all = pd.concat(all_summaries, ignore_index=True)
    seed_all = pd.concat(all_seed_level, ignore_index=True)

    all_auc_csv = AUC_SUMMARY_DIR / "auc_summary_series_long_dft_ALL.csv"
    df_all.to_csv(all_auc_csv, index=False)
    print(f"[OK] Wrote combined AUC summary to: {all_auc_csv}")

    seed_all_csv = AUC_SUMMARY_DIR / "auc_seed_level_dft_ALL.csv"
    seed_all.to_csv(seed_all_csv, index=False)
    print(f"[OK] Wrote combined seed-level diagnostics to: {seed_all_csv}")

    improv_df = build_improvement_table(df_all)

    improv_csv = AUC_SUMMARY_DIR / "improvement_vs_random_targeted_untargeted.csv"
    improv_df.to_csv(improv_csv, index=False)
    print(f"[OK] Wrote improvement summary to: {improv_csv}")

    plot_improvement_figure(improv_df)


if __name__ == "__main__":
    main()

    # Optional diagnostic:
    # Uncomment this to visually check a case .
    #
    diagnostic_plot_one_case(
         ds="jarvis22",
         eval_metric_key="rmse_bulk_modulus",
         acq_target="bulk_modulus",
     )
```

```python
# ----------------------------- XGBoost AUC CODE -----------------------------
from __future__ import annotations

import os
import re
from pathlib import Path
from typing import List, Dict, Tuple

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib.patches import Patch

# =========================================================
# USER CONFIG
# =========================================================
DATASETS = ["jarvis18", "jarvis22", "mp18", "mp21"]
seeds = range(5)

metrics_internal = {
    "rmse_bandgap": "Bandgap",
    "rmse_bulk_modulus": "Bulk Modulus",
    "rmse_eform": "Formation Energy",
}

OUT_BASE = Path(
    r"C:\Users\rafae\OneDrive\Imágenes\Documentos\PhDToronto\OLD_FILES\PhD Toronto\DatasetGeneration\Better_Acqsition"
)

AUC_SUMMARY_DIR = OUT_BASE / "ThermoElectric_NEW" / "AUC_DFT_MULTI_XGBoost"
PLOTS_OUTDIR_BASE = OUT_BASE / "ThermoElectric_NEW" / "Plots" / "Plots_AUC_DFT_MULTI_XGBoost"

DEBUG = True

# Common AUC grid.
# This assumes train_ratio is in [0, 1].
COMMON_GRID = np.linspace(0.01, 1.00, 100)

# =========================================================
# IMPORTANT: BASE ROOT TEMPLATES
# =========================================================
BASE_ROOT_QBC = r"{OUT}\NewAlg_DFT2\al\QBC\{ds}\{target}"
BASE_ROOT_RANDOM = r"{OUT}\NewAlg_DFT2\al\random\{ds}\all"
BASE_ROOT_QBCDIV = r"{OUT}\NewAlg_DFT2\al\QBC_DIVERSITY\{ds}\{target}"

TARGET_TO_FOLDER = {
    "bandgap": "bandgap",
    "bulk_modulus": "bulk_modulus",
    "eform": "e_form",
}

TARGET_DISPLAY_ORDER = ["Bandgap", "Bulk Modulus", "Formation Energy"]

# =========================================================
# XGBoost filename patterns
# =========================================================
def qbc_filename(ds: str, target: str) -> str:
    if target == "eform":
        return f"{ds}_e_form_QBC_xgb.csv"
    return f"{ds}_{target}_QBC_xgb.csv"


def qbcdiv_filename(ds: str, target: str, seed: int) -> str:
    if target == "eform":
        return f"{ds}_e_form_QBCdiversity_xgb{seed}.csv"
    return f"{ds}_{target}_QBCdiversity_xgb{seed}.csv"


def random_filename(ds: str, seed: int) -> str:
    return f"{ds}_all_random_xgb{seed}.csv"


COLMAP_QBC = {
    "rmse_bandgap": "rmse_bandgap",
    "rmse_bulk_modulus": "rmse_bulk_modulus",
    "rmse_eform": "rmse_eform",
}

COLMAP_RANDOM = {
    "rmse_bandgap": "rmse_bandgap",
    "rmse_bulk_modulus": "rmse_bulk_modulus",
    "rmse_eform": "rmse_eform",
}

COLMAP_QBCDIV = {
    "rmse_bandgap": "rmse_bandgap",
    "rmse_bulk_modulus": "rmse_bulk_modulus",
    "rmse_eform": "rmse_e_form",
}

# =========================================================
# STYLE
# =========================================================
FIG_DPI = 100
ERRORBAR_CAPSIZE = 3

TITLE_FONTSIZE = 20
AXIS_LABEL_FONTSIZE = 18
XTICK_FONTSIZE = 15
YTICK_FONTSIZE = 14
GROUP_LABEL_FONTSIZE = 16
LEGEND_FONTSIZE = 14

TARGET_COLORS = {
    "Bandgap": "#749AC4",
    "Bulk Modulus": "#94C48E",
    "Formation Energy": "#E89C9F",
}

STYLE_MAP = {
    "QBC": {"hatch": None, "alpha": 1.0},
    "QBC with diversity": {"hatch": "///", "alpha": 1.0},
}

# =========================================================
# OUTPUT TABLE SCHEMA
# =========================================================
COL_DATASET = "dataset"
COL_TARGET = "targets"
COL_PROPERTY = "RMSE"
COL_POLICY = "policy"
COL_MEAN = "auc_mean"
COL_STD = "auc_std"

# =========================================================
# UTILITIES
# =========================================================
def _normalize_colnames(df: pd.DataFrame) -> pd.DataFrame:
    df = df.copy()
    df.columns = [str(c).strip() for c in df.columns]
    return df


def _find_x_column(df: pd.DataFrame) -> str:
    """
    Strict x-axis detection.

    Important:
    This version prefers train_ratio and does not guess from arbitrary
    numeric columns. If no valid x-axis column is found, it raises an error.
    """
    cols = list(df.columns)
    lower = {c: c.strip().lower() for c in cols}

    preferred = [
        "train_ratio",
        "train_pct",
        "pct",
        "pcts",
        "percent",
        "percentage",
        "data_pct",
        "pool_pct",
        "train_percent",
        "data_percent",
        "fraction",
        "frac",
        "acq_pct",
        "acq_percent",
        "x",
    ]

    for p in preferred:
        for c in cols:
            if lower[c] == p:
                return c

    raise ValueError(
        "Could not safely identify the x-axis column. "
        f"Available columns are: {cols}"
    )


def _to_fraction01(x: np.ndarray) -> np.ndarray:
    """
    Convert x-axis to fraction scale [0, 1].

    If x is already train_ratio, it remains unchanged.
    If x is percent, e.g. 1--100, it is divided by 100.
    """
    x = np.asarray(x, dtype=float)

    finite = x[np.isfinite(x)]
    if finite.size == 0:
        return x

    if np.nanmax(finite) > 1.5:
        return x / 100.0

    return x


def _clean_xy(x: np.ndarray, y: np.ndarray) -> Tuple[np.ndarray, np.ndarray]:
    """
    Remove NaNs, sort by x, and average duplicate x values.
    """
    x = np.asarray(x, dtype=float)
    y = np.asarray(y, dtype=float)

    mask = np.isfinite(x) & np.isfinite(y)
    x = x[mask]
    y = y[mask]

    if x.size < 2:
        return x, y

    order = np.argsort(x)
    x = x[order]
    y = y[order]

    tmp = pd.DataFrame({"x": x, "y": y})
    tmp = tmp.groupby("x", as_index=False)["y"].mean()

    return tmp["x"].to_numpy(), tmp["y"].to_numpy()


def auc_trapz_raw(x: np.ndarray, y: np.ndarray) -> float:
    """
    Raw trapezoidal AUC on each curve's native x-grid.
    Stored only for diagnostics.
    """
    x, y = _clean_xy(x, y)

    if x.size < 2:
        return np.nan

    return float(np.trapz(y, x))


def auc_trapz_common_grid(x: np.ndarray, y: np.ndarray) -> float:
    """
    AUC computed on a common interpolation grid.

    This makes the comparison fair even if files have tiny differences
    in x values or repeated points.
    """
    x, y = _clean_xy(x, y)

    if x.size < 2:
        return np.nan

    grid = COMMON_GRID[
        (COMMON_GRID >= x.min()) &
        (COMMON_GRID <= x.max())
    ]

    if grid.size < 2:
        return np.nan

    y_interp = np.interp(grid, x, y)

    return float(np.trapz(y_interp, grid))


def _resolve_metric_column(df: pd.DataFrame, expected_col: str) -> str:
    """
    Exact match first. Then case-insensitive match.
    """
    if expected_col in df.columns:
        return expected_col

    expected_lower = expected_col.strip().lower()

    for c in df.columns:
        if c.strip().lower() == expected_lower:
            return c

    raise ValueError(
        f"Missing metric column '{expected_col}'. "
        f"Available columns are: {list(df.columns)}"
    )


def infer_policy_from_runlabel(run_label: str) -> str:
    rl = run_label.strip().lower()

    if "random" in rl:
        return "Random"

    if "divers" in rl:
        return "QBC with diversity"

    if rl.startswith("qbc"):
        return "QBC"

    return "Other"


def _target_display_from_runlabel(run_label: str) -> str:
    rl = str(run_label).strip()

    if rl.lower().startswith("random"):
        return "Random"

    m = re.search(r"target\s*:\s*(.*)$", rl, flags=re.IGNORECASE)

    if m:
        return m.group(1).strip()

    return rl


# =========================================================
# BUILD RUNS PER DATASET
# =========================================================
def make_runs_for_dataset(ds: str) -> Dict[str, dict]:
    out = str(OUT_BASE)

    return {
        "QBC target: Bandgap": {
            "base_root": BASE_ROOT_QBC.format(
                OUT=out,
                ds=ds,
                target=TARGET_TO_FOLDER["bandgap"],
            ),
            "path_fn": lambda s, _ds=ds: os.path.join(
                str(s),
                qbc_filename(_ds, "bandgap"),
            ),
            "colmap": COLMAP_QBC,
        },

        "QBC target: Bulk Modulus": {
            "base_root": BASE_ROOT_QBC.format(
                OUT=out,
                ds=ds,
                target=TARGET_TO_FOLDER["bulk_modulus"],
            ),
            "path_fn": lambda s, _ds=ds: os.path.join(
                str(s),
                qbc_filename(_ds, "bulk_modulus"),
            ),
            "colmap": COLMAP_QBC,
        },

        "QBC target: Formation Energy": {
            "base_root": BASE_ROOT_QBC.format(
                OUT=out,
                ds=ds,
                target=TARGET_TO_FOLDER["eform"],
            ),
            "path_fn": lambda s, _ds=ds: os.path.join(
                str(s),
                qbc_filename(_ds, "eform"),
            ),
            "colmap": COLMAP_QBC,
        },

        "Random sampling": {
            "base_root": BASE_ROOT_RANDOM.format(
                OUT=out,
                ds=ds,
            ),
            "path_fn": lambda s, _ds=ds: os.path.join(
                str(s),
                random_filename(_ds, s),
            ),
            "colmap": COLMAP_RANDOM,
        },

        "QBC+Diversity target: Bandgap": {
            "base_root": BASE_ROOT_QBCDIV.format(
                OUT=out,
                ds=ds,
                target=TARGET_TO_FOLDER["bandgap"],
            ),
            "path_fn": lambda s, _ds=ds: os.path.join(
                str(s),
                qbcdiv_filename(_ds, "bandgap", s),
            ),
            "colmap": COLMAP_QBCDIV,
        },

        "QBC+Diversity target: Bulk Modulus": {
            "base_root": BASE_ROOT_QBCDIV.format(
                OUT=out,
                ds=ds,
                target=TARGET_TO_FOLDER["bulk_modulus"],
            ),
            "path_fn": lambda s, _ds=ds: os.path.join(
                str(s),
                qbcdiv_filename(_ds, "bulk_modulus", s),
            ),
            "colmap": COLMAP_QBCDIV,
        },

        "QBC+Diversity target: Formation Energy": {
            "base_root": BASE_ROOT_QBCDIV.format(
                OUT=out,
                ds=ds,
                target=TARGET_TO_FOLDER["eform"],
            ),
            "path_fn": lambda s, _ds=ds: os.path.join(
                str(s),
                qbcdiv_filename(_ds, "eform", s),
            ),
            "colmap": COLMAP_QBCDIV,
        },
    }


# =========================================================
# STEP 1: BUILD LONG AUC SUMMARY FOR ONE DATASET
# =========================================================
def build_auc_summary_long_for_dataset(
    ds: str,
    runs: Dict[str, dict],
) -> Tuple[pd.DataFrame, pd.DataFrame]:

    seed_rows: List[dict] = []

    for run_label, cfg in runs.items():
        base_root = Path(cfg["base_root"])
        path_fn = cfg["path_fn"]
        colmap = cfg["colmap"]
        policy = infer_policy_from_runlabel(run_label)

        for s in seeds:
            csv_path = base_root / path_fn(s)

            if not csv_path.exists():
                print(f"[WARN] Missing file: {csv_path}")
                continue

            df = pd.read_csv(csv_path)
            df = _normalize_colnames(df)

            try:
                xcol = _find_x_column(df)
            except ValueError as e:
                print(f"[WARN] {csv_path}")
                print(f"       {e}")
                continue

            x = _to_fraction01(
                pd.to_numeric(df[xcol], errors="coerce").to_numpy()
            )

            for mkey, display in metrics_internal.items():
                expected_col = colmap.get(mkey, mkey)

                try:
                    actual_col = _resolve_metric_column(df, expected_col)
                except ValueError as e:
                    print(f"[WARN] {csv_path}")
                    print(f"       {e}")
                    continue

                y = pd.to_numeric(df[actual_col], errors="coerce").to_numpy()

                auc_common = auc_trapz_common_grid(x, y)
                auc_raw = auc_trapz_raw(x, y)

                if DEBUG:
                    print(
                        "[DEBUG]",
                        f"dataset={ds}",
                        f"run='{run_label}'",
                        f"policy='{policy}'",
                        f"seed={s}",
                        f"eval='{display}'",
                        f"xcol='{xcol}'",
                        f"ycol='{actual_col}'",
                        f"x_min={np.nanmin(x):.4f}",
                        f"x_max={np.nanmax(x):.4f}",
                        f"n_points={np.isfinite(x).sum()}",
                        f"auc_common={auc_common:.6f}",
                        f"auc_raw={auc_raw:.6f}",
                        f"path='{csv_path}'",
                        sep=" | ",
                    )

                seed_rows.append(
                    {
                        COL_DATASET: ds,
                        COL_TARGET: run_label,
                        COL_PROPERTY: display,
                        COL_POLICY: policy,
                        "seed": int(s),
                        "x_column": xcol,
                        "y_column": actual_col,
                        "x_min": float(np.nanmin(x)),
                        "x_max": float(np.nanmax(x)),
                        "n_points": int(np.isfinite(x).sum()),
                        "auc": auc_common,
                        "auc_raw": auc_raw,
                        "path": str(csv_path),
                    }
                )

    seed_long = pd.DataFrame(seed_rows)

    if seed_long.empty:
        raise RuntimeError(f"No AUC rows were generated for dataset '{ds}'.")

    agg = (
        seed_long
        .groupby(
            [COL_DATASET, COL_TARGET, COL_PROPERTY, COL_POLICY],
            as_index=False,
        )["auc"]
        .agg(auc_mean="mean", auc_std="std")
    )

    agg[COL_STD] = agg[COL_STD].fillna(0.0)

    return agg, seed_long


# =========================================================
# IMPROVEMENT CALCULATION
# =========================================================
def build_improvement_table(df_all: pd.DataFrame) -> pd.DataFrame:
    df = df_all.copy()
    df["target_display"] = df[COL_TARGET].map(_target_display_from_runlabel)

    rows = []

    for ds in DATASETS:
        for eval_prop in TARGET_DISPLAY_ORDER:
            rand = df[
                (df[COL_DATASET] == ds)
                & (df[COL_PROPERTY] == eval_prop)
                & (df[COL_POLICY] == "Random")
            ]

            if rand.empty:
                print(f"[WARN] Missing random baseline for {ds}, {eval_prop}")
                continue

            if len(rand) > 1:
                print(f"[WARN] Multiple random rows for {ds}, {eval_prop}. Using first row.")

            R = float(rand.iloc[0][COL_MEAN])
            sR = float(rand.iloc[0][COL_STD])

            if not np.isfinite(R) or abs(R) < 1e-15:
                print(f"[WARN] Invalid random AUC for {ds}, {eval_prop}: {R}")
                continue

            cand = df[
                (df[COL_DATASET] == ds)
                & (df[COL_PROPERTY] == eval_prop)
                & (df[COL_POLICY].isin(["QBC", "QBC with diversity"]))
            ].copy()

            for _, r in cand.iterrows():
                P = float(r[COL_MEAN])
                sP = float(r[COL_STD])
                target_display = str(r["target_display"])

                improvement = 100.0 * (1.0 - P / R)

                # Error propagation for f = 100 * (1 - P/R)
                dfdP = -100.0 / R
                dfdR = 100.0 * P / (R ** 2)

                s_improvement = float(
                    np.sqrt((dfdP * sP) ** 2 + (dfdR * sR) ** 2)
                )

                section = "Targeted" if target_display == eval_prop else "Untargeted"

                rows.append(
                    {
                        "dataset": ds,
                        "eval_property": eval_prop,
                        "acq_target": target_display,
                        "policy": str(r[COL_POLICY]),
                        "section": section,
                        "auc_policy": P,
                        "auc_policy_std": sP,
                        "auc_random": R,
                        "auc_random_std": sR,
                        "improvement_mean": improvement,
                        "improvement_std": s_improvement,
                    }
                )

    out = pd.DataFrame(rows)

    if out.empty:
        raise RuntimeError("Improvement table is empty. Check random/policy matching.")

    return out


def compute_global_improvement_limits(improv_df: pd.DataFrame) -> Tuple[float, float]:
    vals = improv_df["improvement_mean"].to_numpy(dtype=float)
    errs = improv_df["improvement_std"].to_numpy(dtype=float)

    ymin = np.min(vals - errs)
    ymax = np.max(vals + errs)

    pad = 0.08 * max(ymax - ymin, 1e-12)

    ymin -= pad
    ymax += pad

    ymin = min(ymin, 0.0)
    ymax = max(ymax, 0.0)

    return ymin, ymax


# =========================================================
# PLOTTING HELPERS
# =========================================================
def make_grouped_case_positions():
    n_datasets = len(DATASETS)

    dataset_spacing = 1.7
    group_spacing = 1.45

    cases = []
    xticks = []
    xticklabels = []
    group_centers = {}
    separators = []

    for p_idx, prop in enumerate(TARGET_DISPLAY_ORDER):
        start = p_idx * ((n_datasets - 1) * dataset_spacing + group_spacing + 1.0)

        xs_this_group = []

        for d_idx, ds in enumerate(DATASETS):
            x = start + d_idx * dataset_spacing

            cases.append((prop, ds, x))
            xticks.append(x)
            xticklabels.append(ds.upper())
            xs_this_group.append(x)

        group_centers[prop] = float(np.mean(xs_this_group))

        if p_idx < len(TARGET_DISPLAY_ORDER) - 1:
            separators.append(
                xs_this_group[-1] + 0.5 * dataset_spacing + 0.5 * group_spacing
            )

    return cases, xticks, xticklabels, group_centers, separators


def add_bottom_group_labels(ax, group_centers):
    ymin, ymax = ax.get_ylim()
    y_text = ymin - 0.1 * (ymax - ymin)

    display_map = {
        "Bandgap": "Bandgap",
        "Bulk Modulus": "Bulk Modulus",
        "Formation Energy": "Formation\nEnergy",
    }

    x_shift = {
        "Bandgap": 0.0,
        "Bulk Modulus": 0.5,
        "Formation Energy": 0.0,
    }

    for prop in TARGET_DISPLAY_ORDER:
        ax.text(
            group_centers[prop] + x_shift[prop],
            y_text,
            display_map[prop],
            ha="center",
            va="top",
            fontsize=GROUP_LABEL_FONTSIZE,
            fontweight="bold",
            clip_on=False,
        )


def add_bottom_group_labels_inside(ax, group_centers):
    ymin, ymax = ax.get_ylim()
    y_text = ymin + 0.08 * (ymax - ymin)

    display_map = {
        "Bandgap": "Bandgap",
        "Bulk Modulus": "Bulk Modulus",
        "Formation Energy": "Formation\nEnergy",
    }

    x_shift = {
        "Bandgap": 0.0,
        "Bulk Modulus": -0.5,
        "Formation Energy": 0.0,
    }

    for prop in TARGET_DISPLAY_ORDER:
        ax.text(
            group_centers[prop] + x_shift[prop],
            y_text,
            display_map[prop],
            ha="center",
            va="bottom",
            fontsize=GROUP_LABEL_FONTSIZE,
            fontweight="bold",
        )


def plot_targeted_panel_single_row(ax, improv_df: pd.DataFrame) -> None:
    sub = improv_df[improv_df["section"] == "Targeted"].copy()

    cases, xticks, xticklabels, group_centers, separators = make_grouped_case_positions()

    width = 0.62

    offsets = {
        "QBC": -0.34,
        "QBC with diversity": +0.34,
    }

    for policy in ["QBC", "QBC with diversity"]:
        for eval_prop in TARGET_DISPLAY_ORDER:
            xs, ys, es = [], [], []
            acq_target_for_color = eval_prop

            for prop, ds, x in cases:
                if prop != eval_prop:
                    continue

                r = sub[
                    (sub["dataset"] == ds)
                    & (sub["eval_property"] == prop)
                    & (sub["policy"] == policy)
                ]

                if r.empty:
                    continue

                rr = r.iloc[0]
                acq_target_for_color = str(rr["acq_target"])

                xs.append(x + offsets[policy])
                ys.append(float(rr["improvement_mean"]))
                es.append(float(rr["improvement_std"]))

            if xs:
                ax.bar(
                    xs,
                    ys,
                    width=width,
                    color=TARGET_COLORS[acq_target_for_color],
                    edgecolor="black",
                    linewidth=0.8,
                    hatch=STYLE_MAP[policy]["hatch"],
                    yerr=es,
                    capsize=ERRORBAR_CAPSIZE,
                    error_kw={"elinewidth": 1.0, "capthick": 1.0},
                    zorder=3,
                )

    ax.axhline(0.0, color="black", linewidth=1.0, zorder=2)

    ax.set_xticks(xticks)
    ax.set_xticklabels(
        xticklabels,
        fontsize=XTICK_FONTSIZE,
        rotation=45,
        ha="right",
        fontweight="bold",
    )

    ax.tick_params(axis="y", labelsize=YTICK_FONTSIZE)

    ax.set_xlim(min(xticks) - 1.0, max(xticks) + 1.0)
    ax.grid(axis="y", linestyle="--", alpha=0.25, zorder=0)

    for x_sep in separators:
        ax.axvline(x_sep, color="black", lw=1.0, alpha=0.25)

    add_bottom_group_labels(ax, group_centers)


def plot_untargeted_panel_single_row(ax, improv_df: pd.DataFrame) -> None:
    sub = improv_df[improv_df["section"] == "Untargeted"].copy()

    cases, xticks, xticklabels, group_centers, separators = make_grouped_case_positions()

    width = 0.34
    offsets = [-0.57, -0.19, 0.19, 0.57]

    for eval_prop in TARGET_DISPLAY_ORDER:
        for prop, ds, x in cases:
            if prop != eval_prop:
                continue

            others = [t for t in TARGET_DISPLAY_ORDER if t != eval_prop]

            barspecs = [
                (others[0], "QBC", offsets[0]),
                (others[0], "QBC with diversity", offsets[1]),
                (others[1], "QBC", offsets[2]),
                (others[1], "QBC with diversity", offsets[3]),
            ]

            for acq_target, policy, off in barspecs:
                r = sub[
                    (sub["dataset"] == ds)
                    & (sub["eval_property"] == eval_prop)
                    & (sub["acq_target"] == acq_target)
                    & (sub["policy"] == policy)
                ]

                if r.empty:
                    continue

                rr = r.iloc[0]

                ax.bar(
                    x + off,
                    float(rr["improvement_mean"]),
                    width=width,
                    color=TARGET_COLORS[acq_target],
                    edgecolor="black",
                    linewidth=0.8,
                    hatch=STYLE_MAP[policy]["hatch"],
                    yerr=float(rr["improvement_std"]),
                    capsize=ERRORBAR_CAPSIZE,
                    error_kw={"elinewidth": 1.0, "capthick": 1.0},
                    zorder=3,
                )

    ax.axhline(0.0, color="black", linewidth=1.0, zorder=2)

    ax.set_xticks(xticks)
    ax.set_xticklabels(
        xticklabels,
        fontsize=XTICK_FONTSIZE,
        rotation=45,
        ha="right",
        fontweight="bold",
    )

    ax.tick_params(axis="y", labelsize=YTICK_FONTSIZE)

    ax.set_xlim(min(xticks) - 1.0, max(xticks) + 1.0)
    ax.grid(axis="y", linestyle="--", alpha=0.25, zorder=0)

    for x_sep in separators:
        ax.axvline(x_sep, color="black", lw=1.0, alpha=0.25)

    add_bottom_group_labels_inside(ax, group_centers)


# =========================================================
# FULL FIGURE
# =========================================================
def plot_improvement_figure(improv_df: pd.DataFrame) -> None:
    ylims = compute_global_improvement_limits(improv_df)

    fig, axes = plt.subplots(
        nrows=1,
        ncols=2,
        figsize=(15.5, 6.4),
        squeeze=False,
        sharey=True,
        gridspec_kw={"width_ratios": [1, 1.55]},
    )

    axL = axes[0, 0]
    axR = axes[0, 1]

    plot_targeted_panel_single_row(axL, improv_df)
    plot_untargeted_panel_single_row(axR, improv_df)

    axL.set_ylim(*ylims)
    axR.set_ylim(*ylims)

    axL.set_title("Targeted", fontsize=TITLE_FONTSIZE, fontweight="bold")
    axR.set_title("Untargeted", fontsize=TITLE_FONTSIZE, fontweight="bold")

    axL.set_ylabel(
        "Improvement vs random (%)",
        fontsize=AXIS_LABEL_FONTSIZE,
        fontweight="bold",
    )

    policy_handles = [
        Patch(facecolor="white", edgecolor="black", hatch=None, label="QBC"),
        Patch(facecolor="white", edgecolor="black", hatch="///", label="QBC with diversity"),
    ]

    target_handles = [
        Patch(facecolor=TARGET_COLORS["Bandgap"], edgecolor="black", label="Target: Bandgap"),
        Patch(facecolor=TARGET_COLORS["Bulk Modulus"], edgecolor="black", label="Target: Bulk Modulus"),
        Patch(facecolor=TARGET_COLORS["Formation Energy"], edgecolor="black", label="Target: Formation\nEnergy"),
    ]

    fig.legend(
        handles=policy_handles + target_handles,
        loc="lower center",
        bbox_to_anchor=(0.5, 0.03),
        ncol=5,
        frameon=True,
        fontsize=LEGEND_FONTSIZE,
        title_fontsize=12,
    )

    fig.subplots_adjust(
        left=0.07,
        right=0.995,
        top=0.88,
        bottom=0.31,
        wspace=0.035,
    )

    PLOTS_OUTDIR_BASE.mkdir(parents=True, exist_ok=True)

    outpath = (
        PLOTS_OUTDIR_BASE /
        "TARGETED_UNTARGETED_IMPROVEMENT_SINGLE_ROW_GROUPED_BY_PROPERTY_TARGET_COLORED_XGBoost.png"
    )

    fig.savefig(outpath, dpi=FIG_DPI, bbox_inches="tight")
    plt.show()

    print(f"[OK] Wrote improvement plot to: {outpath}")


# =========================================================
# OPTIONAL DIAGNOSTIC PLOT FOR ONE CASE
# =========================================================
def _display_from_target_key(target_key: str) -> str:
    if target_key == "bandgap":
        return "Bandgap"
    if target_key == "bulk_modulus":
        return "Bulk Modulus"
    if target_key == "eform":
        return "Formation Energy"

    raise ValueError(f"Unknown target key: {target_key}")


def diagnostic_plot_one_case(
    ds: str = "jarvis18",
    eval_metric_key: str = "rmse_bandgap",
    acq_target: str = "bandgap",
) -> None:
    """
    Diagnostic plot for one dataset, one evaluated RMSE metric,
    and one acquisition target.

    This assumes all files have the same number of points. It will raise
    an error if that is not true, because you want a rigorous comparison.
    """
    runs = make_runs_for_dataset(ds)

    run_labels = [
        f"QBC target: {_display_from_target_key(acq_target)}",
        "Random sampling",
        f"QBC+Diversity target: {_display_from_target_key(acq_target)}",
    ]

    fig, ax = plt.subplots(figsize=(7, 5))

    colors = {
        "QBC": "tab:blue",
        "Random": "tab:green",
        "QBC with diversity": "tab:orange",
    }

    for run_label in run_labels:
        if run_label not in runs:
            print(f"[WARN] Diagnostic run not found: {run_label}")
            continue

        cfg = runs[run_label]
        base_root = Path(cfg["base_root"])
        path_fn = cfg["path_fn"]
        colmap = cfg["colmap"]
        policy = infer_policy_from_runlabel(run_label)

        curves = []
        x_refs = []

        for s in seeds:
            csv_path = base_root / path_fn(s)

            if not csv_path.exists():
                print(f"[WARN] Missing diagnostic file: {csv_path}")
                continue

            df = pd.read_csv(csv_path)
            df = _normalize_colnames(df)

            xcol = _find_x_column(df)
            x = _to_fraction01(
                pd.to_numeric(df[xcol], errors="coerce").to_numpy()
            )

            expected_col = colmap.get(eval_metric_key, eval_metric_key)
            actual_col = _resolve_metric_column(df, expected_col)
            y = pd.to_numeric(df[actual_col], errors="coerce").to_numpy()

            x_clean, y_clean = _clean_xy(x, y)

            if x_clean.size < 2:
                print(f"[WARN] Not enough valid points in: {csv_path}")
                continue

            x_refs.append(x_clean)
            curves.append(y_clean)

            print(
                "[DIAG]",
                f"dataset={ds}",
                f"run='{run_label}'",
                f"policy='{policy}'",
                f"seed={s}",
                f"xcol='{xcol}'",
                f"ycol='{actual_col}'",
                f"x_min={x_clean.min():.4f}",
                f"x_max={x_clean.max():.4f}",
                f"n_points={x_clean.size}",
                f"path='{csv_path}'",
                sep=" | ",
            )

        if not curves:
            continue

        lengths = [len(c) for c in curves]
        if len(set(lengths)) != 1:
            raise ValueError(
                f"Seed curves do not have the same number of points for {run_label}. "
                f"Lengths: {lengths}"
            )

        x0 = x_refs[0]
        for idx, x_ref in enumerate(x_refs[1:], start=1):
            if not np.allclose(x0, x_ref, rtol=1e-10, atol=1e-12):
                raise ValueError(
                    f"Seed x-grids differ for {run_label}. "
                    f"Seed 0 and seed {idx} do not have identical train_ratio values."
                )

        curves = np.vstack(curves)

        y_mean = np.nanmean(curves, axis=0)
        y_std = np.nanstd(curves, axis=0)

        auc_mean_curve = np.trapz(y_mean, x0)

        ax.plot(
            x0 * 100,
            y_mean,
            label=f"{policy}, AUC={auc_mean_curve:.4f}",
            color=colors.get(policy, None),
            linewidth=2,
        )

        ax.fill_between(
            x0 * 100,
            y_mean - y_std,
            y_mean + y_std,
            color=colors.get(policy, None),
            alpha=0.18,
            linewidth=0,
        )

    ax.set_xlabel("Training data (%)", fontsize=14)
    ax.set_ylabel(f"RMSE {metrics_internal[eval_metric_key]}", fontsize=14)
    ax.set_title(
        f"{ds.upper()} | {metrics_internal[eval_metric_key]} | Target: {_display_from_target_key(acq_target)}",
        fontsize=14,
    )
    ax.grid(alpha=0.3)
    ax.legend()
    plt.show()


# =========================================================
# MAIN
# =========================================================
def main() -> None:
    AUC_SUMMARY_DIR.mkdir(parents=True, exist_ok=True)
    PLOTS_OUTDIR_BASE.mkdir(parents=True, exist_ok=True)

    all_summaries = []
    all_seed_level = []

    for ds in DATASETS:
        runs_ds = make_runs_for_dataset(ds)

        auc_df_ds, seed_level_ds = build_auc_summary_long_for_dataset(ds, runs_ds)

        out_csv = AUC_SUMMARY_DIR / f"auc_summary_series_long_dft_{ds}_XGBoost.csv"
        auc_df_ds.to_csv(out_csv, index=False)
        print(f"[OK] Wrote AUC summary for {ds} to: {out_csv}")

        seed_csv = AUC_SUMMARY_DIR / f"auc_seed_level_dft_{ds}_XGBoost.csv"
        seed_level_ds.to_csv(seed_csv, index=False)
        print(f"[OK] Wrote seed-level AUC diagnostics for {ds} to: {seed_csv}")

        all_summaries.append(auc_df_ds)
        all_seed_level.append(seed_level_ds)

    df_all = pd.concat(all_summaries, ignore_index=True)
    seed_all = pd.concat(all_seed_level, ignore_index=True)

    all_auc_csv = AUC_SUMMARY_DIR / "auc_summary_series_long_dft_ALL_XGBoost.csv"
    df_all.to_csv(all_auc_csv, index=False)
    print(f"[OK] Wrote combined AUC summary to: {all_auc_csv}")

    seed_all_csv = AUC_SUMMARY_DIR / "auc_seed_level_dft_ALL_XGBoost.csv"
    seed_all.to_csv(seed_all_csv, index=False)
    print(f"[OK] Wrote combined seed-level diagnostics to: {seed_all_csv}")

    improv_df = build_improvement_table(df_all)

    improv_csv = AUC_SUMMARY_DIR / "improvement_vs_random_targeted_untargeted_XGBoost.csv"
    improv_df.to_csv(improv_csv, index=False)
    print(f"[OK] Wrote improvement summary to: {improv_csv}")

    plot_improvement_figure(improv_df)


if __name__ == "__main__":
    main()

    # Optional diagnostic:
    # Uncomment and change values as needed.
    #
    # diagnostic_plot_one_case(
    #     ds="jarvis22",
    #     eval_metric_key="rmse_bulk_modulus",
    #     acq_target="bulk_modulus",
    # )
```

```python
#!/usr/bin/env python3
from __future__ import annotations

import os
import re
from pathlib import Path
from typing import List, Dict, Tuple, Callable, Any

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib.patches import Patch


# ============================================================
# USER CONFIG
# ============================================================
SEEDS = range(5)

DATASETS = ["jarvis18", "jarvis22", "mp18", "mp21"]

TARGET_OUTCOMES = [
    "bandgap_bulkmodulus",
    "bandgap_eform",
    "eform_bulk_modulus",
]

DEBUG = True

# Common AUC grid.
# Assumes train_ratio is in [0, 1].
COMMON_GRID = np.linspace(0.01, 1.00, 100)

FIG_DPI = 100
ERRORBAR_CAPSIZE = 4

BASE_OUTDIR = Path(
    r"c:\Users\rafae\OneDrive\Imágenes\Documentos\PhDToronto\OLD_FILES\PhD Toronto"
    r"\DatasetGeneration\Better_Acqsition\NewAlg_DFT2\AUC_2OBJ"
)

PLOT_OUTDIR = Path(
    r"c:\Users\rafae\OneDrive\Imágenes\Documentos\PhDToronto\OLD_FILES\PhD Toronto"
    r"\DatasetGeneration\Better_Acqsition\NewAlg_DFT2\Plots\Plots_AUC"
)

# ============================================================
# COLUMN NAMES
# ============================================================
COL_DATASET = "dataset"
COL_TARGET = "targets"
COL_PROPERTY = "RMSE"
COL_POLICY = "policy"
COL_MEAN = "auc_mean"
COL_STD = "auc_std"

# ============================================================
# DISPLAY NAMES
# ============================================================
metrics_internal: Dict[str, str] = {
    "bandgap": "Bandgap",
    "bulkmodulus": "Bulk Modulus",
    "eform": "Formation Energy",
}

PROPERTY_ORDER = ["Bandgap", "Bulk Modulus", "Formation Energy"]

TARGET_PAIR_DISPLAY = {
    "bandgap_bulkmodulus": "Bandgap + Bulk Modulus",
    "bandgap_eform": "Bandgap + Formation Energy",
    "eform_bulk_modulus": "Formation Energy + Bulk Modulus",
}

TARGET_PAIR_COLORS = {
    "bandgap_bulkmodulus": "#5E81AC",
    "bandgap_eform": "#A3BE8C",
    "eform_bulk_modulus": "#D08770",
}

# ============================================================
# POLICY STYLE
# ============================================================
POLICY_DISPLAY_ORDER = [
    "2QBC",
    "NSGA-II QBC",
    "QBC + Diversity",
]

POLICY_HATCH = {
    "2QBC": None,
    "NSGA-II QBC": "//",
    "QBC + Diversity": "...",
}

# ============================================================
# PATH BUILDERS
# ============================================================
def random_rmse_path(dataset: str, target: str, seed: int) -> str:
    root = (
        r"c:\Users\rafae\OneDrive\Imágenes\Documentos\PhDToronto\OLD_FILES\PhD Toronto"
        r"\DatasetGeneration\Better_Acqsition\NewAlg_DFT2\al\random"
    )
    return os.path.join(
        root,
        dataset,
        "all",
        str(seed),
        f"{dataset}_all_random_rf{seed}.csv",
    )


def pure_qbc_rmse_path(dataset: str, target: str, seed: int) -> str:
    root = (
        r"c:\Users\rafae\OneDrive\Imágenes\Documentos\PhDToronto\OLD_FILES\PhD Toronto"
        r"\DatasetGeneration\Better_Acqsition\NewAlg_DFT2\al\grow_MultiObj2Outcomes"
    )
    return os.path.join(
        root,
        dataset,
        target,
        str(seed),
        f"{dataset}_{target}MultiObjective_OptDiffModelsOnly2Var_rf.csv",
    )


def ea_qbc_only_rmse_path(dataset: str, target: str, seed: int) -> str:
    root = (
        r"c:\Users\rafae\OneDrive\Imágenes\Documentos\PhDToronto\OLD_FILES\PhD Toronto"
        r"\DatasetGeneration\Better_Acqsition\NewAlg_DFT2\al\grow_2TargetsOnlyQBC"
    )
    return os.path.join(
        root,
        dataset,
        target,
        str(seed),
        f"{dataset}_{target}growing_EA_2obj_qbc_only_rf{seed}.csv",
    )


def ea_qbc_diversity_rmse_path(dataset: str, target: str, seed: int) -> str:
    root = (
        r"c:\Users\rafae\OneDrive\Imágenes\Documentos\PhDToronto\OLD_FILES\PhD Toronto"
        r"\DatasetGeneration\Better_Acqsition\NewAlg_DFT2\al\grow_2TargetsDiversity"
    )
    return os.path.join(
        root,
        dataset,
        target,
        str(seed),
        f"{dataset}_{target}growing_EA_2obj_qbc_Diversity_rf{seed}.csv",
    )


# ============================================================
# HELPERS
# ============================================================
def safe_filename(s: str, max_len: int = 180) -> str:
    s = s.strip()
    s = re.sub(r"\s+", " ", s)
    s = re.sub(r'[<>:"/\\|?*\n\r\t]', "_", s)
    return s[:max_len]


def _normalize_colnames(df: pd.DataFrame) -> pd.DataFrame:
    out = df.copy()
    out.columns = [str(c).strip() for c in out.columns]
    return out


def _find_x_column(df: pd.DataFrame) -> str:
    """
    Strict x-axis detection.

    train_ratio is preferred.
    The function does NOT fall back to arbitrary numeric columns.
    """
    cols = list(df.columns)
    lower = {c: str(c).strip().lower() for c in cols}

    preferred = [
        "train_ratio",
        "train_pct",
        "pct",
        "pcts",
        "percent",
        "percentage",
        "data_pct",
        "pool_pct",
        "train_percent",
        "data_percent",
        "fraction",
        "frac",
        "acq_pct",
        "acq_percent",
        "x",
    ]

    for p in preferred:
        for c in cols:
            if lower[c] == p:
                return c

    raise ValueError(
        "Could not safely identify the x-axis column. "
        f"Available columns are: {cols}"
    )


def _to_fraction01(x: np.ndarray) -> np.ndarray:
    x = pd.to_numeric(pd.Series(x), errors="coerce").to_numpy(dtype=float)
    finite = np.isfinite(x)

    if finite.any() and np.nanmax(x[finite]) > 1.5:
        x = x / 100.0

    return x


def _clean_xy(x: np.ndarray, y: np.ndarray) -> Tuple[np.ndarray, np.ndarray]:
    """
    Remove NaNs, sort by x, and average duplicated x values.
    """
    x = np.asarray(x, dtype=float)
    y = np.asarray(y, dtype=float)

    mask = np.isfinite(x) & np.isfinite(y)
    x = x[mask]
    y = y[mask]

    if x.size < 2:
        return x, y

    order = np.argsort(x)
    x = x[order]
    y = y[order]

    tmp = pd.DataFrame({"x": x, "y": y})
    tmp = tmp.groupby("x", as_index=False)["y"].mean()

    return tmp["x"].to_numpy(), tmp["y"].to_numpy()


def auc_trapz_raw(x: np.ndarray, y: np.ndarray) -> float:
    """
    Raw trapezoidal AUC on the native x-grid.
    Saved only for diagnostics.
    """
    x, y = _clean_xy(x, y)

    if x.size < 2:
        return np.nan

    return float(np.trapz(y, x))


def auc_trapz_common_grid(x: np.ndarray, y: np.ndarray) -> float:
    """
    AUC computed after interpolating onto a common grid.
    """
    x, y = _clean_xy(x, y)

    if x.size < 2:
        return np.nan

    grid = COMMON_GRID[
        (COMMON_GRID >= x.min()) &
        (COMMON_GRID <= x.max())
    ]

    if grid.size < 2:
        return np.nan

    y_interp = np.interp(grid, x, y)

    return float(np.trapz(y_interp, grid))


def resolve_metric_column(df: pd.DataFrame, wanted: str) -> str:
    cols = list(df.columns)

    def norm(s: str) -> str:
        return re.sub(r"[^a-z0-9]+", "", str(s).lower())

    wanted_norm = norm(wanted)

    for c in cols:
        if norm(c) == wanted_norm:
            return c

    for c in cols:
        cn = norm(c)
        if wanted_norm in cn or cn in wanted_norm:
            return c

    raise ValueError(
        f"Could not find metric column '{wanted}'. "
        f"Available columns are: {cols}"
    )


def infer_policy_from_runlabel(run_label: str) -> str:
    rl = run_label.lower()

    if "diversity" in rl:
        return "QBC + Diversity"

    if "ea" in rl and "qbc_only" in rl:
        return "NSGA-II QBC"

    if "pure_qbc" in rl or rl == "qbc":
        return "2QBC"

    if "random" in rl:
        return "Random"

    return run_label


def target_pair_members(target_pair: str) -> List[str]:
    tp = target_pair.strip().lower()

    mapping = {
        "bandgap_bulkmodulus": ["Bandgap", "Bulk Modulus"],
        "bandgap_eform": ["Bandgap", "Formation Energy"],
        "eform_bulk_modulus": ["Formation Energy", "Bulk Modulus"],
    }

    if tp not in mapping:
        raise ValueError(f"Unrecognized target pair: {target_pair}")

    return mapping[tp]


def dataset_label(ds: str) -> str:
    mapping = {
        "jarvis18": "JARVIS18",
        "jarvis22": "JARVIS22",
        "mp18": "MP18",
        "mp21": "MP21",
    }

    return mapping.get(ds, ds.upper())


def format_property_label(prop: str) -> str:
    if prop == "Bulk Modulus":
        return "Bulk\nModulus"

    if prop == "Formation Energy":
        return "Formation\nEnergy"

    return prop


# ============================================================
# STEP 1: BUILD AUC TABLES
# ============================================================
def build_auc_summary_long_for_dataset(
    ds: str,
    runs: Dict[str, dict],
) -> Tuple[pd.DataFrame, pd.DataFrame]:

    seed_rows: List[dict] = []

    for run_label, cfg in runs.items():
        path_fn: Callable[[str, str, int], str] = cfg["path_fn"]
        colmap: Dict[str, Any] = cfg["colmap"]
        policy = infer_policy_from_runlabel(run_label)

        for target in TARGET_OUTCOMES:
            for s in SEEDS:
                csv_path = Path(path_fn(ds, target, s))

                if not csv_path.exists():
                    print(f"[WARN] Missing file: {csv_path}")
                    continue

                df = pd.read_csv(csv_path)
                df = _normalize_colnames(df)

                try:
                    xcol = _find_x_column(df)
                except ValueError as e:
                    print(f"[WARN] {csv_path}")
                    print(f"       {e}")
                    continue

                x = _to_fraction01(df[xcol].to_numpy())

                for mkey, display in metrics_internal.items():
                    candidate_cols = colmap.get(mkey, [mkey])

                    if isinstance(candidate_cols, str):
                        candidate_cols = [candidate_cols]

                    resolved_col = None
                    last_error = None

                    for candidate in candidate_cols:
                        try:
                            resolved_col = resolve_metric_column(df, candidate)
                            break
                        except ValueError as e:
                            last_error = e

                    if resolved_col is None:
                        print(f"[WARN] {csv_path}")
                        print(f"       Missing metric candidates {candidate_cols}")
                        if last_error is not None:
                            print(f"       {last_error}")
                        continue

                    y = pd.to_numeric(df[resolved_col], errors="coerce").to_numpy()

                    auc_common = auc_trapz_common_grid(x, y)
                    auc_raw = auc_trapz_raw(x, y)

                    if DEBUG:
                        finite_x = x[np.isfinite(x)]
                        print(
                            "[DEBUG]",
                            f"dataset={ds}",
                            f"target_pair={target}",
                            f"run='{run_label}'",
                            f"policy='{policy}'",
                            f"seed={s}",
                            f"eval='{display}'",
                            f"xcol='{xcol}'",
                            f"ycol='{resolved_col}'",
                            f"x_min={np.nanmin(finite_x):.4f}",
                            f"x_max={np.nanmax(finite_x):.4f}",
                            f"n_points={np.isfinite(x).sum()}",
                            f"auc_common={auc_common:.6f}",
                            f"auc_raw={auc_raw:.6f}",
                            f"path='{csv_path}'",
                            sep=" | ",
                        )

                    seed_rows.append(
                        {
                            COL_DATASET: ds,
                            COL_TARGET: target,
                            COL_PROPERTY: display,
                            COL_POLICY: policy,
                            "seed": int(s),
                            "x_column": xcol,
                            "y_column": resolved_col,
                            "x_min": float(np.nanmin(x[np.isfinite(x)])),
                            "x_max": float(np.nanmax(x[np.isfinite(x)])),
                            "n_points": int(np.isfinite(x).sum()),
                            "auc": auc_common,
                            "auc_raw": auc_raw,
                            "path": str(csv_path),
                        }
                    )

    seed_long = pd.DataFrame(seed_rows)

    if seed_long.empty:
        raise RuntimeError(f"No AUC rows were generated for dataset '{ds}'.")

    agg = (
        seed_long
        .groupby(
            [COL_DATASET, COL_TARGET, COL_PROPERTY, COL_POLICY],
            as_index=False,
        )["auc"]
        .agg(auc_mean="mean", auc_std="std")
    )

    agg[COL_STD] = agg[COL_STD].fillna(0.0)

    return seed_long, agg


def build_all_auc_tables() -> Tuple[pd.DataFrame, pd.DataFrame]:
    common_colmap = {
        "bandgap": ["rmse_bandgap"],
        "bulkmodulus": ["rmse_bulk_modulus"],
        "eform": ["rmse_eform", "rmse_e_form"],
    }

    runs = {
        "pure_qbc": {
            "path_fn": pure_qbc_rmse_path,
            "colmap": common_colmap,
        },
        "ea_qbc_only": {
            "path_fn": ea_qbc_only_rmse_path,
            "colmap": common_colmap,
        },
        "ea_qbc_diversity": {
            "path_fn": ea_qbc_diversity_rmse_path,
            "colmap": common_colmap,
        },
        "random": {
            "path_fn": random_rmse_path,
            "colmap": common_colmap,
        },
    }

    all_seed_level = []
    all_agg = []

    for ds in DATASETS:
        print(f"\n[INFO] Processing dataset: {ds}")

        seed_ds, agg_ds = build_auc_summary_long_for_dataset(ds, runs)

        seed_path = BASE_OUTDIR / f"auc_seed_level_2objectives_{ds}.csv"
        agg_path = BASE_OUTDIR / f"auc_summary_2objectives_{ds}.csv"

        seed_ds.to_csv(seed_path, index=False)
        agg_ds.to_csv(agg_path, index=False)

        print(f"[OK] Saved seed-level AUC table for {ds} to: {seed_path}")
        print(f"[OK] Saved aggregated AUC summary for {ds} to: {agg_path}")

        all_seed_level.append(seed_ds)
        all_agg.append(agg_ds)

    seed_all = pd.concat(all_seed_level, ignore_index=True)
    agg_all = pd.concat(all_agg, ignore_index=True)

    BASE_OUTDIR.mkdir(parents=True, exist_ok=True)

    seed_all_path = BASE_OUTDIR / "auc_seed_level_2objectives_ALL.csv"
    agg_all_path = BASE_OUTDIR / "auc_summary_long_mod.csv"

    seed_all.to_csv(seed_all_path, index=False)
    agg_all.to_csv(agg_all_path, index=False)

    print(f"[OK] Saved combined seed-level AUC table to: {seed_all_path}")
    print(f"[OK] Saved combined aggregated AUC summary to: {agg_all_path}")

    return seed_all, agg_all


# ============================================================
# STEP 2: IMPROVEMENT VS RANDOM
# ============================================================
def build_improvement_table(df_all: pd.DataFrame) -> pd.DataFrame:
    rows = []

    for ds in DATASETS:
        for eval_prop in PROPERTY_ORDER:
            sub_eval = df_all[
                (df_all[COL_DATASET] == ds) &
                (df_all[COL_PROPERTY] == eval_prop)
            ].copy()

            if sub_eval.empty:
                print(f"[WARN] No rows found for dataset={ds}, eval property={eval_prop}")
                continue

            rand = sub_eval[sub_eval[COL_POLICY] == "Random"]

            if rand.empty:
                print(f"[WARN] Missing random row for dataset={ds}, eval property={eval_prop}")
                continue

            R = float(rand.iloc[0][COL_MEAN])
            sR = float(rand.iloc[0][COL_STD])

            if not np.isfinite(R) or abs(R) < 1e-15:
                print(f"[WARN] Invalid random baseline for dataset={ds}, eval property={eval_prop}")
                continue

            cand = sub_eval[sub_eval[COL_POLICY].isin(POLICY_DISPLAY_ORDER)].copy()

            for _, r in cand.iterrows():
                tp = str(r[COL_TARGET]).strip()

                if tp not in TARGET_PAIR_DISPLAY:
                    print(f"[WARN] Unknown target pair key: {tp}")
                    continue

                pair_members = target_pair_members(tp)

                P = float(r[COL_MEAN])
                sP = float(r[COL_STD])

                improvement = 100.0 * (1.0 - P / R)

                dfdP = -100.0 / R
                dfdR = 100.0 * P / (R ** 2)

                s_improvement = float(
                    np.sqrt((dfdP * sP) ** 2 + (dfdR * sR) ** 2)
                )

                section = "Targeted" if eval_prop in pair_members else "Untargeted"

                rows.append(
                    {
                        "dataset": ds,
                        "eval_property": eval_prop,
                        "target_pair_key": tp,
                        "target_pair": TARGET_PAIR_DISPLAY[tp],
                        "policy": str(r[COL_POLICY]),
                        "section": section,
                        "auc_policy": P,
                        "auc_policy_std": sP,
                        "auc_random": R,
                        "auc_random_std": sR,
                        "improvement_mean": improvement,
                        "improvement_std": s_improvement,
                    }
                )

    out = pd.DataFrame(rows)

    if out.empty:
        raise RuntimeError("Improvement table is empty. Check target-pair parsing and random rows.")

    return out


def compute_global_improvement_limits(improv_df: pd.DataFrame) -> Tuple[float, float]:
    vals = improv_df["improvement_mean"].to_numpy(dtype=float)
    errs = improv_df["improvement_std"].to_numpy(dtype=float)

    ymin = np.min(vals - errs)
    ymax = np.max(vals + errs)

    pad = 0.08 * max(ymax - ymin, 1e-12)

    ymin -= pad
    ymax += pad

    ymin = min(ymin, 0.0)
    ymax = max(ymax, 0.0)

    return ymin, ymax


# ============================================================
# PLOTTING HELPERS
# Group by single evaluated property
# ============================================================
def cases_for_section(section: str) -> List[Tuple[str, str]]:
    return [(prop, ds) for prop in PROPERTY_ORDER for ds in DATASETS]


def target_pairs_for_property_and_section(prop: str, section: str) -> List[str]:
    out = []

    for tp in TARGET_OUTCOMES:
        members = target_pair_members(tp)
        is_targeted = prop in members

        if section == "Targeted" and is_targeted:
            out.append(tp)

        elif section == "Untargeted" and not is_targeted:
            out.append(tp)

    return out


def draw_property_group_labels(ax, base_x: np.ndarray) -> None:
    n_ds = len(DATASETS)

    x0, x1 = ax.get_xlim()

    for i, prop in enumerate(PROPERTY_ORDER):
        group_start = i * n_ds
        group_end = group_start + n_ds - 1
        center_data = 0.5 * (base_x[group_start] + base_x[group_end])
        x_frac = (center_data - x0) / (x1 - x0) - 0.03

        ax.text(
            x_frac,
            0.88,
            format_property_label(prop),
            transform=ax.transAxes,
            ha="center",
            va="bottom",
            fontsize=15,
            fontweight="bold",
            clip_on=False,
        )


def draw_group_separators(ax, group_spacing: float) -> None:
    n_ds = len(DATASETS)

    for g in range(len(PROPERTY_ORDER)):
        start = g * n_ds

        for j in range(1, n_ds):
            ax.axvline(
                (start + j - 0.5) * group_spacing,
                color="black",
                lw=0.6,
                alpha=0.15,
                zorder=1,
            )

    for g in range(1, len(PROPERTY_ORDER)):
        ax.axvline(
            (g * n_ds - 0.5) * group_spacing,
            color="black",
            lw=1.0,
            alpha=0.28,
            zorder=1,
        )


def plot_panel(ax, improv_df: pd.DataFrame, section: str) -> None:
    sub = improv_df[improv_df["section"] == section].copy()

    cases = cases_for_section(section)

    group_spacing = 1.6
    base_x = np.arange(len(cases), dtype=float) * group_spacing

    if section == "Targeted":
        width = 0.22
        offsets = np.array([-2.5, -1.5, -0.5, 0.5, 1.5, 2.5]) * width
    else:
        width = 0.35
        offsets = np.array([-1.0, 0.0, 1.0]) * width

    for i, (prop, ds) in enumerate(cases):
        valid_pairs = target_pairs_for_property_and_section(prop, section)

        barspecs = []
        k = 0

        for tp in valid_pairs:
            for policy in POLICY_DISPLAY_ORDER:
                barspecs.append((tp, policy, offsets[k]))
                k += 1

        for tp, policy, off in barspecs:
            r = sub[
                (sub["dataset"] == ds) &
                (sub["eval_property"] == prop) &
                (sub["target_pair_key"] == tp) &
                (sub["policy"] == policy)
            ]

            if r.empty:
                print(
                    f"[WARN] Missing row: section={section}, dataset={ds}, "
                    f"property={prop}, target_pair={tp}, policy={policy}"
                )
                continue

            rr = r.iloc[0]

            ax.bar(
                base_x[i] + off,
                float(rr["improvement_mean"]),
                width=width,
                color=TARGET_PAIR_COLORS[tp],
                edgecolor="black",
                linewidth=0.8,
                hatch=POLICY_HATCH[policy],
                yerr=float(rr["improvement_std"]),
                capsize=ERRORBAR_CAPSIZE,
                error_kw={"elinewidth": 1.0, "capthick": 1.0},
                zorder=3,
            )

    ax.axhline(0.0, color="black", linewidth=1.0, zorder=2)

    ax.set_xticks(base_x)
    ax.set_xticklabels(
        [dataset_label(ds) for prop, ds in cases],
        fontsize=14,
        rotation=35,
        ha="right",
        rotation_mode="anchor",
        fontweight="bold",
    )

    ax.grid(axis="y", linestyle="--", alpha=0.28, zorder=0)
    draw_group_separators(ax, group_spacing=group_spacing)

    left_edge = base_x[0] + offsets.min() - width / 2
    right_edge = base_x[-1] + offsets.max() + width / 2
    outer_pad = 0.12 * group_spacing

    ax.set_xlim(left_edge - outer_pad, right_edge + outer_pad)

    draw_property_group_labels(ax, base_x=base_x)


def plot_improvement_figure(improv_df: pd.DataFrame) -> None:
    ylims = compute_global_improvement_limits(improv_df)

    fig, axes = plt.subplots(
        nrows=1,
        ncols=2,
        figsize=(16, 7.2),
        squeeze=False,
        sharey=True,
        gridspec_kw={"width_ratios": [1.65, 1.0]},
    )

    axL = axes[0, 0]
    axR = axes[0, 1]

    plot_panel(axL, improv_df, section="Targeted")
    plot_panel(axR, improv_df, section="Untargeted")

    axL.set_ylim(*ylims)
    axR.set_ylim(*ylims)

    axL.set_title("Targeted", fontsize=18, fontweight="bold")
    axR.set_title("Untargeted", fontsize=18, fontweight="bold")

    axL.set_ylabel("Improvement vs random (%)", fontsize=18, fontweight="bold")

    policy_handles = [
        Patch(facecolor="white", edgecolor="black", hatch=POLICY_HATCH["2QBC"], label="2QBC"),
        Patch(facecolor="white", edgecolor="black", hatch=POLICY_HATCH["NSGA-II QBC"], label="NSGA-II QBC"),
        Patch(facecolor="white", edgecolor="black", hatch=POLICY_HATCH["QBC + Diversity"], label="NSGA-II QBC + Diversity"),
    ]

    target_handles = [
        Patch(
            facecolor=TARGET_PAIR_COLORS["bandgap_bulkmodulus"],
            edgecolor="black",
            label="Bandgap + Bulk Modulus",
        ),
        Patch(
            facecolor=TARGET_PAIR_COLORS["bandgap_eform"],
            edgecolor="black",
            label="Bandgap + Formation Energy",
        ),
        Patch(
            facecolor=TARGET_PAIR_COLORS["eform_bulk_modulus"],
            edgecolor="black",
            label="Formation Energy + Bulk Modulus",
        ),
    ]

    fig.legend(
        handles=policy_handles + target_handles,
        loc="lower center",
        bbox_to_anchor=(0.5, -0.02),
        ncol=3,
        frameon=True,
        fontsize=15,
        title_fontsize=11,
        borderpad=0.4,
        labelspacing=0.4,
        handletextpad=0.5,
    )

    fig.subplots_adjust(
        left=0.065,
        right=0.995,
        top=0.84,
        bottom=0.22,
        wspace=0.02,
    )

    PLOT_OUTDIR.mkdir(parents=True, exist_ok=True)

    outpath = PLOT_OUTDIR / "TARGETED_UNTARGETED_IMPROVEMENT_GROUPED_BY_PROPERTY.png"

    fig.savefig(outpath, dpi=FIG_DPI, bbox_inches="tight")
    plt.show()

    print(f"[OK] Wrote improvement plot to: {outpath}")


# ============================================================
# MAIN
# ============================================================
def main() -> None:
    BASE_OUTDIR.mkdir(parents=True, exist_ok=True)
    PLOT_OUTDIR.mkdir(parents=True, exist_ok=True)

    seed_df, agg_df = build_all_auc_tables()

    improv_df = build_improvement_table(agg_df)

    improv_csv = BASE_OUTDIR / "improvement_vs_random_targeted_untargeted_2objectives.csv"
    improv_df.to_csv(improv_csv, index=False)

    print(f"[OK] Saved improvement table to: {improv_csv}")

    print("\n[DEBUG] Improvement table:")
    print(
        improv_df.sort_values(
            ["dataset", "eval_property", "section", "target_pair", "policy"]
        ).to_string(index=False)
    )

    plot_improvement_figure(improv_df)


if __name__ == "__main__":
    main()
```

```python
#!/usr/bin/env python3
from __future__ import annotations

import os
import re
from pathlib import Path
from typing import List, Dict, Tuple, Callable, Any

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib.patches import Patch


# ============================================================
# USER CONFIG
# ============================================================
SEEDS = range(5)

DATASETS = ["jarvis18", "jarvis22", "mp18", "mp21"]

TARGET_OUTCOMES = [
    "bandgap_bulkmodulus",
    "bandgap_eform",
    "eform_bulk_modulus",
]

DEBUG = True

# Common AUC grid.
# Assumes train_ratio is in [0, 1].
COMMON_GRID = np.linspace(0.01, 1.00, 100)

FIG_DPI = 100
ERRORBAR_CAPSIZE = 4

ROOT_BASE = (
    r"c:\Users\rafae\OneDrive\Imágenes\Documentos\PhDToronto\OLD_FILES\PhD Toronto"
    r"\DatasetGeneration\Better_Acqsition\NewAlg_DFT2"
)

BASE_OUTDIR = Path(ROOT_BASE) / "AUC_2OBJ_XGBoost"

PLOT_OUTDIR = Path(ROOT_BASE) / "Plots" / "Plots_AUC_XGBoost"

# ============================================================
# COLUMN NAMES
# ============================================================
COL_DATASET = "dataset"
COL_TARGET = "targets"
COL_PROPERTY = "RMSE"
COL_POLICY = "policy"
COL_MEAN = "auc_mean"
COL_STD = "auc_std"

# ============================================================
# DISPLAY NAMES
# ============================================================
metrics_internal: Dict[str, str] = {
    "bandgap": "Bandgap",
    "bulkmodulus": "Bulk Modulus",
    "eform": "Formation Energy",
}

PROPERTY_ORDER = ["Bandgap", "Bulk Modulus", "Formation Energy"]

TARGET_PAIR_DISPLAY = {
    "bandgap_bulkmodulus": "Bandgap + Bulk Modulus",
    "bandgap_eform": "Bandgap + Formation Energy",
    "eform_bulk_modulus": "Formation Energy + Bulk Modulus",
}

TARGET_PAIR_COLORS = {
    "bandgap_bulkmodulus": "#5E81AC",
    "bandgap_eform": "#A3BE8C",
    "eform_bulk_modulus": "#D08770",
}

# ============================================================
# POLICY STYLE
# ============================================================
POLICY_DISPLAY_ORDER = [
    "2QBC",
    "NSGA-II QBC",
    "QBC + Diversity",
]

POLICY_HATCH = {
    "2QBC": None,
    "NSGA-II QBC": "//",
    "QBC + Diversity": "...",
}

# ============================================================
# PATH BUILDERS - XGBoost
# ============================================================
def random_rmse_path(dataset: str, target: str, seed: int) -> str:
    root = os.path.join(ROOT_BASE, "al", "random")

    return os.path.join(
        root,
        dataset,
        "all",
        str(seed),
        f"{dataset}_all_random_xgb{seed}.csv",
    )


def pure_qbc_rmse_path(dataset: str, target: str, seed: int) -> str:
    root = os.path.join(ROOT_BASE, "al", "grow_MultiObj2Outcomes")

    return os.path.join(
        root,
        dataset,
        target,
        str(seed),
        f"{dataset}_{target}MultiObjective_OptDiffModelsOnly2Var_xgb.csv",
    )


def ea_qbc_only_rmse_path(dataset: str, target: str, seed: int) -> str:
    root = os.path.join(ROOT_BASE, "al", "grow_2TargetsOnlyQBC")

    return os.path.join(
        root,
        dataset,
        target,
        str(seed),
        f"{dataset}_{target}growing_EA_2obj_qbc_only_xgb{seed}.csv",
    )


def ea_qbc_diversity_rmse_path(dataset: str, target: str, seed: int) -> str:
    root = os.path.join(ROOT_BASE, "al", "grow_2TargetsDiversity")

    return os.path.join(
        root,
        dataset,
        target,
        str(seed),
        f"{dataset}_{target}growing_EA_2obj_qbc_Diversity_xgb{seed}.csv",
    )


# ============================================================
# HELPERS
# ============================================================
def safe_filename(s: str, max_len: int = 180) -> str:
    s = s.strip()
    s = re.sub(r"\s+", " ", s)
    s = re.sub(r'[<>:"/\\|?*\n\r\t]', "_", s)
    return s[:max_len]


def _normalize_colnames(df: pd.DataFrame) -> pd.DataFrame:
    out = df.copy()
    out.columns = [str(c).strip() for c in out.columns]
    return out


def _find_x_column(df: pd.DataFrame) -> str:
    """
    Strict x-axis detection.

    train_ratio is preferred.
    This function does NOT fall back to arbitrary numeric columns.
    """
    cols = list(df.columns)
    lower = {c: str(c).strip().lower() for c in cols}

    preferred = [
        "train_ratio",
        "train_pct",
        "pct",
        "pcts",
        "percent",
        "percentage",
        "data_pct",
        "pool_pct",
        "train_percent",
        "data_percent",
        "fraction",
        "frac",
        "acq_pct",
        "acq_percent",
        "x",
    ]

    for p in preferred:
        for c in cols:
            if lower[c] == p:
                return c

    raise ValueError(
        "Could not safely identify the x-axis column. "
        f"Available columns are: {cols}"
    )


def _to_fraction01(x: np.ndarray) -> np.ndarray:
    """
    Convert x-axis to fraction scale [0, 1].

    If x is already train_ratio, it remains unchanged.
    If x is percentage, e.g. 1--100, it is divided by 100.
    """
    x = pd.to_numeric(pd.Series(x), errors="coerce").to_numpy(dtype=float)
    finite = np.isfinite(x)

    if finite.any() and np.nanmax(x[finite]) > 1.5:
        x = x / 100.0

    return x


def _clean_xy(x: np.ndarray, y: np.ndarray) -> Tuple[np.ndarray, np.ndarray]:
    """
    Remove NaNs, sort by x, and average duplicated x values.
    """
    x = np.asarray(x, dtype=float)
    y = np.asarray(y, dtype=float)

    mask = np.isfinite(x) & np.isfinite(y)
    x = x[mask]
    y = y[mask]

    if x.size < 2:
        return x, y

    order = np.argsort(x)
    x = x[order]
    y = y[order]

    tmp = pd.DataFrame({"x": x, "y": y})
    tmp = tmp.groupby("x", as_index=False)["y"].mean()

    return tmp["x"].to_numpy(), tmp["y"].to_numpy()


def auc_trapz_raw(x: np.ndarray, y: np.ndarray) -> float:
    """
    Raw trapezoidal AUC on the native x-grid.
    Saved only for diagnostics.
    """
    x, y = _clean_xy(x, y)

    if x.size < 2:
        return np.nan

    return float(np.trapz(y, x))


def auc_trapz_common_grid(x: np.ndarray, y: np.ndarray) -> float:
    """
    AUC computed after interpolating onto a common grid.
    """
    x, y = _clean_xy(x, y)

    if x.size < 2:
        return np.nan

    grid = COMMON_GRID[
        (COMMON_GRID >= x.min()) &
        (COMMON_GRID <= x.max())
    ]

    if grid.size < 2:
        return np.nan

    y_interp = np.interp(grid, x, y)

    return float(np.trapz(y_interp, grid))


def resolve_metric_column(df: pd.DataFrame, wanted: str) -> str:
    """
    Resolve metric column robustly.
    Uses normalized exact match first, then partial normalized match.
    """
    cols = list(df.columns)

    def norm(s: str) -> str:
        return re.sub(r"[^a-z0-9]+", "", str(s).lower())

    wanted_norm = norm(wanted)

    for c in cols:
        if norm(c) == wanted_norm:
            return c

    for c in cols:
        cn = norm(c)
        if wanted_norm in cn or cn in wanted_norm:
            return c

    raise ValueError(
        f"Could not find metric column '{wanted}'. "
        f"Available columns are: {cols}"
    )


def infer_policy_from_runlabel(run_label: str) -> str:
    rl = run_label.lower()

    if "diversity" in rl:
        return "QBC + Diversity"

    if "ea" in rl and "qbc_only" in rl:
        return "NSGA-II QBC"

    if "pure_qbc" in rl or rl == "qbc":
        return "2QBC"

    if "random" in rl:
        return "Random"

    return run_label


def target_pair_members(target_pair: str) -> List[str]:
    tp = target_pair.strip().lower()

    mapping = {
        "bandgap_bulkmodulus": ["Bandgap", "Bulk Modulus"],
        "bandgap_eform": ["Bandgap", "Formation Energy"],
        "eform_bulk_modulus": ["Formation Energy", "Bulk Modulus"],
    }

    if tp not in mapping:
        raise ValueError(f"Unrecognized target pair: {target_pair}")

    return mapping[tp]


def dataset_label(ds: str) -> str:
    mapping = {
        "jarvis18": "JARVIS18",
        "jarvis22": "JARVIS22",
        "mp18": "MP18",
        "mp21": "MP21",
    }

    return mapping.get(ds, ds.upper())


def format_property_label(prop: str) -> str:
    if prop == "Bulk Modulus":
        return "Bulk\nModulus"

    if prop == "Formation Energy":
        return "Formation\nEnergy"

    return prop


# ============================================================
# STEP 1: BUILD AUC TABLES
# ============================================================
def build_auc_summary_long_for_dataset(
    ds: str,
    runs: Dict[str, dict],
) -> Tuple[pd.DataFrame, pd.DataFrame]:

    seed_rows: List[dict] = []

    for run_label, cfg in runs.items():
        path_fn: Callable[[str, str, int], str] = cfg["path_fn"]
        colmap: Dict[str, Any] = cfg["colmap"]
        policy = infer_policy_from_runlabel(run_label)

        for target in TARGET_OUTCOMES:
            for s in SEEDS:
                csv_path = Path(path_fn(ds, target, s))

                if not csv_path.exists():
                    print(f"[WARN] Missing file: {csv_path}")
                    continue

                df = pd.read_csv(csv_path)
                df = _normalize_colnames(df)

                try:
                    xcol = _find_x_column(df)
                except ValueError as e:
                    print(f"[WARN] {csv_path}")
                    print(f"       {e}")
                    continue

                x = _to_fraction01(df[xcol].to_numpy())

                finite_x = x[np.isfinite(x)]
                if finite_x.size < 2:
                    print(f"[WARN] Not enough finite x values in: {csv_path}")
                    continue

                for mkey, display in metrics_internal.items():
                    candidate_cols = colmap.get(mkey, [mkey])

                    if isinstance(candidate_cols, str):
                        candidate_cols = [candidate_cols]

                    resolved_col = None
                    last_error = None

                    for candidate in candidate_cols:
                        try:
                            resolved_col = resolve_metric_column(df, candidate)
                            break
                        except ValueError as e:
                            last_error = e

                    if resolved_col is None:
                        print(f"[WARN] {csv_path}")
                        print(f"       Missing metric candidates {candidate_cols}")
                        if last_error is not None:
                            print(f"       {last_error}")
                        continue

                    y = pd.to_numeric(df[resolved_col], errors="coerce").to_numpy()

                    auc_common = auc_trapz_common_grid(x, y)
                    auc_raw = auc_trapz_raw(x, y)

                    if DEBUG:
                        print(
                            "[DEBUG]",
                            f"dataset={ds}",
                            f"target_pair={target}",
                            f"run='{run_label}'",
                            f"policy='{policy}'",
                            f"seed={s}",
                            f"eval='{display}'",
                            f"xcol='{xcol}'",
                            f"ycol='{resolved_col}'",
                            f"x_min={np.nanmin(finite_x):.4f}",
                            f"x_max={np.nanmax(finite_x):.4f}",
                            f"n_points={np.isfinite(x).sum()}",
                            f"auc_common={auc_common:.6f}",
                            f"auc_raw={auc_raw:.6f}",
                            f"path='{csv_path}'",
                            sep=" | ",
                        )

                    seed_rows.append(
                        {
                            COL_DATASET: ds,
                            COL_TARGET: target,
                            COL_PROPERTY: display,
                            COL_POLICY: policy,
                            "seed": int(s),
                            "x_column": xcol,
                            "y_column": resolved_col,
                            "x_min": float(np.nanmin(finite_x)),
                            "x_max": float(np.nanmax(finite_x)),
                            "n_points": int(np.isfinite(x).sum()),
                            "auc": auc_common,
                            "auc_raw": auc_raw,
                            "path": str(csv_path),
                        }
                    )

    seed_long = pd.DataFrame(seed_rows)

    if seed_long.empty:
        raise RuntimeError(f"No AUC rows were generated for dataset '{ds}'.")

    agg = (
        seed_long
        .groupby(
            [COL_DATASET, COL_TARGET, COL_PROPERTY, COL_POLICY],
            as_index=False,
        )["auc"]
        .agg(auc_mean="mean", auc_std="std")
    )

    agg[COL_STD] = agg[COL_STD].fillna(0.0)

    return seed_long, agg


def build_all_auc_tables() -> Tuple[pd.DataFrame, pd.DataFrame]:
    common_colmap = {
        "bandgap": ["rmse_bandgap"],
        "bulkmodulus": ["rmse_bulk_modulus"],
        "eform": ["rmse_eform", "rmse_e_form"],
    }

    runs = {
        "pure_qbc": {
            "path_fn": pure_qbc_rmse_path,
            "colmap": common_colmap,
        },
        "ea_qbc_only": {
            "path_fn": ea_qbc_only_rmse_path,
            "colmap": common_colmap,
        },
        "ea_qbc_diversity": {
            "path_fn": ea_qbc_diversity_rmse_path,
            "colmap": common_colmap,
        },
        "random": {
            "path_fn": random_rmse_path,
            "colmap": common_colmap,
        },
    }

    all_seed_level = []
    all_agg = []

    BASE_OUTDIR.mkdir(parents=True, exist_ok=True)

    for ds in DATASETS:
        print(f"\n[INFO] Processing dataset: {ds}")

        seed_ds, agg_ds = build_auc_summary_long_for_dataset(ds, runs)

        seed_path = BASE_OUTDIR / f"auc_seed_level_2objectives_{ds}_xgboost.csv"
        agg_path = BASE_OUTDIR / f"auc_summary_2objectives_{ds}_xgboost.csv"

        seed_ds.to_csv(seed_path, index=False)
        agg_ds.to_csv(agg_path, index=False)

        print(f"[OK] Saved seed-level AUC table for {ds} to: {seed_path}")
        print(f"[OK] Saved aggregated AUC summary for {ds} to: {agg_path}")

        all_seed_level.append(seed_ds)
        all_agg.append(agg_ds)

    seed_all = pd.concat(all_seed_level, ignore_index=True)
    agg_all = pd.concat(all_agg, ignore_index=True)

    seed_all_path = BASE_OUTDIR / "auc_seed_level_2objectives_ALL_xgboost.csv"
    agg_all_path = BASE_OUTDIR / "auc_summary_long_mod_xgboost.csv"

    seed_all.to_csv(seed_all_path, index=False)
    agg_all.to_csv(agg_all_path, index=False)

    print(f"[OK] Saved combined seed-level AUC table to: {seed_all_path}")
    print(f"[OK] Saved combined aggregated AUC summary to: {agg_all_path}")

    return seed_all, agg_all


# ============================================================
# STEP 2: IMPROVEMENT VS RANDOM
# ============================================================
def build_improvement_table(df_all: pd.DataFrame) -> pd.DataFrame:
    rows = []

    for ds in DATASETS:
        for eval_prop in PROPERTY_ORDER:
            sub_eval = df_all[
                (df_all[COL_DATASET] == ds) &
                (df_all[COL_PROPERTY] == eval_prop)
            ].copy()

            if sub_eval.empty:
                print(f"[WARN] No rows found for dataset={ds}, eval property={eval_prop}")
                continue

            rand = sub_eval[sub_eval[COL_POLICY] == "Random"]

            if rand.empty:
                print(f"[WARN] Missing random row for dataset={ds}, eval property={eval_prop}")
                continue

            R = float(rand.iloc[0][COL_MEAN])
            sR = float(rand.iloc[0][COL_STD])

            if not np.isfinite(R) or abs(R) < 1e-15:
                print(f"[WARN] Invalid random baseline for dataset={ds}, eval property={eval_prop}")
                continue

            cand = sub_eval[sub_eval[COL_POLICY].isin(POLICY_DISPLAY_ORDER)].copy()

            for _, r in cand.iterrows():
                tp = str(r[COL_TARGET]).strip()

                if tp not in TARGET_PAIR_DISPLAY:
                    print(f"[WARN] Unknown target pair key: {tp}")
                    continue

                pair_members = target_pair_members(tp)

                P = float(r[COL_MEAN])
                sP = float(r[COL_STD])

                improvement = 100.0 * (1.0 - P / R)

                dfdP = -100.0 / R
                dfdR = 100.0 * P / (R ** 2)

                s_improvement = float(
                    np.sqrt((dfdP * sP) ** 2 + (dfdR * sR) ** 2)
                )

                section = "Targeted" if eval_prop in pair_members else "Untargeted"

                rows.append(
                    {
                        "dataset": ds,
                        "eval_property": eval_prop,
                        "target_pair_key": tp,
                        "target_pair": TARGET_PAIR_DISPLAY[tp],
                        "policy": str(r[COL_POLICY]),
                        "section": section,
                        "auc_policy": P,
                        "auc_policy_std": sP,
                        "auc_random": R,
                        "auc_random_std": sR,
                        "improvement_mean": improvement,
                        "improvement_std": s_improvement,
                    }
                )

    out = pd.DataFrame(rows)

    if out.empty:
        raise RuntimeError("Improvement table is empty. Check target-pair parsing and random rows.")

    return out


def compute_global_improvement_limits(improv_df: pd.DataFrame) -> Tuple[float, float]:
    vals = improv_df["improvement_mean"].to_numpy(dtype=float)
    errs = improv_df["improvement_std"].to_numpy(dtype=float)

    ymin = np.min(vals - errs)
    ymax = np.max(vals + errs)

    pad = 0.08 * max(ymax - ymin, 1e-12)

    ymin -= pad
    ymax += pad

    ymin = min(ymin, 0.0)
    ymax = max(ymax, 0.0)

    return ymin, ymax


# ============================================================
# PLOTTING HELPERS
# ============================================================
def cases_for_section(section: str) -> List[Tuple[str, str]]:
    return [(prop, ds) for prop in PROPERTY_ORDER for ds in DATASETS]


def target_pairs_for_property_and_section(prop: str, section: str) -> List[str]:
    out = []

    for tp in TARGET_OUTCOMES:
        members = target_pair_members(tp)
        is_targeted = prop in members

        if section == "Targeted" and is_targeted:
            out.append(tp)

        elif section == "Untargeted" and not is_targeted:
            out.append(tp)

    return out


def draw_property_group_labels(ax, base_x: np.ndarray) -> None:
    n_ds = len(DATASETS)

    x0, x1 = ax.get_xlim()

    for i, prop in enumerate(PROPERTY_ORDER):
        group_start = i * n_ds
        group_end = group_start + n_ds - 1
        center_data = 0.5 * (base_x[group_start] + base_x[group_end])
        x_frac = (center_data - x0) / (x1 - x0) - 0.03

        ax.text(
            x_frac,
            0.88,
            format_property_label(prop),
            transform=ax.transAxes,
            ha="center",
            va="bottom",
            fontsize=15,
            fontweight="bold",
            clip_on=False,
        )


def draw_group_separators(ax, group_spacing: float) -> None:
    n_ds = len(DATASETS)

    for g in range(len(PROPERTY_ORDER)):
        start = g * n_ds

        for j in range(1, n_ds):
            ax.axvline(
                (start + j - 0.5) * group_spacing,
                color="black",
                lw=0.6,
                alpha=0.15,
                zorder=1,
            )

    for g in range(1, len(PROPERTY_ORDER)):
        ax.axvline(
            (g * n_ds - 0.5) * group_spacing,
            color="black",
            lw=1.0,
            alpha=0.28,
            zorder=1,
        )


def plot_panel(ax, improv_df: pd.DataFrame, section: str) -> None:
    sub = improv_df[improv_df["section"] == section].copy()

    cases = cases_for_section(section)

    group_spacing = 1.6
    base_x = np.arange(len(cases), dtype=float) * group_spacing

    if section == "Targeted":
        width = 0.22
        offsets = np.array([-2.5, -1.5, -0.5, 0.5, 1.5, 2.5]) * width
    else:
        width = 0.35
        offsets = np.array([-1.0, 0.0, 1.0]) * width

    for i, (prop, ds) in enumerate(cases):
        valid_pairs = target_pairs_for_property_and_section(prop, section)

        barspecs = []
        k = 0

        for tp in valid_pairs:
            for policy in POLICY_DISPLAY_ORDER:
                barspecs.append((tp, policy, offsets[k]))
                k += 1

        for tp, policy, off in barspecs:
            r = sub[
                (sub["dataset"] == ds) &
                (sub["eval_property"] == prop) &
                (sub["target_pair_key"] == tp) &
                (sub["policy"] == policy)
            ]

            if r.empty:
                print(
                    f"[WARN] Missing row: section={section}, dataset={ds}, "
                    f"property={prop}, target_pair={tp}, policy={policy}"
                )
                continue

            rr = r.iloc[0]

            ax.bar(
                base_x[i] + off,
                float(rr["improvement_mean"]),
                width=width,
                color=TARGET_PAIR_COLORS[tp],
                edgecolor="black",
                linewidth=0.8,
                hatch=POLICY_HATCH[policy],
                yerr=float(rr["improvement_std"]),
                capsize=ERRORBAR_CAPSIZE,
                error_kw={"elinewidth": 1.0, "capthick": 1.0},
                zorder=3,
            )

    ax.axhline(0.0, color="black", linewidth=1.0, zorder=2)

    ax.set_xticks(base_x)

    ax.set_xticklabels(
        [dataset_label(ds) for prop, ds in cases],
        fontsize=14,
        rotation=35,
        ha="right",
        rotation_mode="anchor",
        fontweight="bold",
    )

    ax.grid(axis="y", linestyle="--", alpha=0.28, zorder=0)

    draw_group_separators(ax, group_spacing=group_spacing)

    left_edge = base_x[0] + offsets.min() - width / 2
    right_edge = base_x[-1] + offsets.max() + width / 2
    outer_pad = 0.12 * group_spacing

    ax.set_xlim(left_edge - outer_pad, right_edge + outer_pad)

    draw_property_group_labels(ax, base_x=base_x)


def plot_improvement_figure(improv_df: pd.DataFrame) -> None:
    ylims = compute_global_improvement_limits(improv_df)

    fig, axes = plt.subplots(
        nrows=1,
        ncols=2,
        figsize=(16, 7.2),
        squeeze=False,
        sharey=True,
        gridspec_kw={"width_ratios": [1.65, 1.0]},
    )

    axL = axes[0, 0]
    axR = axes[0, 1]

    plot_panel(axL, improv_df, section="Targeted")
    plot_panel(axR, improv_df, section="Untargeted")

    axL.set_ylim(*ylims)
    axR.set_ylim(*ylims)

    axL.set_title("Targeted", fontsize=18, fontweight="bold")
    axR.set_title("Untargeted", fontsize=18, fontweight="bold")

    axL.set_ylabel(
        "Improvement vs random (%)",
        fontsize=18,
        fontweight="bold",
    )

    policy_handles = [
        Patch(facecolor="white", edgecolor="black", hatch=POLICY_HATCH["2QBC"], label="2QBC"),
        Patch(facecolor="white", edgecolor="black", hatch=POLICY_HATCH["NSGA-II QBC"], label="NSGA-II QBC"),
        Patch(facecolor="white", edgecolor="black", hatch=POLICY_HATCH["QBC + Diversity"], label="NSGA-II QBC + Diversity"),
    ]

    target_handles = [
        Patch(
            facecolor=TARGET_PAIR_COLORS["bandgap_bulkmodulus"],
            edgecolor="black",
            label="Bandgap + Bulk Modulus",
        ),
        Patch(
            facecolor=TARGET_PAIR_COLORS["bandgap_eform"],
            edgecolor="black",
            label="Bandgap + Formation Energy",
        ),
        Patch(
            facecolor=TARGET_PAIR_COLORS["eform_bulk_modulus"],
            edgecolor="black",
            label="Formation Energy + Bulk Modulus",
        ),
    ]

    fig.legend(
        handles=policy_handles + target_handles,
        loc="lower center",
        bbox_to_anchor=(0.5, -0.02),
        ncol=3,
        frameon=True,
        fontsize=15,
        title_fontsize=11,
        borderpad=0.4,
        labelspacing=0.4,
        handletextpad=0.5,
    )

    fig.subplots_adjust(
        left=0.065,
        right=0.995,
        top=0.84,
        bottom=0.22,
        wspace=0.02,
    )

    PLOT_OUTDIR.mkdir(parents=True, exist_ok=True)

    outpath = PLOT_OUTDIR / "TARGETED_UNTARGETED_IMPROVEMENT_GROUPED_BY_PROPERTY_XGBoost.png"

    fig.savefig(outpath, dpi=FIG_DPI, bbox_inches="tight")
    plt.show()

    print(f"[OK] Wrote improvement plot to: {outpath}")


# ============================================================
# MAIN
# ============================================================
def main() -> None:
    BASE_OUTDIR.mkdir(parents=True, exist_ok=True)
    PLOT_OUTDIR.mkdir(parents=True, exist_ok=True)

    seed_df, agg_df = build_all_auc_tables()

    improv_df = build_improvement_table(agg_df)

    improv_csv = BASE_OUTDIR / "improvement_vs_random_targeted_untargeted_2objectives_xgboost.csv"
    improv_df.to_csv(improv_csv, index=False)

    print(f"[OK] Saved improvement table to: {improv_csv}")

    print("\n[DEBUG] Improvement table:")
    print(
        improv_df.sort_values(
            ["dataset", "eval_property", "section", "target_pair", "policy"]
        ).to_string(index=False)
    )

    plot_improvement_figure(improv_df)


if __name__ == "__main__":
    main()
```

```python
import os
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

# ---------------------------------------------------------------------------
# JOINT PLOTS.
# ONE PLOT PER TARGETED PROPERTY AND POLICY.
# Also computes:
#   1. Mean RMSE across seeds at each train_ratio.
#   2. Pairwise RMSE differences at train_ratio = 0.40.
#   3. Improvement at 40% relative to the non-diversity QBC case.
# ---------------------------------------------------------------------------

# ============================================================
# CONFIG
# ============================================================
seeds = range(5)

DATASETS = ["jarvis18", "jarvis22", "mp18", "mp21"]
TARGETS = ["bandgap", "bulk_modulus", "e_form"]

TARGET_RATIO = 0.40
TARGET_RATIO_PERCENT = TARGET_RATIO * 100.0

# Where your CSVs live
ROOT_RANDOM = r"C:\Users\rafae\OneDrive\Imágenes\Documentos\PhDToronto\OLD_FILES\PhD Toronto\DatasetGeneration\Better_Acqsition\NewAlg_DFT2\al\random"
ROOT_QBC    = r"C:\Users\rafae\OneDrive\Imágenes\Documentos\PhDToronto\OLD_FILES\PhD Toronto\DatasetGeneration\Better_Acqsition\NewAlg_DFT2\al\QBC"
ROOT_QBCDIV = r"C:\Users\rafae\OneDrive\Imágenes\Documentos\PhDToronto\OLD_FILES\PhD Toronto\DatasetGeneration\Better_Acqsition\NewAlg_DFT2\al\QBC_DIVERSITY"

# Output base
OUT_DIR = r"C:\Users\rafae\OneDrive\Imágenes\Documentos\PhDToronto\OLD_FILES\PhD Toronto\DatasetGeneration\Better_Acqsition\Plots_DFT"

# Tables output
TABLE_OUT_DIR = os.path.join(OUT_DIR, "RMSE_40pct_tables")
os.makedirs(TABLE_OUT_DIR, exist_ok=True)

# ============================================================
# Metrics
# ============================================================
metrics_internal = {
    "rmse_bandgap": "Bandgap (eV)",
    "rmse_bulk_modulus": "Bulk Modulus (GPa)",
    "rmse_eform": r"Formation Energy (eV atom$^{-1}$)",
}

metric_short_names = {
    "rmse_bandgap": "Bandgap",
    "rmse_bulk_modulus": "Bulk Modulus",
    "rmse_eform": "Formation Energy",
}

# Column maps
COLMAP_RANDOM_QBC = {
    "rmse_bandgap": "rmse_bandgap",
    "rmse_bulk_modulus": "rmse_bulk_modulus",
    "rmse_eform": "rmse_eform",
}

COLMAP_QBCDIV = {
    "rmse_bandgap": "rmse_bandgap",
    "rmse_bulk_modulus": "rmse_bulk_modulus",
    "rmse_eform": "rmse_e_form",
}

# ============================================================
# Filename conventions
# ============================================================
def random_csv_name(dataset: str, seed: int) -> str:
    return f"{dataset}_all_random_rf{seed}.csv"


def qbc_csv_name(dataset: str, target: str) -> str:
    return f"{dataset}_{target}_QBC_rf.csv"


def qbcdiv_csv_name(dataset: str, target: str, seed: int) -> str:
    return f"{dataset}_{target}_QBCdiversity_rf{seed}.csv"


def target_label(target: str) -> str:
    if target == "bandgap":
        return "Bandgap"
    if target == "bulk_modulus":
        return "Bulk Modulus"
    if target == "e_form":
        return "Formation Energy"
    return target


# ============================================================
# IO helpers
# ============================================================
def read_run_csv(path: str, colmap: dict) -> pd.DataFrame:
    df = pd.read_csv(path)
    df.columns = [str(c).strip() for c in df.columns]

    if "train_ratio" not in df.columns:
        raise ValueError(f"Missing 'train_ratio' in {path}")

    for internal_name, file_col in colmap.items():
        if file_col not in df.columns:
            raise ValueError(
                f"Missing '{file_col}' mapped to {internal_name} in {path}. "
                f"Available columns: {list(df.columns)}"
            )

    out = pd.DataFrame({
        "train_ratio": pd.to_numeric(df["train_ratio"], errors="coerce")
    })

    for internal_name, file_col in colmap.items():
        out[internal_name] = pd.to_numeric(df[file_col], errors="coerce")

    return out


def aggregate_over_seeds(paths: list, colmap: dict) -> tuple[pd.DataFrame, pd.DataFrame]:
    """
    Returns:
        data_long:
            One row per seed and train_ratio.
        agg:
            Mean and std RMSE at each train_ratio across seeds.
    """
    per_seed = []

    for seed, p in paths:
        tmp = read_run_csv(p, colmap)
        tmp["seed"] = seed
        tmp["path"] = p
        per_seed.append(tmp)

    data_long = pd.concat(per_seed, ignore_index=True)

    agg = (
        data_long
        .groupby("train_ratio")[list(metrics_internal.keys())]
        .agg(["mean", "std"])
        .reset_index()
        .sort_values("train_ratio")
    )

    return data_long, agg


# ============================================================
# 40% RMSE extraction and comparison helpers
# ============================================================
def get_mean_rmse_at_ratio(
    agg: pd.DataFrame,
    metric_col: str,
    target_ratio: float = TARGET_RATIO,
    strict_exact: bool = False,
) -> tuple[float, str]:
    """
    Extract the mean RMSE at target_ratio.

    If strict_exact=True, this raises an error when target_ratio is absent.
    If strict_exact=False, it interpolates if target_ratio is not exactly present.

    Returns:
        value, mode
        mode is "exact" or "interpolated".
    """
    x = agg["train_ratio"].to_numpy(dtype=float)
    y = agg[(metric_col, "mean")].to_numpy(dtype=float)

    mask = np.isfinite(x) & np.isfinite(y)
    x = x[mask]
    y = y[mask]

    if x.size < 2:
        raise ValueError(f"Not enough finite points for {metric_col}.")

    order = np.argsort(x)
    x = x[order]
    y = y[order]

    exact_mask = np.isclose(x, target_ratio, rtol=1e-10, atol=1e-12)

    if exact_mask.any():
        return float(y[exact_mask][0]), "exact"

    if strict_exact:
        raise ValueError(
            f"train_ratio={target_ratio} was not found exactly. "
            f"Available train_ratio values: {x}"
        )

    if target_ratio < x.min() or target_ratio > x.max():
        raise ValueError(
            f"target_ratio={target_ratio} is outside available range "
            f"[{x.min()}, {x.max()}]."
        )

    return float(np.interp(target_ratio, x, y)), "interpolated"


def flatten_mean_rmse_table(
    dataset: str,
    target: str,
    agg_by_run: dict,
) -> pd.DataFrame:
    """
    Saves the mean RMSE at every train_ratio for every policy.
    """
    rows = []

    for run_name, agg in agg_by_run.items():
        x = agg["train_ratio"].to_numpy(dtype=float)

        for metric_col in metrics_internal:
            mean_vals = agg[(metric_col, "mean")].to_numpy(dtype=float)
            std_vals = agg[(metric_col, "std")].to_numpy(dtype=float)

            for tr, mean_rmse, std_rmse in zip(x, mean_vals, std_vals):
                rows.append({
                    "dataset": dataset,
                    "target": target,
                    "target_label": target_label(target),
                    "policy": run_name,
                    "metric": metric_short_names[metric_col],
                    "metric_col": metric_col,
                    "train_ratio": float(tr),
                    "training_percent": float(tr * 100.0),
                    "mean_rmse": float(mean_rmse),
                    "std_rmse": float(std_rmse),
                })

    return pd.DataFrame(rows)


def compute_40pct_comparisons(
    dataset: str,
    target: str,
    agg_by_run: dict,
) -> pd.DataFrame:
    """
    Computes pairwise differences between policies at 40% training data.

    Also computes improvement of QBC+Diversity relative to QBC:

        100 * (1 - RMSE_QBCDiv / RMSE_QBC)

    Positive improvement means diversity has lower RMSE than non-diversity QBC.
    """
    run_random = "Random sampling"
    run_qbc = f"QBC target: {target_label(target)}"
    run_div = f"QBC+Diversity target: {target_label(target)}"

    required = [run_random, run_qbc, run_div]
    for r in required:
        if r not in agg_by_run:
            raise KeyError(f"Missing run '{r}' for {dataset}, target={target}")

    rows = []

    for metric_col in metrics_internal:
        metric_name = metric_short_names[metric_col]

        random_rmse, random_mode = get_mean_rmse_at_ratio(
            agg_by_run[run_random],
            metric_col,
            target_ratio=TARGET_RATIO,
            strict_exact=False,
        )

        qbc_rmse, qbc_mode = get_mean_rmse_at_ratio(
            agg_by_run[run_qbc],
            metric_col,
            target_ratio=TARGET_RATIO,
            strict_exact=False,
        )

        div_rmse, div_mode = get_mean_rmse_at_ratio(
            agg_by_run[run_div],
            metric_col,
            target_ratio=TARGET_RATIO,
            strict_exact=False,
        )

        # Signed differences.
        # Positive value means first policy has higher RMSE than second policy.
        diff_random_minus_qbc = random_rmse - qbc_rmse
        diff_random_minus_div = random_rmse - div_rmse
        diff_qbc_minus_div = qbc_rmse - div_rmse

        # Improvement relative to non-diversity case.
        # Positive means diversity improves over QBC.
        improvement_div_vs_qbc = 100.0 * (1.0 - div_rmse / qbc_rmse)

        rows.append({
            "dataset": dataset,
            "target": target,
            "target_label": target_label(target),
            "metric": metric_name,
            "metric_col": metric_col,
            "train_ratio": TARGET_RATIO,
            "training_percent": TARGET_RATIO_PERCENT,

            "random_mean_rmse_40": random_rmse,
            "qbc_mean_rmse_40": qbc_rmse,
            "qbc_diversity_mean_rmse_40": div_rmse,

            "random_extraction": random_mode,
            "qbc_extraction": qbc_mode,
            "qbc_diversity_extraction": div_mode,

            "diff_random_minus_qbc": diff_random_minus_qbc,
            "diff_random_minus_qbc_abs": abs(diff_random_minus_qbc),

            "diff_random_minus_qbc_diversity": diff_random_minus_div,
            "diff_random_minus_qbc_diversity_abs": abs(diff_random_minus_div),

            "diff_qbc_minus_qbc_diversity": diff_qbc_minus_div,
            "diff_qbc_minus_qbc_diversity_abs": abs(diff_qbc_minus_div),

            "improvement_qbc_diversity_vs_qbc_percent": improvement_div_vs_qbc,
        })

    return pd.DataFrame(rows)


def print_40pct_comparisons(comp_df: pd.DataFrame) -> None:
    """
    Prints all calculated information clearly.
    """
    for _, r in comp_df.iterrows():
        print("\n" + "=" * 100)
        print(
            f"[40% RMSE SUMMARY] "
            f"dataset={r['dataset'].upper()} | "
            f"target={r['target_label']} | "
            f"metric={r['metric']}"
        )
        print("-" * 100)

        print(f"Random sampling mean RMSE @ 40%:      {r['random_mean_rmse_40']:.8f} ({r['random_extraction']})")
        print(f"QBC mean RMSE @ 40%:                  {r['qbc_mean_rmse_40']:.8f} ({r['qbc_extraction']})")
        print(f"QBC + Diversity mean RMSE @ 40%:      {r['qbc_diversity_mean_rmse_40']:.8f} ({r['qbc_diversity_extraction']})")

        print("\nPairwise signed differences at 40%:")
        print(
            f"Random - QBC:                         {r['diff_random_minus_qbc']:.8f} "
            f"| abs={r['diff_random_minus_qbc_abs']:.8f}"
        )
        print(
            f"Random - QBC+Diversity:               {r['diff_random_minus_qbc_diversity']:.8f} "
            f"| abs={r['diff_random_minus_qbc_diversity_abs']:.8f}"
        )
        print(
            f"QBC - QBC+Diversity:                  {r['diff_qbc_minus_qbc_diversity']:.8f} "
            f"| abs={r['diff_qbc_minus_qbc_diversity_abs']:.8f}"
        )

        print("\nImprovement of QBC+Diversity relative to non-diversity QBC at 40%:")
        print(
            f"100 * (1 - RMSE_div / RMSE_qbc) = "
            f"{r['improvement_qbc_diversity_vs_qbc_percent']:.4f}%"
        )


# ============================================================
# Plotting: 3 metrics in ONE single image per target
# ============================================================
def plot_all_metrics_for_dataset_target(
    dataset: str,
    target: str,
    agg_by_run: dict,
    dataset_out_dir: str,
) -> None:
    metric_cols = list(metrics_internal.keys())
    panel_labels = ["a)", "b)", "c)"]

    fig, axes = plt.subplots(1, 3, figsize=(21, 10), sharex=True)

    for i, (ax, metric_col) in enumerate(zip(axes, metric_cols)):
        for run_name, agg in agg_by_run.items():
            x_percent = agg["train_ratio"].to_numpy(dtype=float) * 100.0
            mean = agg[(metric_col, "mean")].to_numpy(dtype=float)
            std = agg[(metric_col, "std")].to_numpy(dtype=float)

            ax.plot(x_percent, mean, linewidth=2, label=run_name)
            ax.fill_between(x_percent, mean - std, mean + std, alpha=0.18)

        ax.axvline(
            TARGET_RATIO_PERCENT,
            color="black",
            linestyle="--",
            linewidth=1.2,
            alpha=0.6,
        )

        ax.set_xlabel("Training data (%)", fontsize=25)
        ax.set_ylabel(f"RMSE {metrics_internal[metric_col]}", fontsize=25)

        ax.set_title(
            f"{panel_labels[i]}",
            fontsize=25,
            fontweight="bold",
            loc="left"
        )

        xticks = np.arange(10, 101, 10)
        ax.set_xticks(xticks)
        ax.tick_params(axis="x", labelrotation=45, labelsize=20)
        ax.tick_params(axis="y", labelsize=20)

        ax.margins(x=0)
        ax.grid(alpha=0.3)

    fig.suptitle(
        f"{dataset.upper()} — Target: {target_label(target)}",
        fontsize=30,
        fontweight="bold",
        y=0.925,
    )

    handles, labels = axes[0].get_legend_handles_labels()

    fig.legend(
        handles,
        labels,
        loc="lower center",
        ncol=3,
        fontsize=25,
        frameon=False,
        bbox_to_anchor=(0.5, -0.02)
    )

    fig.tight_layout(rect=[0, 0.05, 1, 0.93])

    out_path = os.path.join(
        dataset_out_dir,
        f"{dataset}_target_{target}__all_rmse.png"
    )

    fig.savefig(out_path, dpi=100, bbox_inches="tight")
    plt.close(fig)


# ============================================================
# Main loop
# ============================================================
os.makedirs(OUT_DIR, exist_ok=True)
os.makedirs(TABLE_OUT_DIR, exist_ok=True)

all_mean_rmse_rows = []
all_comparison_rows = []

for dataset in DATASETS:
    dataset_out_dir = os.path.join(OUT_DIR, dataset)
    os.makedirs(dataset_out_dir, exist_ok=True)

    for target in TARGETS:
        runs = {
            "Random sampling": {
                "paths": [
                    (
                        seed,
                        os.path.join(
                            ROOT_RANDOM,
                            dataset,
                            "all",
                            str(seed),
                            random_csv_name(dataset, seed),
                        )
                    )
                    for seed in seeds
                ],
                "colmap": COLMAP_RANDOM_QBC,
            },

            f"QBC target: {target_label(target)}": {
                "paths": [
                    (
                        seed,
                        os.path.join(
                            ROOT_QBC,
                            dataset,
                            target,
                            str(seed),
                            qbc_csv_name(dataset, target),
                        )
                    )
                    for seed in seeds
                ],
                "colmap": COLMAP_RANDOM_QBC,
            },

            f"QBC+Diversity target: {target_label(target)}": {
                "paths": [
                    (
                        seed,
                        os.path.join(
                            ROOT_QBCDIV,
                            dataset,
                            target,
                            str(seed),
                            qbcdiv_csv_name(dataset, target, seed),
                        )
                    )
                    for seed in seeds
                ],
                "colmap": COLMAP_QBCDIV,
            },
        }

        agg_by_run = {}

        for run_name, info in runs.items():
            missing = [p for (_, p) in info["paths"] if not os.path.exists(p)]

            if missing:
                preview = "\n".join(missing[:8])
                raise FileNotFoundError(
                    f"[{dataset} | target={target}] "
                    f"Missing {len(missing)} file(s) for run '{run_name}'.\n"
                    f"First missing paths:\n{preview}"
                )

            data_long, agg = aggregate_over_seeds(info["paths"], info["colmap"])
            agg_by_run[run_name] = agg

        # Save mean RMSE at each train_ratio for this dataset-target.
        mean_df = flatten_mean_rmse_table(
            dataset=dataset,
            target=target,
            agg_by_run=agg_by_run,
        )

        mean_csv = os.path.join(
            TABLE_OUT_DIR,
            f"{dataset}_target_{target}_mean_rmse_by_train_ratio.csv",
        )
        mean_df.to_csv(mean_csv, index=False)
        all_mean_rmse_rows.append(mean_df)

        # Compute and print 40% comparisons.
        comp_df = compute_40pct_comparisons(
            dataset=dataset,
            target=target,
            agg_by_run=agg_by_run,
        )

        print_40pct_comparisons(comp_df)
        all_comparison_rows.append(comp_df)

        # Plot original figure.
        plot_all_metrics_for_dataset_target(
            dataset=dataset,
            target=target,
            agg_by_run=agg_by_run,
            dataset_out_dir=dataset_out_dir,
        )

# ============================================================
# Save global summary tables
# ============================================================
all_mean_rmse_df = pd.concat(all_mean_rmse_rows, ignore_index=True)
all_comparison_df = pd.concat(all_comparison_rows, ignore_index=True)

all_mean_rmse_csv = os.path.join(
    TABLE_OUT_DIR,
    "ALL_mean_rmse_by_train_ratio.csv",
)

all_comparison_csv = os.path.join(
    TABLE_OUT_DIR,
    "ALL_40pct_pairwise_rmse_differences_and_improvement_vs_qbc.csv",
)

all_mean_rmse_df.to_csv(all_mean_rmse_csv, index=False)
all_comparison_df.to_csv(all_comparison_csv, index=False)

print("\n" + "#" * 100)
print("[DONE]")
print(f"Saved combined plots to:\n{OUT_DIR}")
print(f"Saved mean RMSE-by-train-ratio table to:\n{all_mean_rmse_csv}")
print(f"Saved 40% comparison table to:\n{all_comparison_csv}")
print("#" * 100)
```

```python
import os
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

# ---------------------------------------------------------------------------
# JOINT PLOTS.
# ONE PLOT PER TARGETED PROPERTY AND POLICY.
#
# Additional calculations:
#   1. Mean RMSE across seeds at each train_ratio for each policy.
#   2. Pairwise RMSE differences at train_ratio = 0.40.
#   3. Improvement at 40% relative to non-diversity QBC.
#   4. At train_ratio = 1.00:
#      - mean RMSE across Random, QBC, and QBC+Diversity.
#      - difference of QBC from that policy mean.
#      - difference of QBC+Diversity from that policy mean.
#      - difference between those deviations.
# ---------------------------------------------------------------------------

# ============================================================
# CONFIG
# ============================================================
seeds = range(5)

DATASETS = ["jarvis18", "jarvis22", "mp18", "mp21"]
TARGETS = ["bandgap", "bulk_modulus", "e_form"]

TARGET_RATIO_40 = 0.40
TARGET_RATIO_100 = 1.00

TARGET_RATIO_40_PERCENT = TARGET_RATIO_40 * 100.0
TARGET_RATIO_100_PERCENT = TARGET_RATIO_100 * 100.0

# Where your CSVs live
ROOT_RANDOM = (
    r"C:\Users\rafae\OneDrive\Imágenes\Documentos\PhDToronto\OLD_FILES\PhD Toronto"
    r"\DatasetGeneration\Better_Acqsition\NewAlg_DFT2\al\random"
)

ROOT_QBC = (
    r"C:\Users\rafae\OneDrive\Imágenes\Documentos\PhDToronto\OLD_FILES\PhD Toronto"
    r"\DatasetGeneration\Better_Acqsition\NewAlg_DFT2\al\QBC"
)

ROOT_QBCDIV = (
    r"C:\Users\rafae\OneDrive\Imágenes\Documentos\PhDToronto\OLD_FILES\PhD Toronto"
    r"\DatasetGeneration\Better_Acqsition\NewAlg_DFT2\al\QBC_DIVERSITY"
)

# Output base
OUT_DIR = (
    r"C:\Users\rafae\OneDrive\Imágenes\Documentos\PhDToronto\OLD_FILES\PhD Toronto"
    r"\DatasetGeneration\Better_Acqsition\Plots_DFT"
)

# Tables output
TABLE_OUT_DIR = os.path.join(OUT_DIR, "RMSE_40pct_100pct_tables")
os.makedirs(TABLE_OUT_DIR, exist_ok=True)

# ============================================================
# Metrics
# ============================================================
metrics_internal = {
    "rmse_bandgap": "Bandgap (eV)",
    "rmse_bulk_modulus": "Bulk Modulus (GPa)",
    "rmse_eform": r"Formation Energy (eV atom$^{-1}$)",
}

metric_short_names = {
    "rmse_bandgap": "Bandgap",
    "rmse_bulk_modulus": "Bulk Modulus",
    "rmse_eform": "Formation Energy",
}

# Column maps
COLMAP_RANDOM_QBC = {
    "rmse_bandgap": "rmse_bandgap",
    "rmse_bulk_modulus": "rmse_bulk_modulus",
    "rmse_eform": "rmse_eform",
}

COLMAP_QBCDIV = {
    "rmse_bandgap": "rmse_bandgap",
    "rmse_bulk_modulus": "rmse_bulk_modulus",
    "rmse_eform": "rmse_e_form",
}

# ============================================================
# Filename conventions
# ============================================================
def random_csv_name(dataset: str, seed: int) -> str:
    return f"{dataset}_all_random_rf{seed}.csv"


def qbc_csv_name(dataset: str, target: str) -> str:
    return f"{dataset}_{target}_QBC_rf.csv"


def qbcdiv_csv_name(dataset: str, target: str, seed: int) -> str:
    return f"{dataset}_{target}_QBCdiversity_rf{seed}.csv"


def target_label(target: str) -> str:
    if target == "bandgap":
        return "Bandgap"
    if target == "bulk_modulus":
        return "Bulk Modulus"
    if target == "e_form":
        return "Formation Energy"
    return target


# ============================================================
# IO helpers
# ============================================================
def read_run_csv(path: str, colmap: dict) -> pd.DataFrame:
    df = pd.read_csv(path)
    df.columns = [str(c).strip() for c in df.columns]

    if "train_ratio" not in df.columns:
        raise ValueError(
            f"Missing 'train_ratio' in {path}. "
            f"Available columns: {list(df.columns)}"
        )

    for internal_name, file_col in colmap.items():
        if file_col not in df.columns:
            raise ValueError(
                f"Missing '{file_col}' mapped to {internal_name} in {path}. "
                f"Available columns: {list(df.columns)}"
            )

    out = pd.DataFrame({
        "train_ratio": pd.to_numeric(df["train_ratio"], errors="coerce")
    })

    for internal_name, file_col in colmap.items():
        out[internal_name] = pd.to_numeric(df[file_col], errors="coerce")

    return out


def aggregate_over_seeds(paths: list, colmap: dict) -> tuple[pd.DataFrame, pd.DataFrame]:
    """
    Returns:
        data_long:
            One row per seed and train_ratio.
        agg:
            Mean and std RMSE at each train_ratio across seeds.
    """
    per_seed = []

    for seed, p in paths:
        tmp = read_run_csv(p, colmap)
        tmp["seed"] = seed
        tmp["path"] = p
        per_seed.append(tmp)

    data_long = pd.concat(per_seed, ignore_index=True)

    agg = (
        data_long
        .groupby("train_ratio")[list(metrics_internal.keys())]
        .agg(["mean", "std"])
        .reset_index()
        .sort_values("train_ratio")
    )

    return data_long, agg


# ============================================================
# Ratio extraction helpers
# ============================================================
def get_mean_rmse_at_ratio(
    agg: pd.DataFrame,
    metric_col: str,
    target_ratio: float,
    strict_exact: bool = False,
) -> tuple[float, str]:
    """
    Extract the mean RMSE at target_ratio.

    If strict_exact=True, this raises an error when target_ratio is absent.
    If strict_exact=False, it interpolates if target_ratio is not exactly present.

    Returns:
        value, mode
        mode is "exact" or "interpolated".
    """
    x = agg["train_ratio"].to_numpy(dtype=float)
    y = agg[(metric_col, "mean")].to_numpy(dtype=float)

    mask = np.isfinite(x) & np.isfinite(y)
    x = x[mask]
    y = y[mask]

    if x.size < 2:
        raise ValueError(f"Not enough finite points for {metric_col}.")

    order = np.argsort(x)
    x = x[order]
    y = y[order]

    exact_mask = np.isclose(x, target_ratio, rtol=1e-10, atol=1e-12)

    if exact_mask.any():
        return float(y[exact_mask][0]), "exact"

    if strict_exact:
        raise ValueError(
            f"train_ratio={target_ratio} was not found exactly. "
            f"Available train_ratio values: {x}"
        )

    if target_ratio < x.min() or target_ratio > x.max():
        raise ValueError(
            f"target_ratio={target_ratio} is outside available range "
            f"[{x.min()}, {x.max()}]."
        )

    return float(np.interp(target_ratio, x, y)), "interpolated"


def flatten_mean_rmse_table(
    dataset: str,
    target: str,
    agg_by_run: dict,
) -> pd.DataFrame:
    """
    Saves the mean RMSE at every train_ratio for every policy.
    """
    rows = []

    for run_name, agg in agg_by_run.items():
        x = agg["train_ratio"].to_numpy(dtype=float)

        for metric_col in metrics_internal:
            mean_vals = agg[(metric_col, "mean")].to_numpy(dtype=float)
            std_vals = agg[(metric_col, "std")].to_numpy(dtype=float)

            for tr, mean_rmse, std_rmse in zip(x, mean_vals, std_vals):
                rows.append({
                    "dataset": dataset,
                    "target": target,
                    "target_label": target_label(target),
                    "policy": run_name,
                    "metric": metric_short_names[metric_col],
                    "metric_col": metric_col,
                    "train_ratio": float(tr),
                    "training_percent": float(tr * 100.0),
                    "mean_rmse": float(mean_rmse),
                    "std_rmse": float(std_rmse),
                })

    return pd.DataFrame(rows)


# ============================================================
# 40% and 100% comparison calculations
# ============================================================
def compute_ratio_comparisons(
    dataset: str,
    target: str,
    agg_by_run: dict,
) -> pd.DataFrame:
    """
    Computes:

    At 40%:
        - mean RMSE for Random, QBC, QBC+Diversity.
        - pairwise signed and absolute differences.
        - improvement of QBC+Diversity relative to QBC.

    At 100%:
        - mean RMSE for Random, QBC, QBC+Diversity.
        - mean RMSE over the three policies.
        - deviation of each policy from that mean.
        - percent deviation of each policy from that mean.
        - difference between QBC and QBC+Diversity deviations.

    Positive improvement at 40% means QBC+Diversity has lower RMSE than QBC.

    For 100% deviations:
        policy_minus_mean < 0 means that policy has lower RMSE than the
        average over Random, QBC, and QBC+Diversity.
    """
    run_random = "Random sampling"
    run_qbc = f"QBC target: {target_label(target)}"
    run_div = f"QBC+Diversity target: {target_label(target)}"

    required = [run_random, run_qbc, run_div]
    for r in required:
        if r not in agg_by_run:
            raise KeyError(f"Missing run '{r}' for {dataset}, target={target}")

    rows = []

    for metric_col in metrics_internal:
        metric_name = metric_short_names[metric_col]

        # ====================================================
        # 40% calculations
        # ====================================================
        random_rmse_40, random_mode_40 = get_mean_rmse_at_ratio(
            agg_by_run[run_random],
            metric_col,
            target_ratio=TARGET_RATIO_40,
            strict_exact=False,
        )

        qbc_rmse_40, qbc_mode_40 = get_mean_rmse_at_ratio(
            agg_by_run[run_qbc],
            metric_col,
            target_ratio=TARGET_RATIO_40,
            strict_exact=False,
        )

        div_rmse_40, div_mode_40 = get_mean_rmse_at_ratio(
            agg_by_run[run_div],
            metric_col,
            target_ratio=TARGET_RATIO_40,
            strict_exact=False,
        )

        diff_random_minus_qbc_40 = random_rmse_40 - qbc_rmse_40
        diff_random_minus_div_40 = random_rmse_40 - div_rmse_40
        diff_qbc_minus_div_40 = qbc_rmse_40 - div_rmse_40

        improvement_div_vs_qbc_40 = 100.0 * (1.0 - div_rmse_40 / qbc_rmse_40)

        # ====================================================
        # 100% calculations
        # ====================================================
        random_rmse_100, random_mode_100 = get_mean_rmse_at_ratio(
            agg_by_run[run_random],
            metric_col,
            target_ratio=TARGET_RATIO_100,
            strict_exact=False,
        )

        qbc_rmse_100, qbc_mode_100 = get_mean_rmse_at_ratio(
            agg_by_run[run_qbc],
            metric_col,
            target_ratio=TARGET_RATIO_100,
            strict_exact=False,
        )

        div_rmse_100, div_mode_100 = get_mean_rmse_at_ratio(
            agg_by_run[run_div],
            metric_col,
            target_ratio=TARGET_RATIO_100,
            strict_exact=False,
        )

        mean_policy_rmse_100 = np.mean([
            random_rmse_100,
            qbc_rmse_100,
            div_rmse_100,
        ])

        random_minus_policy_mean_100 = random_rmse_100 - mean_policy_rmse_100
        qbc_minus_policy_mean_100 = qbc_rmse_100 - mean_policy_rmse_100
        div_minus_policy_mean_100 = div_rmse_100 - mean_policy_rmse_100

        random_minus_policy_mean_100_percent = (
            100.0 * random_minus_policy_mean_100 / mean_policy_rmse_100
        )

        qbc_minus_policy_mean_100_percent = (
            100.0 * qbc_minus_policy_mean_100 / mean_policy_rmse_100
        )

        div_minus_policy_mean_100_percent = (
            100.0 * div_minus_policy_mean_100 / mean_policy_rmse_100
        )

        # Because both are deviations from the same mean, this is also:
        # qbc_rmse_100 - div_rmse_100
        diff_qbc_deviation_minus_div_deviation_100 = (
            qbc_minus_policy_mean_100 - div_minus_policy_mean_100
        )

        diff_qbc_minus_div_100 = qbc_rmse_100 - div_rmse_100
        improvement_div_vs_qbc_100 = 100.0 * (1.0 - div_rmse_100 / qbc_rmse_100)

        rows.append({
            "dataset": dataset,
            "target": target,
            "target_label": target_label(target),
            "metric": metric_name,
            "metric_col": metric_col,

            # ====================================================
            # 40% RMSE values
            # ====================================================
            "train_ratio_40": TARGET_RATIO_40,
            "training_percent_40": TARGET_RATIO_40_PERCENT,

            "random_mean_rmse_40": random_rmse_40,
            "qbc_mean_rmse_40": qbc_rmse_40,
            "qbc_diversity_mean_rmse_40": div_rmse_40,

            "random_extraction_40": random_mode_40,
            "qbc_extraction_40": qbc_mode_40,
            "qbc_diversity_extraction_40": div_mode_40,

            "diff_random_minus_qbc_40": diff_random_minus_qbc_40,
            "diff_random_minus_qbc_40_abs": abs(diff_random_minus_qbc_40),

            "diff_random_minus_qbc_diversity_40": diff_random_minus_div_40,
            "diff_random_minus_qbc_diversity_40_abs": abs(diff_random_minus_div_40),

            "diff_qbc_minus_qbc_diversity_40": diff_qbc_minus_div_40,
            "diff_qbc_minus_qbc_diversity_40_abs": abs(diff_qbc_minus_div_40),

            "improvement_qbc_diversity_vs_qbc_percent_40": improvement_div_vs_qbc_40,

            # ====================================================
            # 100% RMSE values
            # ====================================================
            "train_ratio_100": TARGET_RATIO_100,
            "training_percent_100": TARGET_RATIO_100_PERCENT,

            "random_mean_rmse_100": random_rmse_100,
            "qbc_mean_rmse_100": qbc_rmse_100,
            "qbc_diversity_mean_rmse_100": div_rmse_100,

            "random_extraction_100": random_mode_100,
            "qbc_extraction_100": qbc_mode_100,
            "qbc_diversity_extraction_100": div_mode_100,

            "mean_rmse_over_random_qbc_qbcdiv_100": mean_policy_rmse_100,

            "random_minus_policy_mean_100": random_minus_policy_mean_100,
            "qbc_minus_policy_mean_100": qbc_minus_policy_mean_100,
            "qbc_diversity_minus_policy_mean_100": div_minus_policy_mean_100,

            "random_minus_policy_mean_100_percent": random_minus_policy_mean_100_percent,
            "qbc_minus_policy_mean_100_percent": qbc_minus_policy_mean_100_percent,
            "qbc_diversity_minus_policy_mean_100_percent": div_minus_policy_mean_100_percent,

            "diff_qbc_deviation_minus_qbc_diversity_deviation_100": (
                diff_qbc_deviation_minus_div_deviation_100
            ),

            "diff_qbc_minus_qbc_diversity_100": diff_qbc_minus_div_100,
            "diff_qbc_minus_qbc_diversity_100_abs": abs(diff_qbc_minus_div_100),

            "improvement_qbc_diversity_vs_qbc_percent_100": improvement_div_vs_qbc_100,
        })

    return pd.DataFrame(rows)


def print_ratio_comparisons(comp_df: pd.DataFrame) -> None:
    """
    Prints all calculated information clearly.
    """
    for _, r in comp_df.iterrows():
        print("\n" + "=" * 120)
        print(
            f"[RMSE SUMMARY] "
            f"dataset={str(r['dataset']).upper()} | "
            f"target={r['target_label']} | "
            f"metric={r['metric']}"
        )
        print("-" * 120)

        # ====================================================
        # 40% section
        # ====================================================
        print("[40% training ratio]")
        print(
            f"Random sampling mean RMSE @ 40%:      "
            f"{r['random_mean_rmse_40']:.8f} ({r['random_extraction_40']})"
        )
        print(
            f"QBC mean RMSE @ 40%:                  "
            f"{r['qbc_mean_rmse_40']:.8f} ({r['qbc_extraction_40']})"
        )
        print(
            f"QBC + Diversity mean RMSE @ 40%:      "
            f"{r['qbc_diversity_mean_rmse_40']:.8f} ({r['qbc_diversity_extraction_40']})"
        )

        print("\nPairwise signed differences at 40%:")
        print(
            f"Random - QBC:                         "
            f"{r['diff_random_minus_qbc_40']:.8f} "
            f"| abs={r['diff_random_minus_qbc_40_abs']:.8f}"
        )
        print(
            f"Random - QBC+Diversity:               "
            f"{r['diff_random_minus_qbc_diversity_40']:.8f} "
            f"| abs={r['diff_random_minus_qbc_diversity_40_abs']:.8f}"
        )
        print(
            f"QBC - QBC+Diversity:                  "
            f"{r['diff_qbc_minus_qbc_diversity_40']:.8f} "
            f"| abs={r['diff_qbc_minus_qbc_diversity_40_abs']:.8f}"
        )

        print("\nImprovement of QBC+Diversity relative to non-diversity QBC at 40%:")
        print(
            f"100 * (1 - RMSE_div / RMSE_qbc) = "
            f"{r['improvement_qbc_diversity_vs_qbc_percent_40']:.4f}%"
        )

        # ====================================================
        # 100% section
        # ====================================================
        print("\n[100% training ratio]")
        print(
            f"Random sampling mean RMSE @ 100%:     "
            f"{r['random_mean_rmse_100']:.8f} ({r['random_extraction_100']})"
        )
        print(
            f"QBC mean RMSE @ 100%:                 "
            f"{r['qbc_mean_rmse_100']:.8f} ({r['qbc_extraction_100']})"
        )
        print(
            f"QBC + Diversity mean RMSE @ 100%:     "
            f"{r['qbc_diversity_mean_rmse_100']:.8f} ({r['qbc_diversity_extraction_100']})"
        )

        print(
            f"\nMean RMSE over Random, QBC, and QBC+Diversity @ 100%: "
            f"{r['mean_rmse_over_random_qbc_qbcdiv_100']:.8f}"
        )

        print("\nDifference from 100% policy mean:")
        print(
            f"Random - policy mean:                 "
            f"{r['random_minus_policy_mean_100']:.8f} "
            f"({r['random_minus_policy_mean_100_percent']:.4f}%)"
        )
        print(
            f"QBC - policy mean:                    "
            f"{r['qbc_minus_policy_mean_100']:.8f} "
            f"({r['qbc_minus_policy_mean_100_percent']:.4f}%)"
        )
        print(
            f"QBC+Diversity - policy mean:          "
            f"{r['qbc_diversity_minus_policy_mean_100']:.8f} "
            f"({r['qbc_diversity_minus_policy_mean_100_percent']:.4f}%)"
        )

        print("\nDifference between QBC and QBC+Diversity deviations from the 100% policy mean:")
        print(
            f"(QBC - mean) - (QBC+Diversity - mean) = "
            f"{r['diff_qbc_deviation_minus_qbc_diversity_deviation_100']:.8f}"
        )

        print("\nDirect QBC vs QBC+Diversity comparison at 100%:")
        print(
            f"QBC - QBC+Diversity:                  "
            f"{r['diff_qbc_minus_qbc_diversity_100']:.8f} "
            f"| abs={r['diff_qbc_minus_qbc_diversity_100_abs']:.8f}"
        )
        print(
            f"100 * (1 - RMSE_div / RMSE_qbc) = "
            f"{r['improvement_qbc_diversity_vs_qbc_percent_100']:.4f}%"
        )


# ============================================================
# Plotting: 3 metrics in ONE single image per target
# ============================================================
def plot_all_metrics_for_dataset_target(
    dataset: str,
    target: str,
    agg_by_run: dict,
    dataset_out_dir: str,
) -> None:
    metric_cols = list(metrics_internal.keys())
    panel_labels = ["a)", "b)", "c)"]

    fig, axes = plt.subplots(1, 3, figsize=(21, 10), sharex=True)

    for i, (ax, metric_col) in enumerate(zip(axes, metric_cols)):
        for run_name, agg in agg_by_run.items():
            x_percent = agg["train_ratio"].to_numpy(dtype=float) * 100.0
            mean = agg[(metric_col, "mean")].to_numpy(dtype=float)
            std = agg[(metric_col, "std")].to_numpy(dtype=float)

            ax.plot(x_percent, mean, linewidth=2, label=run_name)
            ax.fill_between(x_percent, mean - std, mean + std, alpha=0.18)

        ax.axvline(
            TARGET_RATIO_40_PERCENT,
            color="black",
            linestyle="--",
            linewidth=1.2,
            alpha=0.6,
        )

        ax.axvline(
            TARGET_RATIO_100_PERCENT,
            color="black",
            linestyle=":",
            linewidth=1.2,
            alpha=0.6,
        )

        ax.set_xlabel("Training data (%)", fontsize=25)
        ax.set_ylabel(f"RMSE {metrics_internal[metric_col]}", fontsize=25)

        ax.set_title(
            f"{panel_labels[i]}",
            fontsize=25,
            fontweight="bold",
            loc="left"
        )

        xticks = np.arange(10, 101, 10)
        ax.set_xticks(xticks)
        ax.tick_params(axis="x", labelrotation=45, labelsize=20)
        ax.tick_params(axis="y", labelsize=20)

        ax.margins(x=0)
        ax.grid(alpha=0.3)

    fig.suptitle(
        f"{dataset.upper()} — Target: {target_label(target)}",
        fontsize=30,
        fontweight="bold",
        y=0.925,
    )

    handles, labels = axes[0].get_legend_handles_labels()

    fig.legend(
        handles,
        labels,
        loc="lower center",
        ncol=3,
        fontsize=25,
        frameon=False,
        bbox_to_anchor=(0.5, -0.02)
    )

    fig.tight_layout(rect=[0, 0.05, 1, 0.93])

    out_path = os.path.join(
        dataset_out_dir,
        f"{dataset}_target_{target}__all_rmse.png"
    )

    fig.savefig(out_path, dpi=100, bbox_inches="tight")
    plt.close(fig)


# ============================================================
# MAIN LOOP
# ============================================================
os.makedirs(OUT_DIR, exist_ok=True)
os.makedirs(TABLE_OUT_DIR, exist_ok=True)

all_mean_rmse_rows = []
all_comparison_rows = []

for dataset in DATASETS:
    dataset_out_dir = os.path.join(OUT_DIR, dataset)
    os.makedirs(dataset_out_dir, exist_ok=True)

    for target in TARGETS:
        runs = {
            "Random sampling": {
                "paths": [
                    (
                        seed,
                        os.path.join(
                            ROOT_RANDOM,
                            dataset,
                            "all",
                            str(seed),
                            random_csv_name(dataset, seed),
                        )
                    )
                    for seed in seeds
                ],
                "colmap": COLMAP_RANDOM_QBC,
            },

            f"QBC target: {target_label(target)}": {
                "paths": [
                    (
                        seed,
                        os.path.join(
                            ROOT_QBC,
                            dataset,
                            target,
                            str(seed),
                            qbc_csv_name(dataset, target),
                        )
                    )
                    for seed in seeds
                ],
                "colmap": COLMAP_RANDOM_QBC,
            },

            f"QBC+Diversity target: {target_label(target)}": {
                "paths": [
                    (
                        seed,
                        os.path.join(
                            ROOT_QBCDIV,
                            dataset,
                            target,
                            str(seed),
                            qbcdiv_csv_name(dataset, target, seed),
                        )
                    )
                    for seed in seeds
                ],
                "colmap": COLMAP_QBCDIV,
            },
        }

        agg_by_run = {}

        for run_name, info in runs.items():
            missing = [p for (_, p) in info["paths"] if not os.path.exists(p)]

            if missing:
                preview = "\n".join(missing[:8])
                raise FileNotFoundError(
                    f"[{dataset} | target={target}] "
                    f"Missing {len(missing)} file(s) for run '{run_name}'.\n"
                    f"First missing paths:\n{preview}"
                )

            data_long, agg = aggregate_over_seeds(info["paths"], info["colmap"])
            agg_by_run[run_name] = agg

        # Save mean RMSE at each train_ratio for this dataset-target.
        mean_df = flatten_mean_rmse_table(
            dataset=dataset,
            target=target,
            agg_by_run=agg_by_run,
        )

        mean_csv = os.path.join(
            TABLE_OUT_DIR,
            f"{dataset}_target_{target}_mean_rmse_by_train_ratio.csv",
        )

        mean_df.to_csv(mean_csv, index=False)
        all_mean_rmse_rows.append(mean_df)

        # Compute and print 40% and 100% comparisons.
        comp_df = compute_ratio_comparisons(
            dataset=dataset,
            target=target,
            agg_by_run=agg_by_run,
        )

        print_ratio_comparisons(comp_df)
        all_comparison_rows.append(comp_df)

        # Plot original figure.
        plot_all_metrics_for_dataset_target(
            dataset=dataset,
            target=target,
            agg_by_run=agg_by_run,
            dataset_out_dir=dataset_out_dir,
        )

# ============================================================
# Save global summary tables
# ============================================================
all_mean_rmse_df = pd.concat(all_mean_rmse_rows, ignore_index=True)
all_comparison_df = pd.concat(all_comparison_rows, ignore_index=True)

all_mean_rmse_csv = os.path.join(
    TABLE_OUT_DIR,
    "ALL_mean_rmse_by_train_ratio.csv",
)

all_comparison_csv = os.path.join(
    TABLE_OUT_DIR,
    "ALL_40pct_and_100pct_rmse_differences_improvement_vs_qbc.csv",
)

all_mean_rmse_df.to_csv(all_mean_rmse_csv, index=False)
all_comparison_df.to_csv(all_comparison_csv, index=False)

print("\n" + "#" * 120)
print("[DONE]")
print(f"Saved combined plots to:\n{OUT_DIR}")
print(f"Saved mean RMSE-by-train-ratio table to:\n{all_mean_rmse_csv}")
print(f"Saved 40% and 100% comparison table to:\n{all_comparison_csv}")
print("#" * 120)
```

```python
import os
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

# ---------------------------------------------------------------------------
# JOINT PLOTS.
# ONE PLOT PER TARGETED PROPERTY AND POLICY.
#
# Additional calculations:
#   1. Mean RMSE across seeds at each train_ratio for each policy.
#   2. Pairwise RMSE differences at train_ratio = 0.40.
#   3. Improvement at 40% of QBC+Diversity relative to non-diversity QBC.
#   4. Full-training reference:
#        mean RMSE at 100% over Random, QBC, and QBC+Diversity.
#   5. Improvement of QBC at 40% relative to that 100% full-training policy mean.
#   6. Improvement of QBC+Diversity at 40% relative to that 100% full-training policy mean.
# ---------------------------------------------------------------------------

# ============================================================
# CONFIG
# ============================================================
seeds = range(5)

DATASETS = ["jarvis18", "jarvis22", "mp18", "mp21"]
TARGETS = ["bandgap", "bulk_modulus", "e_form"]

TARGET_RATIO_40 = 0.40
TARGET_RATIO_100 = 1.00

TARGET_RATIO_40_PERCENT = TARGET_RATIO_40 * 100.0
TARGET_RATIO_100_PERCENT = TARGET_RATIO_100 * 100.0

# Where your CSVs live
ROOT_RANDOM = (
    r"C:\Users\rafae\OneDrive\Imágenes\Documentos\PhDToronto\OLD_FILES\PhD Toronto"
    r"\DatasetGeneration\Better_Acqsition\NewAlg_DFT2\al\random"
)

ROOT_QBC = (
    r"C:\Users\rafae\OneDrive\Imágenes\Documentos\PhDToronto\OLD_FILES\PhD Toronto"
    r"\DatasetGeneration\Better_Acqsition\NewAlg_DFT2\al\QBC"
)

ROOT_QBCDIV = (
    r"C:\Users\rafae\OneDrive\Imágenes\Documentos\PhDToronto\OLD_FILES\PhD Toronto"
    r"\DatasetGeneration\Better_Acqsition\NewAlg_DFT2\al\QBC_DIVERSITY"
)

# Output base
OUT_DIR = (
    r"C:\Users\rafae\OneDrive\Imágenes\Documentos\PhDToronto\OLD_FILES\PhD Toronto"
    r"\DatasetGeneration\Better_Acqsition\Plots_DFT"
)

# Tables output
TABLE_OUT_DIR = os.path.join(OUT_DIR, "RMSE_40pct_full_training_reference_tables")
os.makedirs(TABLE_OUT_DIR, exist_ok=True)

# ============================================================
# Metrics
# ============================================================
metrics_internal = {
    "rmse_bandgap": "Bandgap (eV)",
    "rmse_bulk_modulus": "Bulk Modulus (GPa)",
    "rmse_eform": r"Formation Energy (eV atom$^{-1}$)",
}

metric_short_names = {
    "rmse_bandgap": "Bandgap",
    "rmse_bulk_modulus": "Bulk Modulus",
    "rmse_eform": "Formation Energy",
}

# Column maps
COLMAP_RANDOM_QBC = {
    "rmse_bandgap": "rmse_bandgap",
    "rmse_bulk_modulus": "rmse_bulk_modulus",
    "rmse_eform": "rmse_eform",
}

COLMAP_QBCDIV = {
    "rmse_bandgap": "rmse_bandgap",
    "rmse_bulk_modulus": "rmse_bulk_modulus",
    "rmse_eform": "rmse_e_form",
}

# ============================================================
# Filename conventions
# ============================================================
def random_csv_name(dataset: str, seed: int) -> str:
    return f"{dataset}_all_random_rf{seed}.csv"


def qbc_csv_name(dataset: str, target: str) -> str:
    return f"{dataset}_{target}_QBC_rf.csv"


def qbcdiv_csv_name(dataset: str, target: str, seed: int) -> str:
    return f"{dataset}_{target}_QBCdiversity_rf{seed}.csv"


def target_label(target: str) -> str:
    if target == "bandgap":
        return "Bandgap"
    if target == "bulk_modulus":
        return "Bulk Modulus"
    if target == "e_form":
        return "Formation Energy"
    return target


# ============================================================
# IO helpers
# ============================================================
def read_run_csv(path: str, colmap: dict) -> pd.DataFrame:
    df = pd.read_csv(path)
    df.columns = [str(c).strip() for c in df.columns]

    if "train_ratio" not in df.columns:
        raise ValueError(
            f"Missing 'train_ratio' in {path}. "
            f"Available columns: {list(df.columns)}"
        )

    for internal_name, file_col in colmap.items():
        if file_col not in df.columns:
            raise ValueError(
                f"Missing '{file_col}' mapped to {internal_name} in {path}. "
                f"Available columns: {list(df.columns)}"
            )

    out = pd.DataFrame({
        "train_ratio": pd.to_numeric(df["train_ratio"], errors="coerce")
    })

    for internal_name, file_col in colmap.items():
        out[internal_name] = pd.to_numeric(df[file_col], errors="coerce")

    return out


def aggregate_over_seeds(paths: list, colmap: dict) -> tuple[pd.DataFrame, pd.DataFrame]:
    """
    Returns:
        data_long:
            One row per seed and train_ratio.
        agg:
            Mean and std RMSE at each train_ratio across seeds.
    """
    per_seed = []

    for seed, p in paths:
        tmp = read_run_csv(p, colmap)
        tmp["seed"] = seed
        tmp["path"] = p
        per_seed.append(tmp)

    data_long = pd.concat(per_seed, ignore_index=True)

    agg = (
        data_long
        .groupby("train_ratio")[list(metrics_internal.keys())]
        .agg(["mean", "std"])
        .reset_index()
        .sort_values("train_ratio")
    )

    return data_long, agg


# ============================================================
# Ratio extraction helpers
# ============================================================
def get_mean_rmse_at_ratio(
    agg: pd.DataFrame,
    metric_col: str,
    target_ratio: float,
    strict_exact: bool = False,
) -> tuple[float, str]:
    """
    Extract the mean RMSE at target_ratio.

    If strict_exact=True, this raises an error when target_ratio is absent.
    If strict_exact=False, it interpolates if target_ratio is not exactly present.

    Returns:
        value, mode
        mode is "exact" or "interpolated".
    """
    x = agg["train_ratio"].to_numpy(dtype=float)
    y = agg[(metric_col, "mean")].to_numpy(dtype=float)

    mask = np.isfinite(x) & np.isfinite(y)
    x = x[mask]
    y = y[mask]

    if x.size < 2:
        raise ValueError(f"Not enough finite points for {metric_col}.")

    order = np.argsort(x)
    x = x[order]
    y = y[order]

    exact_mask = np.isclose(x, target_ratio, rtol=1e-10, atol=1e-12)

    if exact_mask.any():
        return float(y[exact_mask][0]), "exact"

    if strict_exact:
        raise ValueError(
            f"train_ratio={target_ratio} was not found exactly. "
            f"Available train_ratio values: {x}"
        )

    if target_ratio < x.min() or target_ratio > x.max():
        raise ValueError(
            f"target_ratio={target_ratio} is outside available range "
            f"[{x.min()}, {x.max()}]."
        )

    return float(np.interp(target_ratio, x, y)), "interpolated"


def flatten_mean_rmse_table(
    dataset: str,
    target: str,
    agg_by_run: dict,
) -> pd.DataFrame:
    """
    Saves the mean RMSE at every train_ratio for every policy.
    """
    rows = []

    for run_name, agg in agg_by_run.items():
        x = agg["train_ratio"].to_numpy(dtype=float)

        for metric_col in metrics_internal:
            mean_vals = agg[(metric_col, "mean")].to_numpy(dtype=float)
            std_vals = agg[(metric_col, "std")].to_numpy(dtype=float)

            for tr, mean_rmse, std_rmse in zip(x, mean_vals, std_vals):
                rows.append({
                    "dataset": dataset,
                    "target": target,
                    "target_label": target_label(target),
                    "policy": run_name,
                    "metric": metric_short_names[metric_col],
                    "metric_col": metric_col,
                    "train_ratio": float(tr),
                    "training_percent": float(tr * 100.0),
                    "mean_rmse": float(mean_rmse),
                    "std_rmse": float(std_rmse),
                })

    return pd.DataFrame(rows)


# ============================================================
# 40% and full-training reference calculations
# ============================================================
def compute_ratio_comparisons(
    dataset: str,
    target: str,
    agg_by_run: dict,
) -> pd.DataFrame:
    """
    Computes:

    At 40%:
        - mean RMSE for Random, QBC, QBC+Diversity.
        - pairwise signed and absolute differences.
        - improvement of QBC+Diversity relative to QBC.

    At 100%:
        - mean RMSE for Random, QBC, QBC+Diversity.
        - full-training policy mean RMSE:
              mean(Random_100, QBC_100, QBCDiv_100)

    Requested new calculations:
        - improvement of QBC at 40% relative to the 100% full-training policy mean.
        - improvement of QBC+Diversity at 40% relative to the 100% full-training policy mean.

    Formulas:
        improvement_qbc_40_vs_full_avg_100 =
            100 * (1 - RMSE_QBC_40 / full_training_policy_mean_100)

        improvement_qbcdiv_40_vs_full_avg_100 =
            100 * (1 - RMSE_QBCDiv_40 / full_training_policy_mean_100)
    """
    run_random = "Random sampling"
    run_qbc = f"QBC target: {target_label(target)}"
    run_div = f"QBC+Diversity target: {target_label(target)}"

    required = [run_random, run_qbc, run_div]
    for r in required:
        if r not in agg_by_run:
            raise KeyError(f"Missing run '{r}' for {dataset}, target={target}")

    rows = []

    for metric_col in metrics_internal:
        metric_name = metric_short_names[metric_col]

        # ====================================================
        # 40% mean RMSE values
        # ====================================================
        random_rmse_40, random_mode_40 = get_mean_rmse_at_ratio(
            agg_by_run[run_random],
            metric_col,
            target_ratio=TARGET_RATIO_40,
            strict_exact=False,
        )

        qbc_rmse_40, qbc_mode_40 = get_mean_rmse_at_ratio(
            agg_by_run[run_qbc],
            metric_col,
            target_ratio=TARGET_RATIO_40,
            strict_exact=False,
        )

        div_rmse_40, div_mode_40 = get_mean_rmse_at_ratio(
            agg_by_run[run_div],
            metric_col,
            target_ratio=TARGET_RATIO_40,
            strict_exact=False,
        )

        # ====================================================
        # 40% pairwise differences
        # Positive Random - QBC means QBC has lower RMSE than Random.
        # Positive QBC - QBCDiv means QBCDiv has lower RMSE than QBC.
        # ====================================================
        diff_random_minus_qbc_40 = random_rmse_40 - qbc_rmse_40
        diff_random_minus_div_40 = random_rmse_40 - div_rmse_40
        diff_qbc_minus_div_40 = qbc_rmse_40 - div_rmse_40

        improvement_div_vs_qbc_40 = 100.0 * (1.0 - div_rmse_40 / qbc_rmse_40)

        # ====================================================
        # 100% mean RMSE values
        # ====================================================
        random_rmse_100, random_mode_100 = get_mean_rmse_at_ratio(
            agg_by_run[run_random],
            metric_col,
            target_ratio=TARGET_RATIO_100,
            strict_exact=False,
        )

        qbc_rmse_100, qbc_mode_100 = get_mean_rmse_at_ratio(
            agg_by_run[run_qbc],
            metric_col,
            target_ratio=TARGET_RATIO_100,
            strict_exact=False,
        )

        div_rmse_100, div_mode_100 = get_mean_rmse_at_ratio(
            agg_by_run[run_div],
            metric_col,
            target_ratio=TARGET_RATIO_100,
            strict_exact=False,
        )

        # Full-training reference:
        # mean of the average RMSE at 100% over Random, QBC, and QBC+Diversity.
        full_training_policy_mean_rmse_100 = np.mean([
            random_rmse_100,
            qbc_rmse_100,
            div_rmse_100,
        ])

        # ====================================================
        # Requested improvements relative to full-training reference
        # ====================================================
        improvement_qbc_40_vs_full_training_policy_mean_100 = (
            100.0 * (1.0 - qbc_rmse_40 / full_training_policy_mean_rmse_100)
        )

        improvement_qbc_diversity_40_vs_full_training_policy_mean_100 = (
            100.0 * (1.0 - div_rmse_40 / full_training_policy_mean_rmse_100)
        )

        diff_improvement_qbcdiv_minus_qbc_40_vs_full_avg_100 = (
            improvement_qbc_diversity_40_vs_full_training_policy_mean_100
            - improvement_qbc_40_vs_full_training_policy_mean_100
        )

        # Also useful: direct differences from the full-training reference.
        qbc_40_minus_full_training_policy_mean_100 = (
            qbc_rmse_40 - full_training_policy_mean_rmse_100
        )

        qbcdiv_40_minus_full_training_policy_mean_100 = (
            div_rmse_40 - full_training_policy_mean_rmse_100
        )

        diff_qbc40_minus_qbcdiv40_relative_to_same_full_avg = (
            qbc_40_minus_full_training_policy_mean_100
            - qbcdiv_40_minus_full_training_policy_mean_100
        )

        rows.append({
            "dataset": dataset,
            "target": target,
            "target_label": target_label(target),
            "metric": metric_name,
            "metric_col": metric_col,

            # ====================================================
            # 40% values
            # ====================================================
            "train_ratio_40": TARGET_RATIO_40,
            "training_percent_40": TARGET_RATIO_40_PERCENT,

            "random_mean_rmse_40": random_rmse_40,
            "qbc_mean_rmse_40": qbc_rmse_40,
            "qbc_diversity_mean_rmse_40": div_rmse_40,

            "random_extraction_40": random_mode_40,
            "qbc_extraction_40": qbc_mode_40,
            "qbc_diversity_extraction_40": div_mode_40,

            "diff_random_minus_qbc_40": diff_random_minus_qbc_40,
            "diff_random_minus_qbc_40_abs": abs(diff_random_minus_qbc_40),

            "diff_random_minus_qbc_diversity_40": diff_random_minus_div_40,
            "diff_random_minus_qbc_diversity_40_abs": abs(diff_random_minus_div_40),

            "diff_qbc_minus_qbc_diversity_40": diff_qbc_minus_div_40,
            "diff_qbc_minus_qbc_diversity_40_abs": abs(diff_qbc_minus_div_40),

            "improvement_qbc_diversity_vs_qbc_percent_40": improvement_div_vs_qbc_40,

            # ====================================================
            # 100% values
            # ====================================================
            "train_ratio_100": TARGET_RATIO_100,
            "training_percent_100": TARGET_RATIO_100_PERCENT,

            "random_mean_rmse_100": random_rmse_100,
            "qbc_mean_rmse_100": qbc_rmse_100,
            "qbc_diversity_mean_rmse_100": div_rmse_100,

            "random_extraction_100": random_mode_100,
            "qbc_extraction_100": qbc_mode_100,
            "qbc_diversity_extraction_100": div_mode_100,

            "full_training_policy_mean_rmse_100": full_training_policy_mean_rmse_100,

            # ====================================================
            # Requested improvement values
            # ====================================================
            "improvement_qbc_40_vs_full_training_policy_mean_100_percent": (
                improvement_qbc_40_vs_full_training_policy_mean_100
            ),

            "improvement_qbc_diversity_40_vs_full_training_policy_mean_100_percent": (
                improvement_qbc_diversity_40_vs_full_training_policy_mean_100
            ),

            "diff_improvement_qbcdiv_minus_qbc_40_vs_full_avg_100_percent": (
                diff_improvement_qbcdiv_minus_qbc_40_vs_full_avg_100
            ),

            # ====================================================
            # Direct RMSE differences from full-training reference
            # ====================================================
            "qbc_40_minus_full_training_policy_mean_rmse_100": (
                qbc_40_minus_full_training_policy_mean_100
            ),

            "qbc_diversity_40_minus_full_training_policy_mean_rmse_100": (
                qbcdiv_40_minus_full_training_policy_mean_100
            ),

            "diff_qbc40_minus_qbcdiv40_relative_to_same_full_avg_rmse": (
                diff_qbc40_minus_qbcdiv40_relative_to_same_full_avg
            ),
        })

    return pd.DataFrame(rows)


def print_ratio_comparisons(comp_df: pd.DataFrame) -> None:
    """
    Prints all calculated information clearly.
    """
    for _, r in comp_df.iterrows():
        print("\n" + "=" * 125)
        print(
            f"[RMSE SUMMARY] "
            f"dataset={str(r['dataset']).upper()} | "
            f"target={r['target_label']} | "
            f"metric={r['metric']}"
        )
        print("-" * 125)

        # ====================================================
        # 40% section
        # ====================================================
        print("[40% training ratio]")
        print(
            f"Random sampling mean RMSE @ 40%:      "
            f"{r['random_mean_rmse_40']:.8f} ({r['random_extraction_40']})"
        )
        print(
            f"QBC mean RMSE @ 40%:                  "
            f"{r['qbc_mean_rmse_40']:.8f} ({r['qbc_extraction_40']})"
        )
        print(
            f"QBC + Diversity mean RMSE @ 40%:      "
            f"{r['qbc_diversity_mean_rmse_40']:.8f} ({r['qbc_diversity_extraction_40']})"
        )

        print("\nPairwise signed differences at 40%:")
        print(
            f"Random - QBC:                         "
            f"{r['diff_random_minus_qbc_40']:.8f} "
            f"| abs={r['diff_random_minus_qbc_40_abs']:.8f}"
        )
        print(
            f"Random - QBC+Diversity:               "
            f"{r['diff_random_minus_qbc_diversity_40']:.8f} "
            f"| abs={r['diff_random_minus_qbc_diversity_40_abs']:.8f}"
        )
        print(
            f"QBC - QBC+Diversity:                  "
            f"{r['diff_qbc_minus_qbc_diversity_40']:.8f} "
            f"| abs={r['diff_qbc_minus_qbc_diversity_40_abs']:.8f}"
        )

        print("\nImprovement of QBC+Diversity relative to non-diversity QBC at 40%:")
        print(
            f"100 * (1 - RMSE_div_40 / RMSE_qbc_40) = "
            f"{r['improvement_qbc_diversity_vs_qbc_percent_40']:.4f}%"
        )

        # ====================================================
        # 100% full-training reference section
        # ====================================================
        print("\n[100% full-training reference]")
        print(
            f"Random sampling mean RMSE @ 100%:     "
            f"{r['random_mean_rmse_100']:.8f} ({r['random_extraction_100']})"
        )
        print(
            f"QBC mean RMSE @ 100%:                 "
            f"{r['qbc_mean_rmse_100']:.8f} ({r['qbc_extraction_100']})"
        )
        print(
            f"QBC + Diversity mean RMSE @ 100%:     "
            f"{r['qbc_diversity_mean_rmse_100']:.8f} ({r['qbc_diversity_extraction_100']})"
        )

        print(
            f"\nFull-training policy mean RMSE @ 100% "
            f"over Random, QBC, and QBC+Diversity: "
            f"{r['full_training_policy_mean_rmse_100']:.8f}"
        )

        print("\nRequested improvements at 40% relative to the 100% full-training policy mean:")
        print(
            f"QBC @ 40% vs full-training policy mean @ 100%: "
            f"{r['improvement_qbc_40_vs_full_training_policy_mean_100_percent']:.4f}%"
        )
        print(
            f"QBC+Diversity @ 40% vs full-training policy mean @ 100%: "
            f"{r['improvement_qbc_diversity_40_vs_full_training_policy_mean_100_percent']:.4f}%"
        )
        print(
            f"Difference: QBC+Diversity improvement - QBC improvement = "
            f"{r['diff_improvement_qbcdiv_minus_qbc_40_vs_full_avg_100_percent']:.4f}%"
        )

        print("\nDirect RMSE difference from full-training policy mean:")
        print(
            f"QBC_40 - full_training_policy_mean_100:          "
            f"{r['qbc_40_minus_full_training_policy_mean_rmse_100']:.8f}"
        )
        print(
            f"QBCDiv_40 - full_training_policy_mean_100:       "
            f"{r['qbc_diversity_40_minus_full_training_policy_mean_rmse_100']:.8f}"
        )
        print(
            f"(QBC_40 - full_avg_100) - (QBCDiv_40 - full_avg_100): "
            f"{r['diff_qbc40_minus_qbcdiv40_relative_to_same_full_avg_rmse']:.8f}"
        )


# ============================================================
# Plotting: 3 metrics in ONE single image per target
# ============================================================
def plot_all_metrics_for_dataset_target(
    dataset: str,
    target: str,
    agg_by_run: dict,
    dataset_out_dir: str,
) -> None:
    metric_cols = list(metrics_internal.keys())
    panel_labels = ["a)", "b)", "c)"]

    fig, axes = plt.subplots(1, 3, figsize=(21, 10), sharex=True)

    for i, (ax, metric_col) in enumerate(zip(axes, metric_cols)):
        for run_name, agg in agg_by_run.items():
            x_percent = agg["train_ratio"].to_numpy(dtype=float) * 100.0
            mean = agg[(metric_col, "mean")].to_numpy(dtype=float)
            std = agg[(metric_col, "std")].to_numpy(dtype=float)

            ax.plot(x_percent, mean, linewidth=2, label=run_name)
            ax.fill_between(x_percent, mean - std, mean + std, alpha=0.18)

        ax.axvline(
            TARGET_RATIO_40_PERCENT,
            color="black",
            linestyle="--",
            linewidth=1.2,
            alpha=0.6,
        )

        ax.axvline(
            TARGET_RATIO_100_PERCENT,
            color="black",
            linestyle=":",
            linewidth=1.2,
            alpha=0.6,
        )

        ax.set_xlabel("Training data (%)", fontsize=25)
        ax.set_ylabel(f"RMSE {metrics_internal[metric_col]}", fontsize=25)

        ax.set_title(
            f"{panel_labels[i]}",
            fontsize=25,
            fontweight="bold",
            loc="left"
        )

        xticks = np.arange(10, 101, 10)
        ax.set_xticks(xticks)
        ax.tick_params(axis="x", labelrotation=45, labelsize=20)
        ax.tick_params(axis="y", labelsize=20)

        ax.margins(x=0)
        ax.grid(alpha=0.3)

    fig.suptitle(
        f"{dataset.upper()} — Target: {target_label(target)}",
        fontsize=30,
        fontweight="bold",
        y=0.925,
    )

    handles, labels = axes[0].get_legend_handles_labels()

    fig.legend(
        handles,
        labels,
        loc="lower center",
        ncol=3,
        fontsize=25,
        frameon=False,
        bbox_to_anchor=(0.5, -0.02)
    )

    fig.tight_layout(rect=[0, 0.05, 1, 0.93])

    out_path = os.path.join(
        dataset_out_dir,
        f"{dataset}_target_{target}__all_rmse.png"
    )

    fig.savefig(out_path, dpi=100, bbox_inches="tight")
    plt.close(fig)


# ============================================================
# MAIN LOOP
# ============================================================
os.makedirs(OUT_DIR, exist_ok=True)
os.makedirs(TABLE_OUT_DIR, exist_ok=True)

all_mean_rmse_rows = []
all_comparison_rows = []

for dataset in DATASETS:
    dataset_out_dir = os.path.join(OUT_DIR, dataset)
    os.makedirs(dataset_out_dir, exist_ok=True)

    for target in TARGETS:
        runs = {
            "Random sampling": {
                "paths": [
                    (
                        seed,
                        os.path.join(
                            ROOT_RANDOM,
                            dataset,
                            "all",
                            str(seed),
                            random_csv_name(dataset, seed),
                        )
                    )
                    for seed in seeds
                ],
                "colmap": COLMAP_RANDOM_QBC,
            },

            f"QBC target: {target_label(target)}": {
                "paths": [
                    (
                        seed,
                        os.path.join(
                            ROOT_QBC,
                            dataset,
                            target,
                            str(seed),
                            qbc_csv_name(dataset, target),
                        )
                    )
                    for seed in seeds
                ],
                "colmap": COLMAP_RANDOM_QBC,
            },

            f"QBC+Diversity target: {target_label(target)}": {
                "paths": [
                    (
                        seed,
                        os.path.join(
                            ROOT_QBCDIV,
                            dataset,
                            target,
                            str(seed),
                            qbcdiv_csv_name(dataset, target, seed),
                        )
                    )
                    for seed in seeds
                ],
                "colmap": COLMAP_QBCDIV,
            },
        }

        agg_by_run = {}

        for run_name, info in runs.items():
            missing = [p for (_, p) in info["paths"] if not os.path.exists(p)]

            if missing:
                preview = "\n".join(missing[:8])
                raise FileNotFoundError(
                    f"[{dataset} | target={target}] "
                    f"Missing {len(missing)} file(s) for run '{run_name}'.\n"
                    f"First missing paths:\n{preview}"
                )

            data_long, agg = aggregate_over_seeds(info["paths"], info["colmap"])
            agg_by_run[run_name] = agg

        # Save mean RMSE at each train_ratio for this dataset-target.
        mean_df = flatten_mean_rmse_table(
            dataset=dataset,
            target=target,
            agg_by_run=agg_by_run,
        )

        mean_csv = os.path.join(
            TABLE_OUT_DIR,
            f"{dataset}_target_{target}_mean_rmse_by_train_ratio.csv",
        )

        mean_df.to_csv(mean_csv, index=False)
        all_mean_rmse_rows.append(mean_df)

        # Compute and print 40% and full-training-reference comparisons.
        comp_df = compute_ratio_comparisons(
            dataset=dataset,
            target=target,
            agg_by_run=agg_by_run,
        )

        print_ratio_comparisons(comp_df)
        all_comparison_rows.append(comp_df)

        # Plot original figure.
        plot_all_metrics_for_dataset_target(
            dataset=dataset,
            target=target,
            agg_by_run=agg_by_run,
            dataset_out_dir=dataset_out_dir,
        )

# ============================================================
# Save global summary tables
# ============================================================
all_mean_rmse_df = pd.concat(all_mean_rmse_rows, ignore_index=True)
all_comparison_df = pd.concat(all_comparison_rows, ignore_index=True)

all_mean_rmse_csv = os.path.join(
    TABLE_OUT_DIR,
    "ALL_mean_rmse_by_train_ratio.csv",
)

all_comparison_csv = os.path.join(
    TABLE_OUT_DIR,
    "ALL_40pct_improvement_vs_full_training_policy_mean_100.csv",
)

all_mean_rmse_df.to_csv(all_mean_rmse_csv, index=False)
all_comparison_df.to_csv(all_comparison_csv, index=False)

print("\n" + "#" * 125)
print("[DONE]")
print(f"Saved combined plots to:\n{OUT_DIR}")
print(f"Saved mean RMSE-by-train-ratio table to:\n{all_mean_rmse_csv}")
print(f"Saved 40% improvement vs full-training reference table to:\n{all_comparison_csv}")
print("#" * 125)
```
